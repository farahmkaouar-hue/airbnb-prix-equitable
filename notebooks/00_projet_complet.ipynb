<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-top:2px solid #3E4A55;border-bottom:1px solid #DDE2E7;padding:12px 2px;margin:18px 0;color:#3E4A55;font-size:16px;line-height:1.55">
<div style="font-size:11px;letter-spacing:.14em;text-transform:uppercase;color:#7B8794;margin-bottom:4px">Problématique</div>Comment aider un hôte ou une plateforme à estimer le <b>prix juste</b> d'un logement et à identifier les <b>facteurs de qualité</b>, de manière <b>généralisable à plusieurs villes</b> ? Pour répondre, il faut d'abord rendre les données lisibles, puis les observer, puis seulement les modéliser, et enfin vérifier que ce que l'on a appris tient sur une ville que le modèle n'a jamais vue.</div>

### À qui sert ce travail ?
| Acteur | Question | Où la trouver |
|---|---|---|
| **Hôtes** | À quel prix mettre mon logement, avec quelle fourchette ? Quel équipement vaut le coup ? | chapitres 11 à 13, 21 |
| **Plateforme** | Quelles annonces sont fantômes, mal évaluées ? Que reprochent les voyageurs ? | chapitres 8, 14, 15, 16, 22 |
| **Investisseurs et villes** | Quels quartiers sont sous-évalués ? Quelle saisonnalité ? Quelle concentration des hôtes ? | chapitres 8, 17, 22 |

### Trois règles tenues tout au long
1. **Aucune fuite de données.** Tout ce qui apprend sur les données (imputation, encodage, sélection, seuils de nettoyage, TF-IDF, LDA) est ajusté **sur le train uniquement**, dans des `Pipeline`, y compris dans la validation croisée. Le **jeu de test est mis de côté dès le chapitre 3** et ouvert une seule fois, au chapitre 19.
2. **Un hôte n'est jamais à la fois dans le train et dans le test** : il possède souvent plusieurs annonces presque identiques. Toutes les validations sont **groupées par hôte**.
3. **Aucun chiffre inventé.** Les chiffres de ce notebook viennent de code exécuté. Les calculs lourds (réglage, SHAP, NLP) sont faits par les modules de `src/` (dont le **code complet est reproduit à la fin du chapitre concerné**, en bloc repliable) et mis en cache ; le notebook les relit et les analyse. Tout échantillonnage est signalé.

Puis la vue globale : les 15 villes, leur taille, leur devise et la date de collecte.

````python
cols = ["city", "country", "currency", "scrape_date", "has_price", "price_eur", "price_local", "fx_rate_per_eur", "is_test", "in_reg_set", "host_is_superhost", "host_id"]
vue = pd.read_parquet(PROC / "listings_features.parquet", columns=cols)
t = vue.groupby("city").agg(pays=("country", "first"), devise=("currency", "first"), date_collecte=("scrape_date", lambda s: str(pd.to_datetime(s).iloc[0].date())),
                            annonces=("city", "size"), prix_manquant_pct=("has_price", lambda s: (1 - s.mean()) * 100),
                            prix_median_eur=("price_eur", "median"), superhosts_pct=("host_is_superhost", lambda s: s.mean() * 100)).sort_values("annonces", ascending=False)
display(t.style.format({"annonces": "{:,.0f}", "prix_manquant_pct": "{:.1f}", "prix_median_eur": "{:.0f}", "superhosts_pct": "{:.1f}"}))
fig, ax = plt.subplots(1, 2, figsize=(15, 5))
o = t.index
ax[0].barh(o[::-1], t.annonces[::-1], color=PAL["jaune"], edgecolor="white", linewidth=0.8); ax[0].set(title="Annonces par ville", xlabel="nombre d'annonces")
ax[1].barh(o[::-1], t.prix_manquant_pct[::-1], color=[PAL["rose"] if v > 30 else PAL["bleu"] for v in t.prix_manquant_pct[::-1]], edgecolor="white", linewidth=0.8)
ax[1].set(title="Part des annonces SANS prix (%)", xlabel="% sans prix")
plt.tight_layout(); plt.show()
print(f"Total : {fmt(len(vue))} annonces pour {fmt(vue.groupby(['city','host_id']).ngroups)} hôtes distincts. Annonces avec prix : {fmt(vue.has_price.sum())} ({vue.has_price.mean():.1%}).")
````

```yaml
# Point de configuration unique du projet (Phase 0)
seed: 42

paths:
  raw: data/raw
  processed: data/processed
  listings_raw: data/processed/listings_all.parquet
  reviews_raw: data/processed/reviews_all.parquet
  listings_clean: data/processed/listings_clean.parquet
  models: models
  figures: reports/figures
  reports: reports

# --- Taux de change vers EUR ---
# Source : taux de référence BCE (data-api.ecb.europa.eu, série EXR D.<devise>.EUR.SP00.A), récupérés le 2026-10-15.
# Fichier : data/external/ecb_fx_2026-09.csv (unités de devise pour 1 EUR). Pour chaque ville, on applique le taux
# du dernier jour ouvré <= date de collecte de la ville (price_eur = price / taux). EUR = 1.
fx:
  file: data/external/ecb_fx_2026-09.csv
  source: "BCE, taux de référence quotidiens (publiés vers 14h15 CET)"
  fetched_on: "2026-10-15"
  rule: "dernier fixing <= scrape_date de la ville"

# --- Stratégie de validation (voir reports/validation_strategy.md) ---
validation:
  final_test:
    n_splits: 7              # 1 pli sur 7 = ~14,3 % des annonces, groupé par (city, host_id), stratifié par ville
    group_by: [city, host_id]
    stratify_by: city        # fixé une fois (data/processed/split.parquet), utilisé une seule fois (phase 14)
  scheme_A: {type: stratified_kfold, k: 5, stratify_by: city}
  scheme_B: {type: leave_one_city_out, group: city}
  scheme_C: {type: temporal_by_quote_checkin, enabled: true, cutoff: "2026-10-01", note: "1 seule collecte par ville : variante temporelle sur la date de check-in du devis, hôtes disjoints ; vrai split temporel possible avec le calendar (phase 10)"}

targets:
  regression: log_price_eur
  classif_superhost: host_is_superhost
  classif_price_tier: {by: city, quantiles: 3, labels: [bas, moyen, haut]}

# Échantillonnage pour les étapes lourdes (SFS, SHAP, NLP...) ; à ajuster
sampling:
  heavy_steps_n: 50000
  selection_n: 60000

# --- Nettoyage (phase 2) ---
cleaning:
  amenity_min_freq: 0.01          # équipements conservés si présents dans >= 1 % des annonces (train)
  price_iqr_k: 3.0                # clôtures de Tukey élargies (3 x IQR) sur log(price_eur), par ville, calculées sur le train
  geo_max_km_from_center: 100     # au-delà : coordonnées jugées aberrantes
  max_nights_sentinel: 1125       # maximum_nights plafonné (valeurs sentinelles type 2147483647)
  clip_quantile: 0.999            # winsorisation de bedrooms / beds (quantile train)

# Colonnes qui ne doivent JAMAIS être des features de la régression de prix (fuite / dérivées de la cible)
leaky_columns: [estimated_revenue_l365d, estimated_occupancy_l365d, price, price_eur, log_price_eur]
```

<summary><code>download_airbnb.py</code> (295 lignes) : Téléchargement des fichiers Inside Airbnb (listings, reviews, calendar) par ville</summary>

```python
#!/usr/bin/env python3
"""Télécharge les données Inside Airbnb (listings / reviews / calendar) pour une liste de villes.

Exemples :
    python download_airbnb.py --cities Paris          # test sur une ville
    python download_airbnb.py                          # les 15 villes, collecte la plus récente
    python download_airbnb.py --with-calendar --all-dates
"""
from __future__ import annotations

import argparse
import csv
import gzip
import logging
import random
import re
import sys
import time
import zlib
from dataclasses import asdict, dataclass
from pathlib import Path

import pandas as pd
import requests
from bs4 import BeautifulSoup
from tqdm import tqdm

BASE_URL = "https://insideairbnb.com/get-the-data/"
DEFAULT_UA = "InsideAirbnbResearchDownloader/1.0 (personal ML project; polite: 1-2 s delay)"

# Nom affiché -> (slug pays, slug ville) tels qu'ils apparaissent dans les URL du site.
CITIES: dict[str, tuple[str, str]] = {
    "Paris": ("france", "paris"),
    "Lyon": ("france", "lyon"),
    "Barcelona": ("spain", "barcelona"),
    "Madrid": ("spain", "madrid"),
    "London": ("united-kingdom", "london"),
    "Amsterdam": ("the-netherlands", "amsterdam"),
    "Berlin": ("germany", "berlin"),
    "Rome": ("italy", "rome"),
    "Lisbon": ("portugal", "lisbon"),
    "New York City": ("united-states", "new-york-city"),
    "Los Angeles": ("united-states", "los-angeles"),
    "Toronto": ("canada", "toronto"),
    "Sydney": ("australia", "sydney"),
    "Tokyo": ("japan", "tokyo"),
    "Mexico City": ("mexico", "mexico-city"),
}

URL_RE = re.compile(
    r"https://data\.insideairbnb\.com/(?P<country>[^/\"'<>\s]+)/(?P<region>[^/\"'<>\s]+)/"
    r"(?P<city>[^/\"'<>\s]+)/(?P<date>\d{4}-\d{2}-\d{2})/data/"
    r"(?P<kind>listings|reviews|calendar)\.csv\.gz"
)

log = logging.getLogger("airbnb")


@dataclass(frozen=True)
class Entry:
    city: str  # slug de la ville
    country: str  # slug du pays
    date: str
    kind: str  # listings | reviews | calendar
    url: str


# --------------------------------------------------------------------------- réseau

class Client:
    """Session HTTP polie : User-Agent explicite, pause 1-2 s, retries à backoff exponentiel."""

    def __init__(self, user_agent: str, retries: int = 5, pause: tuple[float, float] = (1.0, 2.0)):
        self.session = requests.Session()
        self.session.headers["User-Agent"] = user_agent
        self.retries = retries
        self.pause = pause

    def get(self, url: str, **kwargs) -> requests.Response:
        for attempt in range(1, self.retries + 1):
            time.sleep(random.uniform(*self.pause))
            try:
                resp = self.session.get(url, timeout=(15, 120), **kwargs)
                if resp.status_code in (404, 403):  # définitif : inutile de réessayer
                    resp.raise_for_status()
                if resp.status_code == 429 or resp.status_code >= 500:
                    raise requests.HTTPError(f"HTTP {resp.status_code}", response=resp)
                resp.raise_for_status()
                return resp
            except requests.RequestException as exc:
                status = getattr(exc.response, "status_code", None)
                if status in (403, 404) or attempt == self.retries:
                    raise
                wait = 2 ** attempt + random.random()
                log.warning("Echec %s (%s), nouvel essai %d/%d dans %.0fs", url, exc, attempt, self.retries, wait)
                time.sleep(wait)
        raise RuntimeError("unreachable")


# --------------------------------------------------------------------------- index

def entries_from_text(text: str) -> list[Entry]:
    seen: dict[str, Entry] = {}
    for m in URL_RE.finditer(text):
        seen[m.group(0)] = Entry(m["city"], m["country"], m["date"], m["kind"], m.group(0))
    return list(seen.values())


def entries_from_bs4(html: str) -> list[Entry]:
    hrefs = " ".join(a["href"] for a in BeautifulSoup(html, "html.parser").find_all("a", href=True))
    return entries_from_text(hrefs)


def html_from_playwright(url: str, user_agent: str) -> str:
    from playwright.sync_api import sync_playwright  # import tardif : dépendance optionnelle

    with sync_playwright() as p:
        browser = p.chromium.launch()
        page = browser.new_page(user_agent=user_agent)
        page.goto(url, wait_until="networkidle")
        html = page.content()
        browser.close()
    return html


def build_index(client: Client, user_agent: str) -> list[Entry]:
    """requests + regex, puis BeautifulSoup, puis Playwright en dernier recours."""
    log.info("Téléchargement de la page %s", BASE_URL)
    html = client.get(BASE_URL).content.decode("utf-8", errors="replace")
    for name, parser in (("regex", entries_from_text), ("BeautifulSoup", entries_from_bs4)):
        entries = parser(html)
        if entries:
            log.info("Index construit via %s : %d fichiers", name, len(entries))
            return entries
    log.warning("Aucun lien trouvé en HTML statique, tentative Playwright")
    entries = entries_from_bs4(html_from_playwright(BASE_URL, user_agent))
    log.info("Index construit via Playwright : %d fichiers", len(entries))
    return entries


def save_index(entries: list[Entry], path: Path) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("w", newline="", encoding="utf-8") as f:
        writer = csv.DictWriter(f, fieldnames=list(asdict(entries[0])) if entries else ["url"])
        writer.writeheader()
        writer.writerows(asdict(e) for e in entries)


# --------------------------------------------------------------------------- sélection

def select_entries(index: list[Entry], wanted: dict[str, tuple[str, str]],
                   kinds: set[str], all_dates: bool) -> tuple[list[Entry], list[str]]:
    """Retourne (entrées à télécharger, noms de villes introuvables)."""
    selected, missing = [], []
    for name, (country, slug) in wanted.items():
        city_entries = [e for e in index if e.city == slug and e.country == country and e.kind in kinds]
        if not any(e.kind == "listings" for e in city_entries):
            missing.append(name)
            continue
        dates = sorted({e.date for e in city_entries if e.kind == "listings"}, reverse=True)
        keep = set(dates) if all_dates else {dates[0]}
        selected += [e for e in city_entries if e.date in keep]
    return selected, missing


def target_path(root: Path, e: Entry) -> Path:
    return root / e.city / e.date / f"{e.kind}.csv.gz"


# --------------------------------------------------------------------------- intégrité

def is_valid(path: Path) -> bool:
    """gzip complet (CRC vérifié à la fin du flux) et CSV lisible par pandas."""
    try:
        with gzip.open(path, "rb") as f:
            while f.read(1 << 20):
                pass
        pd.read_csv(path, nrows=100)
        return True
    except (OSError, EOFError, zlib.error, UnicodeDecodeError, pd.errors.ParserError, pd.errors.EmptyDataError):
        return False


def download_file(client: Client, e: Entry, dest: Path) -> None:
    dest.parent.mkdir(parents=True, exist_ok=True)
    tmp = dest.with_suffix(dest.suffix + ".part")
    resp = client.get(e.url, stream=True)
    total = int(resp.headers.get("Content-Length", 0)) or None
    with tmp.open("wb") as f, tqdm(total=total, unit="B", unit_scale=True, leave=False,
                                   desc=f"{e.city}/{e.kind}") as bar:
        for chunk in resp.iter_content(1 << 20):
            f.write(chunk)
            bar.update(len(chunk))
    tmp.replace(dest)


def fetch_entry(client: Client, e: Entry, root: Path) -> str:
    """Retourne 'skipped', 'downloaded' ou 'failed'."""
    dest = target_path(root, e)
    if dest.exists() and is_valid(dest):
        log.info("Déjà présent et valide : %s", dest)
        return "skipped"
    if dest.exists():
        log.warning("Fichier corrompu, retéléchargement : %s", dest)
        dest.unlink()
    try:
        download_file(client, e, dest)
    except requests.RequestException as exc:
        log.error("Echec téléchargement %s : %s", e.url, exc)
        return "failed"
    if not is_valid(dest):
        log.error("Fichier invalide après téléchargement : %s", dest)
        dest.unlink(missing_ok=True)
        return "failed"
    log.info("Téléchargé : %s (%.1f Mo)", dest, dest.stat().st_size / 1e6)
    return "downloaded"


# --------------------------------------------------------------------------- CLI

def parse_args() -> argparse.Namespace:
    p = argparse.ArgumentParser(description=__doc__, formatter_class=argparse.RawDescriptionHelpFormatter)
    p.add_argument("--cities", nargs="+", metavar="VILLE",
                   help="Sous-ensemble de villes (défaut : les 15). Ex : --cities Paris Lyon")
    p.add_argument("--all-dates", action="store_true", help="Toutes les collectes, pas seulement la plus récente")
    p.add_argument("--with-calendar", action="store_true", help="Télécharge aussi calendar.csv.gz (très volumineux)")
    p.add_argument("--output-dir", type=Path, default=Path("data/raw"))
    p.add_argument("--log-file", type=Path, default=Path("logs/download.log"))
    p.add_argument("--user-agent", default=DEFAULT_UA)
    p.add_argument("--retries", type=int, default=5)
    p.add_argument("--dry-run", action="store_true", help="Construit l'index et affiche la sélection sans télécharger")
    return p.parse_args()


def setup_logging(path: Path) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s",
                        handlers=[logging.FileHandler(path, encoding="utf-8")])


def resolve_cities(names: list[str] | None) -> dict[str, tuple[str, str]]:
    if not names:
        return CITIES
    lookup = {k.lower(): k for k in CITIES}
    unknown = [n for n in names if n.lower() not in lookup]
    if unknown:
        sys.exit(f"Villes inconnues : {unknown}. Disponibles : {list(CITIES)}")
    return {lookup[n.lower()]: CITIES[lookup[n.lower()]] for n in names}


def print_summary(selected: list[Entry], missing: list[str], results: dict[str, int],
                  failed: list[Entry], root: Path) -> None:
    found = sorted({e.city for e in selected})
    size = sum(target_path(root, e).stat().st_size for e in selected if target_path(root, e).exists())
    print("\n===== Récapitulatif =====")
    print(f"Villes trouvées   ({len(found)}) : {', '.join(found) or '-'}")
    print(f"Villes manquantes ({len(missing)}) : {', '.join(missing) or '-'}")
    print(f"Fichiers : {results['downloaded']} téléchargés, {results['skipped']} déjà présents, "
          f"{results['failed']} en échec")
    for e in failed:
        print(f"  ECHEC : {e.url}")
    print(f"Taille totale sélection sur disque : {size / 1e6:.1f} Mo ({size / 1e9:.2f} Go)")
    print("Journal : logs/download.log")


def main() -> int:
    args = parse_args()
    setup_logging(args.log_file)
    client = Client(args.user_agent, retries=args.retries)
    index = build_index(client, args.user_agent)
    save_index(index, args.output_dir / "index.csv")

    kinds = {"listings", "reviews"} | ({"calendar"} if args.with_calendar else set())
    selected, missing = select_entries(index, resolve_cities(args.cities), kinds, args.all_dates)
    if missing:
        log.warning("Villes introuvables : %s", missing)
    if args.dry_run:
        for e in selected:
            print(e.url)
        print(f"Manquantes : {missing}")
        return 0

    results = {"downloaded": 0, "skipped": 0, "failed": 0}
    failed: list[Entry] = []
    for e in tqdm(selected, desc="Fichiers", unit="fichier"):
        status = fetch_entry(client, e, args.output_dir)
        results[status] += 1
        if status == "failed":
            failed.append(e)
    print_summary(selected, missing, results, failed, args.output_dir)
    return 1 if failed else 0


if __name__ == "__main__":
    sys.exit(main())

<summary><code>merge_airbnb.py</code> (225 lignes) : Fusion des 15 villes en listings_all.parquet et reviews_all.parquet</summary>

```python
#!/usr/bin/env python3
"""Fusionne les listings / reviews par ville en Parquet et génère un rapport de données.

Aucune ligne n'est supprimée : seules des colonnes sont ajoutées et le prix est converti en float.

Exemple :
    python merge_airbnb.py
    python merge_airbnb.py --cities paris --output-dir data/processed_test
"""
from __future__ import annotations

import argparse
import logging
import re
from dataclasses import dataclass, field
from pathlib import Path

import pandas as pd
import pyarrow as pa
import pyarrow.parquet as pq

from download_airbnb import CITIES

log = logging.getLogger("merge")

CURRENCY_SYMBOLS = {"$": "USD", "€": "EUR", "£": "GBP", "¥": "JPY", "R$": "BRL", "CHF": "CHF"}
SYMBOL_RE = re.compile(r"^\s*([^\d\s.,-]+)")
QUOTE_CURRENCY_RE = re.compile(r'"currency":\s*"([A-Z]{3})"')
REVIEW_CHUNK = 500_000


@dataclass
class CityStats:
    city: str
    country: str
    scrape_date: str
    n_listings: int = 0
    n_reviews: int = 0
    pct_missing: float = float("nan")
    currency: str = ""
    absent_columns: list[str] = field(default_factory=list)


# --------------------------------------------------------------------------- découverte

def find_runs(raw_dir: Path, only: set[str] | None) -> list[tuple[str, str, Path]]:
    """Liste (slug ville, date, dossier) pour chaque collecte contenant un listings.csv.gz."""
    runs = []
    for listing in sorted(raw_dir.glob("*/*/listings.csv.gz")):
        date_dir = listing.parent
        slug = date_dir.parent.name
        if only is None or slug in only:
            runs.append((slug, date_dir.name, date_dir))
    return runs


def country_of(slug: str) -> str:
    for country, city_slug in CITIES.values():
        if city_slug == slug:
            return country
    return ""


# --------------------------------------------------------------------------- nettoyage

def currency_from_df(df: pd.DataFrame) -> str:
    """Devise ISO lue dans price_quote_raw (fiable) ; sinon déduite du symbole du prix ('' si inconnue)."""
    if "price_quote_raw" in df:
        codes = df["price_quote_raw"].dropna().astype(str).str.extract(QUOTE_CURRENCY_RE)[0].dropna()
        if not codes.empty:
            return codes.mode().iloc[0]
    symbols = df["price"].dropna().astype(str).str.extract(SYMBOL_RE)[0].dropna()
    if symbols.empty:
        return ""
    top = symbols.mode().iloc[0]
    return CURRENCY_SYMBOLS.get(top, top)


def clean_price(raw: pd.Series) -> pd.Series:
    """'$1,250.00' -> 1250.0 (retire symboles et séparateurs de milliers)."""
    text = raw.astype("string").str.replace(r"[^\d.\-]", "", regex=True)
    return pd.to_numeric(text, errors="coerce").astype("float64")


def stringify_mixed_objects(df: pd.DataFrame) -> pd.DataFrame:
    """Colonnes de types mélangés entre villes (ex. int dans l'une, texte dans l'autre) -> texte."""
    for col in df.columns[df.dtypes == object]:
        if df[col].map(type).nunique() > 2:  # str + NaN attendus ; au-delà = types mélangés
            df[col] = df[col].map(lambda v: v if pd.isna(v) or isinstance(v, str) else str(v))
    return df


# --------------------------------------------------------------------------- listings

def load_listings(slug: str, date: str, folder: Path) -> tuple[pd.DataFrame, CityStats]:
    df = pd.read_csv(folder / "listings.csv.gz", low_memory=False)
    stats = CityStats(slug, country_of(slug), date, n_listings=len(df))
    stats.pct_missing = float(df.isna().to_numpy().mean() * 100)  # sur les colonnes d'origine
    if "price" in df:
        stats.currency = currency_from_df(df)
        df["currency"] = stats.currency
        df["price"] = clean_price(df["price"])
    df.insert(0, "city", slug)
    df.insert(1, "country", stats.country)
    df.insert(2, "scrape_date", date)
    return df, stats


def merge_listings(runs: list[tuple[str, str, Path]]) -> tuple[pd.DataFrame, list[CityStats]]:
    frames, stats = [], []
    for slug, date, folder in runs:
        log.info("listings : %s %s", slug, date)
        df, st = load_listings(slug, date, folder)
        frames.append(df)
        stats.append(st)
    all_columns = list(dict.fromkeys(c for f in frames for c in f.columns))
    for f, st in zip(frames, stats):
        st.absent_columns = [c for c in all_columns if c not in f.columns]
        if st.absent_columns:
            log.warning("%s : %d colonnes absentes : %s", st.city, len(st.absent_columns), st.absent_columns)
    merged = pd.concat(frames, ignore_index=True)[all_columns]
    return stringify_mixed_objects(merged), stats


# --------------------------------------------------------------------------- reviews

def reviews_chunks(slug: str, date: str, folder: Path):
    for chunk in pd.read_csv(folder / "reviews.csv.gz", chunksize=REVIEW_CHUNK, low_memory=False):
        if "listing_id" not in chunk:
            raise ValueError(f"{slug}: colonne listing_id absente de reviews.csv.gz")
        chunk.insert(0, "city", slug)
        chunk.insert(1, "scrape_date", date)
        yield chunk


def write_reviews(runs: list[tuple[str, str, Path]], out: Path, stats: dict[tuple[str, str], CityStats]) -> None:
    """Ecrit les avis par blocs (fichiers de plusieurs millions de lignes) dans un seul Parquet."""
    writer, schema = None, None
    try:
        for slug, date, folder in runs:
            if not (folder / "reviews.csv.gz").exists():
                log.warning("reviews manquant pour %s %s", slug, date)
                continue
            log.info("reviews : %s %s", slug, date)
            for chunk in reviews_chunks(slug, date, folder):
                table = pa.Table.from_pandas(chunk, preserve_index=False)
                if writer is None:
                    schema = table.schema
                    writer = pq.ParquetWriter(out, schema, compression="zstd")
                writer.write_table(table.cast(schema))
                stats[(slug, date)].n_reviews += len(chunk)
    finally:
        if writer is not None:
            writer.close()


# --------------------------------------------------------------------------- rapport

def write_report(path: Path, stats: list[CityStats], n_listings: int, n_reviews: int) -> None:
    lines = [
        "# Rapport de données Inside Airbnb",
        "",
        f"Total : {n_listings:,} annonces, {n_reviews:,} avis, {len(stats)} collectes.",
        "",
        "| Ville | Pays | Annonces | Avis | % valeurs manquantes | Date de collecte | Devise |",
        "|---|---|---:|---:|---:|---|---|",
    ]
    for s in sorted(stats, key=lambda s: (s.city, s.scrape_date)):
        lines.append(f"| {s.city} | {s.country} | {s.n_listings:,} | {s.n_reviews:,} | "
                     f"{s.pct_missing:.1f} % | {s.scrape_date} | {s.currency or 'inconnue'} |")
    lines += [
        "",
        "## Notes",
        "",
        "- % manquant : cellules vides sur l'ensemble des colonnes d'origine du fichier de la ville "
        "(les colonnes absentes de la ville ne sont pas comptées).",
        "- Devise : code ISO lu dans `price_quote_raw` ; à défaut, déduite du symbole du prix brut "
        "(le `$` est alors ambigu : USD, CAD, AUD, MXN...). Les prix ne sont pas convertis : "
        "ne pas comparer des villes de devises différentes sans conversion.",
        "",
        "## Colonnes absentes de certaines villes",
        "",
    ]
    absent = [s for s in stats if s.absent_columns]
    if absent:
        lines += [f"- **{s.city}** ({s.scrape_date}) : {', '.join(s.absent_columns)}" for s in absent]
    else:
        lines.append("Aucune : toutes les villes ont les mêmes colonnes.")
    path.write_text("\n".join(lines) + "\n", encoding="utf-8")


# --------------------------------------------------------------------------- CLI

def parse_args() -> argparse.Namespace:
    p = argparse.ArgumentParser(description=__doc__, formatter_class=argparse.RawDescriptionHelpFormatter)
    p.add_argument("--raw-dir", type=Path, default=Path("data/raw"))
    p.add_argument("--output-dir", type=Path, default=Path("data/processed"))
    p.add_argument("--cities", nargs="+", metavar="SLUG", help="Slugs de villes (ex : paris new-york-city)")
    p.add_argument("--skip-reviews", action="store_true")
    return p.parse_args()


def main() -> None:
    logging.basicConfig(level=logging.INFO, format="%(asctime)s %(levelname)s %(message)s")
    args = parse_args()
    runs = find_runs(args.raw_dir, set(args.cities) if args.cities else None)
    if not runs:
        raise SystemExit(f"Aucun listings.csv.gz trouvé sous {args.raw_dir}")
    args.output_dir.mkdir(parents=True, exist_ok=True)

    listings, stats = merge_listings(runs)
    listings.to_parquet(args.output_dir / "listings_all.parquet", index=False)
    log.info("listings_all.parquet : %d lignes, %d colonnes", *listings.shape)
    stats_by_key = {(s.city, s.scrape_date): s for s in stats}
    del listings

    if not args.skip_reviews:
        write_reviews(runs, args.output_dir / "reviews_all.parquet", stats_by_key)
    write_report(args.output_dir / "data_report.md", stats,
                 sum(s.n_listings for s in stats), sum(s.n_reviews for s in stats))
    log.info("Rapport : %s", args.output_dir / "data_report.md")


if __name__ == "__main__":
    main()
```

```python
"""Chargement des données et audit (Phase 0)."""
from pathlib import Path

import pandas as pd
import pyarrow.parquet as pq
import yaml

ROOT = Path(__file__).resolve().parents[1]


def load_config(path: str | Path = ROOT / "config.yaml") -> dict:
    with open(path, encoding="utf-8") as f:
        return yaml.safe_load(f)


def load_listings(cfg: dict | None = None, city: str | None = None, columns=None) -> pd.DataFrame:
    """Charge les annonces brutes (optionnellement une seule ville, ex. 'paris')."""
    cfg = cfg or load_config()
    filters = [("city", "==", city)] if city else None
    return pd.read_parquet(ROOT / cfg["paths"]["listings_raw"], columns=columns, filters=filters)


def reviews_metadata(cfg: dict | None = None) -> pq.FileMetaData:
    """Métadonnées des avis sans charger les 2,6 Go en mémoire."""
    cfg = cfg or load_config()
    return pq.ParquetFile(ROOT / cfg["paths"]["reviews_raw"]).metadata


def make_split(df: pd.DataFrame, cfg: dict | None = None) -> pd.DataFrame:
    """Test final : 1 pli sur n_splits, groupé par (city, host_id) et stratifié par ville. Fixé une fois."""
    from sklearn.model_selection import StratifiedGroupKFold

    cfg = cfg or load_config()
    k = cfg["validation"]["final_test"]["n_splits"]
    groups = df["city"].astype(str) + "_" + df["host_id"].astype(str)
    sgkf = StratifiedGroupKFold(n_splits=k, shuffle=True, random_state=cfg["seed"])
    _, test_idx = next(sgkf.split(df, df["city"], groups))
    out = df[["city", "id", "host_id"]].copy()
    out["is_test"] = False
    out.iloc[test_idx, out.columns.get_loc("is_test")] = True
    return out

    out.append("\n### host_is_superhost\n" + sh.to_frame().to_markdown())

# avis
m = pq.ParquetFile(ROOT / cfg["paths"]["reviews_raw"]).metadata
out.append(f"\n## Avis\n- {m.num_rows:,} lignes x {m.num_columns} colonnes (fichier {ROOT.joinpath(cfg['paths']['reviews_raw']).stat().st_size / 1e9:.2f} Go)")

(ROOT / "reports" / "phase0_audit.md").write_text("\n".join(out), encoding="utf-8")

## 2. Diagnostic de qualité des données

Avant de soigner, on mesure. On définit un **score de santé du dataset sur 100**, calculé à partir de quatre critères, que l'on recalculera après nettoyage (chapitre 7).

| Critère | Ce qu'il mesure |
|---|---|
| **Complétude** | part de cellules renseignées, sur les colonnes qui ne sont pas vides pour toutes les villes |
| **Validité** | prix strictement positif, coordonnées dans la ville, capacité entre 1 et 16, durée de séjour min ≤ max |
| **Cohérence** | capacité cohérente avec les lits et les chambres |
| **Unicité** | identifiants uniques et absence de quasi-doublons |

````python
def score_sante(d, empty_cols=()):
    """Score de santé sur 100 : moyenne de quatre critères (chacun entre 0 et 1)."""
    use = [c for c in d.columns if c not in empty_cols]
    complet = 1 - d[use].isna().mean().mean()
    ok = pd.Series(True, index=d.index)
    price = d["price"] if "price" in d else d["price_local"]
    valide = ((price.fillna(1) > 0) & d.accommodates.between(1, 16) & (d.minimum_nights <= d.maximum_nights)).mean()
    beds = d.beds.fillna(0)
    incoh = ((d.accommodates >= 4) & (d.bedrooms.fillna(0) == 0) & (beds == 0)) | (beds > 3 * d.accommodates) | (d.bedrooms > 2 * d.accommodates)
    coherent = 1 - incoh.mean()
    key = ["host_id", "latitude", "longitude", "room_type", "accommodates", "bedrooms"]
    unique = 1 - (d.id.duplicated().mean() + d.duplicated(key).mean())
    parts = {"Complétude": complet, "Validité": valide, "Cohérence": coherent, "Unicité": unique}
    return round(100 * np.mean(list(parts.values())), 1), parts

vides = [c for c in brut.columns if brut[c].isna().all()]
score_avant, parts_avant = score_sante(brut)
print(f"Paris brut : score de santé = {score_avant}/100 | colonnes 100 % vides : {len(vides)}")
{k: round(v * 100, 1) for k, v in parts_avant.items()}
````

## 3. Nettoyage : types, devises, textes, équipements

Chaque transformation est **journalisée** : une petite classe enregistre l'opération, le nombre de lignes touchées et la raison. À la fin, ce journal est la **traçabilité complète** du nettoyage. On la démontre ici sur Paris ; le même code (`src/cleaning.py`) tourne sur les 15 villes.

````python
class JournalDeBord:
    """Enregistre chaque opération de nettoyage : quoi, combien de lignes, pourquoi."""
    def __init__(self): self.etapes = []
    def log(self, nom, n, pourquoi): self.etapes.append({"Opération": nom, "Lignes touchées": int(n), "Pourquoi": pourquoi})
    def tableau(self): return pd.DataFrame(self.etapes).style.format({"Lignes touchées": "{:,.0f}"}).hide(axis="index")

journal = JournalDeBord()
from src.cleaning import norm_amenity, parse_bathrooms
df = brut.copy()
````

### 3.1 La devise et le prix en euros

````python
fx = pd.read_csv(ROOT / "data/external/ecb_fx_2026-09.csv")
display(fx.pivot(index="date", columns="currency", values="rate_per_eur").loc["2026-09-14":"2026-09-25"])
print("Taux de référence de la BCE (unités de devise pour 1 EUR), récupérés par l'API officielle ; pour chaque ville on prend le dernier fixing AVANT sa date de collecte.")
rate = 1.0 if df.currency.iloc[0] == "EUR" else float(fx[(fx.currency == df.currency.iloc[0]) & (fx.date <= str(df.scrape_date.iloc[0]))].iloc[-1].rate_per_eur)
df["price_eur"] = df["price"] / rate
journal.log("prix -> EUR", df.price.notna().sum(), f"price_eur = prix / taux BCE ({df.currency.iloc[0]}, taux {rate}) ; les {int(df.price.isna().sum()):,} annonces sans prix restent NaN (aucun prix récupérable)")
ex = vue.groupby("city").agg(devise=("currency", "first"), taux_BCE=("fx_rate_per_eur", "first"), **{"médiane (devise locale)": ("price_local", "median")})
ex["médiane (EUR)"] = (ex["médiane (devise locale)"] / ex["taux_BCE"]).round(0)
ex.loc[["london", "mexico-city", "tokyo", "paris", "new-york-city"]]
````

````python
for c in ["host_is_superhost", "host_has_profile_pic", "host_identity_verified", "has_availability"]:
    n = df[c].isin(["t", "f"]).sum(); df[c] = df[c].map({"t": True, "f": False}).astype("boolean")
    journal.log(f"booléen {c}", n, "'t'/'f' -> booléen")
for c in ["first_review", "last_review"]:
    df[c] = pd.to_datetime(df[c], errors="coerce")
journal.log("dates first_review, last_review", df.first_review.notna().sum(), "texte -> datetime")
bath = parse_bathrooms(df["bathrooms_text"]); df = pd.concat([df.drop(columns=["bathrooms_text"]), bath], axis=1)
journal.log("bathrooms_text -> bath_count, bath_shared", bath.bath_count.notna().sum(), f"nombre de salles de bain (demi = 0,5) et indicateur partagée ({int(bath.bath_shared.sum()):,} partagées)")
avant_ap = pd.DataFrame({"texte brut": brut.bathrooms_text.value_counts().head(8).index, "annonces": brut.bathrooms_text.value_counts().head(8).values})
avant_ap["bath_count"] = parse_bathrooms(avant_ap["texte brut"]).bath_count.values; avant_ap["bath_shared"] = parse_bathrooms(avant_ap["texte brut"]).bath_shared.values
avant_ap
````

### 3.3 Les équipements : de 32 755 libellés à 164 indicateurs

### 3.4 Les valeurs extrêmes de prix : une règle explicite, jamais en silence

````python
lp = np.log(df.price_eur.dropna())
q1, q3 = lp.quantile([.25, .75]); k = 3.0; lo, hi = q1 - k * (q3 - q1), q3 + k * (q3 - q1)
n_out = int(((lp < lo) | (lp > hi)).sum())
fig, ax = plt.subplots(figsize=(12, 4))
ax.hist(lp, bins=100, color=PAL["jaune"], edgecolor="white", linewidth=0.8)
ax.axvline(lo, color=PAL["rose"], lw=2); ax.axvline(hi, color=PAL["rose"], lw=2)
ax.set(title=f"Paris : log(prix en EUR) et clôtures de Tukey élargies (Q1 − 3·IQR ; Q3 + 3·IQR) : {n_out} annonces hors clôtures", xlabel="log(prix EUR / nuit)", ylabel="annonces")
plt.tight_layout(); plt.show()
print(f"Bornes en euros : {np.exp(lo):,.0f} € – {np.exp(hi):,.0f} € ; {n_out} annonces signalées ({n_out/len(lp):.2%}).")
journal.log("outliers de prix (Paris)", n_out, f"hors [{np.exp(lo):,.0f} € ; {np.exp(hi):,.0f} €] (3 x IQR de log(prix)) : signalés (is_price_outlier), jamais supprimés")
````

#### Code source du chapitre 3

<details>
<summary><code>src/cleaning.py</code> (258 lignes) : Nettoyage complet des 15 villes : conversion en EUR, équipements, outliers, drapeaux, journal</summary>

```python
"""Phase 2 : nettoyage. Produit data/processed/listings_clean.parquet, texts.parquet, split.parquet,
models/amenity_vocab.json, models/price_bounds.csv et reports/cleaning_log.md.

Principe anti-fuite : tout seuil appris sur les données (vocabulaire d'équipements, clôtures de prix,
centre-ville, quantiles de winsorisation) est calculé sur le TRAIN uniquement (split.parquet).
Rien n'est imputé ici (l'imputation se fait dans les Pipelines, ajustée sur le train de chaque pli).
Usage : python -m src.cleaning
"""
import json
import re

import numpy as np
import pandas as pd

from src.data import ROOT, load_config, load_listings, make_split
from src.features import haversine

LOG: list[tuple[str, str, str]] = []  # (étape, lignes/valeurs affectées, pourquoi)


def log(step: str, n, why: str):
    LOG.append((step, n if isinstance(n, str) else f"{int(n):,}", why))
    print(f"[{step}] {LOG[-1][1]} : {why}")


# ---------------------------------------------------------------- utilitaires
def load_fx(cfg: dict) -> pd.DataFrame:
    fx = pd.read_csv(ROOT / cfg["fx"]["file"], parse_dates=["date"])
    return fx.sort_values("date")


def fx_rate_per_city(df: pd.DataFrame, fx: pd.DataFrame) -> pd.Series:
    """Taux (devise pour 1 EUR) par annonce : dernier fixing BCE <= date de collecte de la ville."""
    rates = {}
    for (city, cur), d in df.groupby(["city", "currency"])["scrape_date"].first().items():
        if cur == "EUR":
            rates[city] = 1.0
            continue
        f = fx[(fx.currency == cur) & (fx.date <= pd.Timestamp(d))]
        if f.empty:
            raise ValueError(f"Aucun taux BCE pour {cur} avant {d} ({city}) : à compléter dans data/external/")
        rates[city] = float(f.iloc[-1].rate_per_eur)
    return df["city"].map(rates)


def norm_amenity(a: str) -> str:
    """Normalise un libellé : minuscule, retire les détails après ':' ou ' – ' (marques, débits...)."""
    a = re.split(r"\s[–-]\s|:", str(a).lower().strip())[0].strip()
    return re.sub(r"[^a-z0-9]+", "_", a).strip("_")


def parse_bathrooms(text: pd.Series) -> pd.DataFrame:
    t = text.str.lower()
    num = t.str.extract(r"(\d+(?:\.\d+)?)")[0].astype(float)
    num = num.where(~(t.str.contains("half", na=False) & num.isna()), 0.5)
    num = num.where(~t.str.contains("no bathroom", na=False), 0.0)
    shared = t.str.contains("shared", na=False).astype("float").where(t.notna())
    return pd.DataFrame({"bath_count": num, "bath_shared": shared})


# ---------------------------------------------------------------- nettoyage
def clean_listings(df: pd.DataFrame, split: pd.DataFrame, cfg: dict):
    cc = cfg["cleaning"]
    n0 = len(df)
    df = df.merge(split[["city", "id", "is_test"]], on=["city", "id"], how="left")
    assert df["is_test"].notna().all()
    train = ~df["is_test"].astype(bool)
    log("split", int(df.is_test.sum()), f"annonces du test final ({df.is_test.mean() * 100:.1f} %), groupées par hôte ; tous les seuils ci-dessous sont appris sur les {int(train.sum()):,} annonces train")

    # 1. colonnes entièrement vides
    empty = [c for c in df.columns if df[c].isna().all()]
    df = df.drop(columns=empty)
    log("colonnes 100 % vides", len(empty), "supprimées : " + ", ".join(empty))

    # 1b. contexte du devis de prix : le prix est un devis pour un séjour précis (dates, durée) -> variables saisonnières
    if {"price_quote_checkin_date", "price_quote_checkout_date"} <= set(df.columns):
        ci = pd.to_datetime(df["price_quote_checkin_date"], errors="coerce")
        co = pd.to_datetime(df["price_quote_checkout_date"], errors="coerce")
        df["quote_lead_days"] = (ci - pd.to_datetime(df["scrape_date"])).dt.days
        df["quote_nights"] = (co - ci).dt.days
        df["quote_checkin_month"] = ci.dt.month
        df["quote_checkin_dow"] = ci.dt.dayofweek
        log("contexte du devis", int(ci.notna().sum()), "price_quote_checkin/checkout_date -> quote_lead_days, quote_nights, quote_checkin_month, quote_checkin_dow "
            "(le prix est un devis de 1 à 31 nuits, check-in de 0 à 369 jours après la collecte : saisonnalité et remises mensuelles)")

    # 2. textes conservés à part pour le NLP, puis suppression des textes bruts / données personnelles
    texts = df[["city", "id", "name", "description", "host_about"]].copy()
    drop_cols = ["listing_url", "picture_url", "host_url", "host_profile_url", "host_picture_url", "scrape_id", "last_scraped", "host_profile_id",
                 "calendar_last_scraped", "host_name", "name", "description", "host_about"]
    drop_quote = [c for c in df.columns if c.startswith("price_quote_")]
    absent = [c for c in drop_cols if c not in df.columns]
    df = df.drop(columns=[c for c in drop_cols + drop_quote if c in df.columns])
    log("colonnes inutiles / sensibles", len(drop_cols) - len(absent) + len(drop_quote),
        "URLs, identifiants techniques, price_quote_* (doublon exact du prix, fuite), host_name (RGPD) supprimés ; name/description/host_about copiés dans texts.parquet"
        + (f" (absentes, ignorées : {absent})" if absent else ""))

    # 3. booléens et dates
    for c in ["host_is_superhost", "host_has_profile_pic", "host_identity_verified", "has_availability"]:
        if c in df:
            n_nan = int(df[c].isna().sum())
            df[c] = df[c].map({"t": True, "f": False}).astype("boolean")
            log(f"booléen {c}", len(df) - n_nan, f"'t'/'f' -> booléen ({n_nan} manquants conservés)")
    for c in ["first_review", "last_review", "scrape_date"]:
        df[c] = pd.to_datetime(df[c], errors="coerce")
    log("dates", len(df), "first_review, last_review, scrape_date -> datetime")
    df["is_stale"] = df["source"].notna() if "source" in df else False
    df = df.drop(columns=["source"], errors="ignore")
    log("source -> is_stale", int(df.is_stale.sum()), "'previous scrape' = annonce non rafraîchie à la dernière collecte (80 % sans prix)")
    df["has_license"] = df["license"].notna()
    df = df.drop(columns=["license"])
    log("license -> has_license", int(df.has_license.sum()), "texte libre (numéros d'enregistrement, donnée quasi personnelle) réduit à un indicateur")

    # 4. salles de bain
    b = parse_bathrooms(df["bathrooms_text"])
    df = pd.concat([df.drop(columns=["bathrooms_text"]), b], axis=1)
    log("bathrooms_text -> bath_count, bath_shared", int(b.bath_count.notna().sum()),
        f"parsé en nombre (demi-salle de bain = 0,5) et indicateur partagée ({int(b.bath_shared.sum()):,} partagées) ; {int(b.bath_count.isna().sum())} non parsables laissés NaN")

    # 5. équipements
    lists = [sorted({norm_amenity(x) for x in json.loads(s)} - {""}) if isinstance(s, str) else [] for s in df["amenities"]]
    lens = np.fromiter((len(x) for x in lists), int)
    flat = pd.Series([a for x in lists for a in x])
    row = np.repeat(np.arange(len(df)), lens)
    tr_rows = train.values[row]
    freq = flat[tr_rows].value_counts() / train.sum()
    vocab = freq[freq >= cc["amenity_min_freq"]].index.tolist()
    with open(ROOT / "models" / "amenity_vocab.json", "w", encoding="utf-8") as f:
        json.dump(vocab, f, ensure_ascii=False, indent=1)
    idx = {a: i for i, a in enumerate(vocab)}
    mat = np.zeros((len(df), len(vocab)), dtype="int8")
    col = flat.map(idx)
    ok = col.notna().values
    mat[row[ok], col[ok].astype(int).values] = 1
    am = pd.DataFrame(mat, columns=["am_" + a for a in vocab], index=df.index)
    df["n_amenities"] = lens
    df["n_amenities_other"] = lens - mat.sum(axis=1)
    df = pd.concat([df.drop(columns=["amenities"]), am], axis=1)
    log("équipements", len(vocab), f"libellés normalisés (minuscule, détails retirés) : {flat.nunique():,} libellés distincts -> {len(vocab)} conservés (>= {cc['amenity_min_freq']:.0%} des annonces train), "
        f"les autres regroupés dans n_amenities_other ; vocabulaire dans models/amenity_vocab.json")

    # 6. prix -> EUR
    fx = load_fx(cfg)
    rate = fx_rate_per_city(df, fx)
    df["fx_rate_per_eur"] = rate
    df["has_price"] = df["price"].notna() & (df["price"] > 0)
    df["price_eur"] = np.where(df.has_price, df["price"] / rate, np.nan)
    df["log_price_eur"] = np.log(df["price_eur"])
    df = df.rename(columns={"price": "price_local"})
    log("prix -> EUR", int(df.has_price.sum()), "price_eur = prix local / taux BCE du dernier fixing <= date de collecte de la ville ; devise d'origine conservée dans `currency` ; "
        f"{int((~df.has_price).sum()):,} annonces sans prix (aucun prix récupérable : le JSON price_quote_raw n'a pas de price_per_night pour elles) conservées avec price_eur = NaN")

    # 7. winsorisation de valeurs numériques extrêmes
    n_sent = int((df.maximum_nights > cc["max_nights_sentinel"]).sum())
    for c in ["maximum_nights", "maximum_maximum_nights", "minimum_maximum_nights", "maximum_nights_avg_ntm"]:
        if c in df:
            df[c] = df[c].clip(upper=cc["max_nights_sentinel"])
    log("maximum_nights", n_sent, f"valeurs sentinelles (max 2 147 483 647) plafonnées à {cc['max_nights_sentinel']}")
    for c in ["bedrooms", "beds"]:
        cap = df.loc[train, c].quantile(cc["clip_quantile"])
        n_cl = int((df[c] > cap).sum())
        df[c] = df[c].clip(upper=cap)
        log(f"winsorisation {c}", n_cl, f"plafonné au quantile {cc['clip_quantile']} du train = {cap:g}")

    # 8. annonces incohérentes (signalées, pas supprimées ici)
    beds0 = df.beds.fillna(0) == 0
    i1 = (df.accommodates >= 4) & (df.bedrooms.fillna(0) == 0) & beds0
    i2 = df.beds > 3 * df.accommodates
    i3 = df.bedrooms > 2 * df.accommodates
    i4 = df.minimum_nights > df.maximum_nights
    df["is_inconsistent"] = i1 | i2 | i3 | i4
    log("incohérence : capacité >= 4 sans chambre ni lit", int(i1.sum()), "capacité élevée sans couchage déclaré")
    log("incohérence : lits > 3 x capacité", int(i2.sum()), "nombre de lits incompatible avec la capacité")
    log("incohérence : chambres > 2 x capacité", int(i3.sum()), "nombre de chambres incompatible avec la capacité")
    log("incohérence : minimum_nights > maximum_nights", int(i4.sum()), "contradiction de règles de séjour")
    log("is_inconsistent (union)", int(df.is_inconsistent.sum()), "signalées ; exclues de l'entraînement de la régression, jamais supprimées de listings_clean")

    # 9. géographie : coordonnées aberrantes
    ctr = df[train].groupby("city")[["latitude", "longitude"]].median()
    c_ = df[["city"]].join(ctr, on="city")
    df["dist_center_km"] = haversine(df.latitude, df.longitude, c_.latitude, c_.longitude)
    df["is_geo_outlier"] = df.dist_center_km > cc["geo_max_km_from_center"]
    log("coordonnées aberrantes", int(df.is_geo_outlier.sum()), f"> {cc['geo_max_km_from_center']} km de la médiane des coordonnées de la ville (centre = médiane train)")

    # 10. doublons
    log("doublons exacts (city, id)", int(df.duplicated(["city", "id"]).sum()), "aucun")
    key = ["city", "host_id", "latitude", "longitude", "room_type", "accommodates", "bedrooms"]
    df["is_near_duplicate"] = df.duplicated(key, keep="first")
    log("quasi-doublons (même hôte, mêmes coordonnées, type, capacité, chambres)", int(df.is_near_duplicate.sum()),
        "signalés, non supprimés : souvent des unités distinctes d'un même immeuble (hôtels, résidences) ; le split groupé par hôte neutralise la fuite")

    # 11. outliers de prix : clôtures par ville sur log(price_eur), apprises sur le train
    k = cc["price_iqr_k"]
    tp = df[train & df.has_price]
    q = tp.groupby("city")["log_price_eur"].quantile([0.25, 0.75]).unstack()
    iqr = q[0.75] - q[0.25]
    bounds = pd.DataFrame({"low_log": q[0.25] - k * iqr, "high_log": q[0.75] + k * iqr})
    bounds["low_eur"] = np.exp(bounds.low_log)
    bounds["high_eur"] = np.exp(bounds.high_log)
    bounds.round(3).to_csv(ROOT / "models" / "price_bounds.csv")
    lo, hi = df.city.map(bounds.low_log), df.city.map(bounds.high_log)
    below = df.has_price & (df.log_price_eur < lo)
    above = df.has_price & (df.log_price_eur > hi)
    df["is_price_outlier"] = below | above
    log("outliers de prix", int(df.is_price_outlier.sum()),
        f"hors [Q1 - {k} x IQR ; Q3 + {k} x IQR] de log(price_eur) par ville (train) : {int(below.sum())} trop bas, {int(above.sum())} trop hauts "
        f"({df.is_price_outlier.sum() / df.has_price.sum() * 100:.2f} % des prix) ; signalés (is_price_outlier), exclus de l'entraînement seulement ; impact mesuré par outlier_impact")

    # 12. jeu de régression
    df["in_reg_set"] = df.has_price & ~df.is_price_outlier & ~df.is_geo_outlier & ~df.is_inconsistent
    log("jeu de régression (in_reg_set)", int(df.in_reg_set.sum()), f"= avec prix, sans outlier de prix, sans coordonnées aberrantes, sans incohérence ({df.in_reg_set.mean() * 100:.1f} % des annonces)")
    assert len(df) == n0
    return df, texts


def missing_strategy(df: pd.DataFrame) -> str:
    m = df.isna().mean() * 100
    m = m[m > 0].sort_values(ascending=False)
    rows = ["| Colonne | % manquant | Stratégie (appliquée dans le Pipeline, ajustée sur le train) |", "|---|---:|---|"]
    for c, v in m.items():
        if c in ("price_local", "price_eur", "log_price_eur"):
            s = "cible : jamais imputée ; annonces sans prix exclues de la régression"
        elif c.startswith("review_scores") or c in ("first_review", "last_review", "reviews_per_month"):
            s = "absence d'avis : indicateur `has_reviews` + imputation médiane par ville (linéaires) ou NaN natif (boosting)"
        elif str(df[c].dtype) in ("str", "string", "object", "category"):
            s = "catégorie « manquant »"
        elif str(df[c].dtype) == "boolean":
            s = "mode + indicateur de manquant"
        else:
            s = "médiane par ville (train) + indicateur de manquant"
        rows.append(f"| {c} | {v:.1f} | {s} |")
    return "\n".join(rows)


def write_log(df: pd.DataFrame, extra: str = ""):
    out = ["# Journal de nettoyage (Phase 2)", "",
           f"Base d'entrée : listings_all.parquet. Sortie : {len(df):,} lignes x {df.shape[1]} colonnes (listings_clean.parquet). Aucune ligne n'est supprimée : les annonces douteuses sont signalées par des indicateurs.", "",
           "| Étape | Lignes / valeurs affectées | Pourquoi |", "|---|---:|---|"]
    out += [f"| {a} | {b} | {c} |" for a, b, c in LOG]
    out += ["", "## Stratégie d'imputation par colonne (appliquée en phase 3/6 dans les Pipelines)", "", missing_strategy(df), "", extra]
    (ROOT / "reports" / "cleaning_log.md").write_text("\n".join(out) + "\n", encoding="utf-8")


def run():
    cfg = load_config()
    (ROOT / "models").mkdir(exist_ok=True)
    raw = load_listings(cfg)
    raw["scrape_date"] = pd.to_datetime(raw["scrape_date"])
    split = make_split(raw, cfg)
    split.to_parquet(ROOT / "data/processed/split.parquet", index=False)
    df, texts = clean_listings(raw, split, cfg)
    df.to_parquet(ROOT / cfg["paths"]["listings_clean"], index=False)
    texts.to_parquet(ROOT / "data/processed/texts.parquet", index=False)
    write_log(df)
    print(df.shape)


if __name__ == "__main__":
    run()
```

"""Phase 2 : impact de la règle d'outliers de prix (avec / sans / winsorisation).
Entraînement et validation uniquement sur le TRAIN (le test final n'est jamais utilisé ici).
Validation = 20 % des hôtes du train (groupés). Modèle de référence : LightGBM léger, mêmes hyperparamètres.
Usage : python -m src.outlier_impact   (ajoute une section à reports/cleaning_log.md)
"""
import lightgbm as lgb
import numpy as np
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit

from src.data import ROOT, load_config

cfg = load_config()
df = pd.read_parquet(ROOT / cfg["paths"]["listings_clean"])
df = df[~df.is_test & df.has_price & ~df.is_geo_outlier & ~df.is_inconsistent].copy()
bounds = pd.read_csv(ROOT / "models" / "price_bounds.csv", index_col=0)

feat = ["accommodates", "bedrooms", "beds", "bath_count", "bath_shared", "latitude", "longitude", "dist_center_km", "n_amenities", "room_type", "city"]
X = df[feat].copy()
for c in ["room_type", "city"]:
    X[c] = X[c].astype("category")
y = df["log_price_eur"].values
groups = df["city"].astype(str) + "_" + df["host_id"].astype(str)
tr, va = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=cfg["seed"]).split(X, y, groups))
out_ = df["is_price_outlier"].values
hi = df["city"].map(bounds.high_log).values
y_win = np.minimum(y, hi)

params = dict(n_estimators=400, learning_rate=0.05, num_leaves=63, random_state=cfg["seed"], verbose=-1)
rows = []
for name, mask_tr, ytr in [("Avec outliers (rien retiré)", np.ones(len(df), bool), y),
                           ("Sans outliers (règle 3 x IQR)", ~out_, y),
                           ("Winsorisation (plafonnée à la clôture haute)", np.ones(len(df), bool), y_win)]:
    idx = tr[mask_tr[tr]]
    m = lgb.LGBMRegressor(**params).fit(X.iloc[idx], ytr[idx])
    p = m.predict(X.iloc[va])
    inl = ~out_[va]
    rmse = lambda a, b: float(np.sqrt(np.mean((a - b) ** 2)))
    mae = lambda a, b: float(np.mean(np.abs(a - b)))
    rows.append([name, len(idx), rmse(y[va][inl], p[inl]), mae(y[va][inl], p[inl]), rmse(y[va], p), mae(y[va], p)])
res = pd.DataFrame(rows, columns=["Stratégie d'entraînement", "n train", "RMSE (validation sans outliers)", "MAE (sans outliers)", "RMSE (validation avec outliers)", "MAE (avec outliers)"]).round(4)

txt = ("\n## Impact de la règle d'outliers (LightGBM de référence, validation sur 20 % des hôtes du train ; cibles = log(price_eur))\n\n"
       + res.to_markdown(index=False)
       + f"\n\nValidation : {len(va):,} annonces dont {int(out_[va].sum())} outliers. Le test final n'est pas utilisé.\n")
p = ROOT / "reports" / "cleaning_log.md"
s = p.read_text(encoding="utf-8")
s = s.split("\n## Impact de la règle d'outliers")[0]
p.write_text(s + txt, encoding="utf-8")

### 4.2 Le prétraitement appris sur le train (Pipeline)

- **Numériques** : imputation par la **médiane de la ville** + indicateur de manquant ; standardisation seulement pour les modèles qui en ont besoin (linéaires, kNN, SVM, réseaux).
- **Catégories à faible cardinalité** (type de logement, type de propriété, ville) : *one-hot*, modalités rares (< 1 %) regroupées.
- **Quartier** (1 294 couples ville-quartier) : *target encoding* avec **validation croisée interne à 5 plis**, pour qu'une annonce ne contribue jamais à son propre encodage.

```python
"""Phase 3 : feature engineering réutilisable.

Deux familles de transformations :
  1. SANS apprentissage (formules déterministes par ligne) : `build_feature_table` -> data/processed/listings_features.parquet
  2. AVEC apprentissage (imputation par ville, encodage, scaling) : dans des Pipelines sklearn (`build_preprocessor`),
     à ajuster sur le train uniquement (y compris dans chaque pli de CV).
Usage : python -m src.features
"""
import json

import numpy as np
import pandas as pd
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler, TargetEncoder

from src.data import ROOT, load_config

SEED = 42


def haversine(lat1, lon1, lat2, lon2):
    """Distance en km entre deux points (degrés)."""
    p = np.pi / 180
    a = np.sin((lat2 - lat1) * p / 2) ** 2 + np.cos(lat1 * p) * np.cos(lat2 * p) * np.sin((lon2 - lon1) * p / 2) ** 2
    return 12742 * np.arcsin(np.sqrt(a))


# ---------------------------------------------------------------------------------------------------------------
# 1. Variables dérivées (sans apprentissage)
# ---------------------------------------------------------------------------------------------------------------
def add_text_features(df: pd.DataFrame, texts: pd.DataFrame) -> pd.DataFrame:
    """Longueurs de textes (le texte brut reste dans texts.parquet pour le NLP)."""
    t = texts.set_index(["city", "id"])
    desc = t["description"].fillna("")
    out = pd.DataFrame({
        "desc_len": desc.str.len(),
        "desc_words": desc.str.split().str.len(),
        "name_len": t["name"].fillna("").str.len(),
        "host_about_len": t["host_about"].fillna("").str.len(),
    }, index=t.index)
    out["has_host_about"] = (out["host_about_len"] > 0).astype(float)
    return df.join(out, on=["city", "id"])


def add_derived_features(df: pd.DataFrame) -> pd.DataFrame:
    """Variables dérivées déterministes (aucune statistique apprise)."""
    d = df.copy()
    # booléens nullables -> float (NaN conservés) pour tous les estimateurs
    for c in d.columns:
        if str(d[c].dtype) == "boolean":  # booléens à NaN possibles ; les drapeaux bool (is_test...) restent bool
            d[c] = d[c].astype("float")
    sd = pd.to_datetime(d["scrape_date"])
    # hôte
    d["host_tenure_years"] = d["hosts_time_as_host_years"] + d["hosts_time_as_host_months"] / 12
    d["user_tenure_years"] = d["hosts_time_as_user_years"] + d["hosts_time_as_user_months"] / 12
    d["host_is_pro"] = (d["calculated_host_listings_count"] >= 5).astype(float)
    tot = d["calculated_host_listings_count"].replace(0, np.nan)
    d["host_entire_share"] = d["calculated_host_listings_count_entire_homes"] / tot
    loc = d["host_location"].str.lower()
    cityname = d["city"].str.replace("-", " ", regex=False)
    d["host_is_local"] = [np.nan if pd.isna(l) else float(c.split()[0] in l) for l, c in zip(loc, cityname)]
    # avis
    d["has_reviews"] = (d["number_of_reviews"] > 0).astype(float)
    d["days_since_last_review"] = (sd - pd.to_datetime(d["last_review"])).dt.days
    d["days_since_first_review"] = (sd - pd.to_datetime(d["first_review"])).dt.days
    # logement
    d["is_studio"] = (d["bedrooms"] == 0).astype(float).where(d["bedrooms"].notna())
    acc = d["accommodates"].replace(0, np.nan)
    d["beds_per_person"] = d["beds"] / acc
    d["persons_per_bedroom"] = d["accommodates"] / d["bedrooms"].clip(lower=1)
    d["bath_per_person"] = d["bath_count"] / acc
    # transformations log des variables très asymétriques (utiles aux modèles linéaires, neutres pour les arbres)
    for src, dst in [("minimum_nights", "log_min_nights"), ("maximum_nights", "log_max_nights"), ("calculated_host_listings_count", "log_host_listings"),
                     ("number_of_reviews", "log_n_reviews"), ("number_of_reviews_ltm", "log_n_reviews_ltm"), ("reviews_per_month", "log_reviews_per_month"),
                     ("quote_lead_days", "log_quote_lead_days")]:
        d[dst] = np.log1p(d[src].clip(lower=0))
    # saisonnalité du devis
    m = d["quote_checkin_month"]
    d["quote_month_sin"] = np.sin(2 * np.pi * m / 12)
    d["quote_month_cos"] = np.cos(2 * np.pi * m / 12)
    d["quote_is_weekend"] = d["quote_checkin_dow"].isin([4, 5]).astype(float).where(d["quote_checkin_dow"].notna())
    d["quote_is_monthly"] = (d["quote_nights"] >= 28).astype(float).where(d["quote_nights"].notna())
    # clé quartier (ville + quartier) pour le target encoding
    d["nbhd_key"] = d["city"] + "|" + d["neighbourhood_cleansed"].fillna("missing")
    return d


def make_targets(d: pd.DataFrame, train_mask: pd.Series) -> pd.DataFrame:
    """Cibles : log_price_eur (déjà là), is_superhost, price_tier (terciles de prix PAR VILLE, bornes apprises sur le train).
    Limite : les bornes sont calculées sur tout le train, donc légèrement vues par les plis de validation interne."""
    d = d.copy()
    d["is_superhost"] = d["host_is_superhost"]
    tr = d[train_mask & d["in_reg_set"]]
    edges = tr.groupby("city")["price_eur"].quantile([1 / 3, 2 / 3]).unstack()
    edges.columns = ["q33", "q66"]
    edges.to_csv(ROOT / "models" / "price_tier_edges.csv")
    q33, q66 = d["city"].map(edges.q33), d["city"].map(edges.q66)
    tier = np.where(d["price_eur"] <= q33, 0, np.where(d["price_eur"] <= q66, 1, 2)).astype(float)
    d["price_tier"] = np.where(d["price_eur"].notna(), tier, np.nan)  # 0 = bas, 1 = moyen, 2 = haut
    return d


# ---------------------------------------------------------------------------------------------------------------
# Groupes de variables et jeux de variables (pour la sélection de la phase 4 et le test avec/sans de la phase 7)
# ---------------------------------------------------------------------------------------------------------------
def amenity_cols(columns) -> list[str]:
    return [c for c in columns if c.startswith("am_")]


def feature_groups(columns) -> dict[str, list[str]]:
    cols = set(columns)
    g = {
        "property": ["accommodates", "bedrooms", "beds", "bath_count", "bath_shared", "room_type", "property_type", "is_studio", "beds_per_person",
                     "persons_per_bedroom", "bath_per_person", "log_min_nights", "log_max_nights"],
        "amenities": ["n_amenities", "n_amenities_other"] + amenity_cols(columns),
        "location": ["latitude", "longitude", "dist_center_km", "nbhd_key", "city"],
        "host": ["host_tenure_years", "user_tenure_years", "log_host_listings", "host_is_pro", "host_entire_share", "host_identity_verified",
                 "host_has_profile_pic", "host_is_local", "has_license", "has_host_about", "host_about_len"],
        "text": ["desc_len", "desc_words", "name_len"],
        "quality": ["review_scores_rating", "review_scores_accuracy", "review_scores_cleanliness", "review_scores_checkin",
                    "review_scores_communication", "review_scores_location", "review_scores_value"],
        "host_status": ["host_is_superhost"],
        "activity": ["log_n_reviews", "log_n_reviews_ltm", "log_reviews_per_month", "days_since_last_review", "days_since_first_review", "has_reviews",
                     "number_of_reviews_l30d", "availability_30", "availability_60", "availability_90", "availability_365", "has_availability", "is_stale"],
        "quote": ["log_quote_lead_days", "quote_nights", "quote_month_sin", "quote_month_cos", "quote_is_weekend", "quote_is_monthly"],
    }
    return {k: [c for c in v if c in cols] for k, v in g.items()}


def feature_set(columns, name: str = "core") -> list[str]:
    """core = caractéristiques du logement (aucune conséquence du prix) ; +quote = contexte du devis ; +quality ; +activity ; full = tout sauf colonnes à fuite."""
    g = feature_groups(columns)
    core = g["property"] + g["amenities"] + g["location"] + g["host"] + g["text"]
    sets = {"core": core, "core+quote": core + g["quote"], "core+quote+quality": core + g["quote"] + g["quality"] + g["host_status"],
            "full": core + g["quote"] + g["quality"] + g["host_status"] + g["activity"]}
    return sets[name]


# ---------------------------------------------------------------------------------------------------------------
# 2. Transformations avec apprentissage (dans les Pipelines)
# ---------------------------------------------------------------------------------------------------------------
class CityMedianImputer(BaseEstimator, TransformerMixin):
    """Imputation par la médiane de la ville (repli : médiane globale) + indicateurs de manquant.
    Les médianes et la liste des colonnes à indicateur sont apprises sur le train seulement."""

    def __init__(self, num_cols, city_col="city"):
        self.num_cols = num_cols
        self.city_col = city_col

    def fit(self, X, y=None):
        self.global_med_ = X[self.num_cols].median()
        self.city_med_ = X.groupby(self.city_col)[self.num_cols].median()
        self.ind_cols_ = [c for c in self.num_cols if X[c].isna().any()]
        return self

    def transform(self, X):
        X = X.copy()
        for c in self.ind_cols_:
            X[c + "_missing"] = X[c].isna().astype("float")
        med = X[[self.city_col]].join(self.city_med_, on=self.city_col)[self.num_cols]
        med = med.fillna(self.global_med_)
        X[self.num_cols] = X[self.num_cols].astype(float).fillna(med)
        return X


def split_columns(features: list[str], X: pd.DataFrame):
    cat_low = [c for c in ["room_type", "property_type", "city"] if c in features]
    te = [c for c in ["nbhd_key"] if c in features]
    num = [c for c in features if c not in cat_low + te]
    return num, cat_low, te


def build_preprocessor(features: list[str], X_example: pd.DataFrame, scale: bool = False, target_type: str = "continuous") -> Pipeline:
    """Pipeline de prétraitement complet.
    - numériques : médiane par ville + indicateurs de manquant, standardisation si scale=True (modèles linéaires, kNN, SVM, réseaux)
    - catégories faible cardinalité (room_type, property_type, city) : one-hot, modalités < 1 % regroupées
    - nbhd_key (ville|quartier, forte cardinalité) : TargetEncoder à validation croisée interne (cv=5), lissage automatique
    target_type : 'continuous' (régression), 'binary' ou 'multiclass' (classification).
    Si 'city' n'est pas dans les features (ex. leave-one-city-out sans identifiant de ville), la colonne reste nécessaire en entrée de l'imputeur."""
    num, cat_low, te = split_columns(features, X_example)
    imp_cols = num
    steps_num = [("scale", StandardScaler())] if scale else [("pass", "passthrough")]
    ct = ColumnTransformer(
        [("num", Pipeline(steps_num), make_num_selector(num)),
         ("ind", "passthrough", make_ind_selector(num)),
         ("cat", OneHotEncoder(min_frequency=0.01, handle_unknown="infrequent_if_exist", sparse_output=False), cat_low)]
        + ([("te", TargetEncoder(target_type=target_type, cv=5, shuffle=True, random_state=SEED), te)] if te else []),
        remainder="drop", verbose_feature_names_out=False)
    ct.set_output(transform="pandas")
    return Pipeline([("impute", CityMedianImputer(imp_cols)), ("ct", ct)])


def make_num_selector(num):
    return list(num)


def _ind_cols(X):
    # les colonnes *_missing n'existent qu'après fit de l'imputeur : sélection par motif
    return [c for c in X.columns if c.endswith("_missing")]


def make_ind_selector(num):
    return _ind_cols

def variables_table(d: pd.DataFrame) -> str:
    g = feature_groups(d.columns)
    inv = {c: k for k, v in g.items() for c in v}
    rows = ["| Variable | Famille | Définition | Type | % manquant | Jeu de variables |", "|---|---|---|---|---:|---|"]
    for c in d.columns:
        if c.startswith("am_"):
            continue
        fam, de = DEFS.get(c, ("?", "(non documentée)"))
        rows.append(f"| {c} | {fam} | {de} | {d[c].dtype} | {d[c].isna().mean() * 100:.1f} | {inv.get(c, '—')} |")
    am = amenity_cols(d.columns)
    rows.append(f"| am_* ({len(am)} colonnes) | équipements | Équipement présent (0/1) ; vocabulaire dans models/amenity_vocab.json | int8 | 0.0 | amenities |")
    return "\n".join(rows)


def run():
    cfg = load_config()
    df = pd.read_parquet(ROOT / cfg["paths"]["listings_clean"])
    texts = pd.read_parquet(ROOT / "data/processed/texts.parquet")
    d = add_text_features(df, texts)
    d = add_derived_features(d)
    d = make_targets(d, ~d["is_test"].astype(bool))
    d.to_parquet(ROOT / "data/processed/listings_features.parquet", index=False)
    unk = [c for c in d.columns if not c.startswith("am_") and c not in DEFS]
    print("colonnes non documentées :", unk)
    md = ["# Tableau des variables (Phase 3)", "",
          f"{len(d):,} annonces x {d.shape[1]} colonnes dans `listings_features.parquet`. Jeux de variables : `core` (caractéristiques du logement), `core+quote` (+ contexte du devis), "
          "`core+quote+quality` (+ notes et statut Superhost), `full` (+ activité : disponibilités, avis). Voir `feature_set()`.", "",
          "Non créées : surface par personne (aucune colonne de surface dans les données) ; langue de la description (99,4 % détectée en anglais, texte traduit à la source : aucune variance).", "",
          variables_table(d)]
    (ROOT / "reports" / "features_table.md").write_text("\n".join(md) + "\n", encoding="utf-8")

```python
"""Phase 3 : contrôles des Pipelines de features. Train uniquement (le test final n'est jamais lu).
 1. anti-fuite : le target encoding à CV interne n'est pas « trop beau » par rapport à un encodage naïf
 2. les Pipelines fonctionnent pour les 3 cibles (régression, binaire, multiclasse) et sur modalités inconnues
 3. comparaison indicative des jeux de variables (validation = 20 % des hôtes du train, groupés)
Écrit reports/features_check.md. Usage : python -m src.features_check
"""
import time

import lightgbm as lgb
import numpy as np
import pandas as pd
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.model_selection import GroupShuffleSplit

from src.data import ROOT, load_config
from src.features import build_preprocessor, feature_set

cfg = load_config()
d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
assert d["is_test"].dtype == bool
d = d[~d["is_test"]]  # le test final n'est pas lu
out = ["# Contrôles des features (Phase 3, train uniquement)", ""]

reg = d[d["in_reg_set"]].reset_index(drop=True)
groups = reg["city"] + "_" + reg["host_id"].astype(str)
tr, va = next(GroupShuffleSplit(n_splits=1, test_size=0.2, random_state=42).split(reg, groups=groups))
Xtr, Xva = reg.iloc[tr], reg.iloc[va]
ytr, yva = Xtr["log_price_eur"].values, Xva["log_price_eur"].values

# 1. fuite du target encoding : corrélation encodage / cible sur le train, croisé vs naïf
feats = feature_set(reg.columns, "core")
pp = build_preprocessor(feats, Xtr, scale=False)
Ztr = pp.fit_transform(Xtr, ytr)
Zva = pp.transform(Xva)
naive = Xtr.groupby("nbhd_key")["log_price_eur"].transform("mean")
te_cv = Ztr["nbhd_key"]
te_val = Zva["nbhd_key"]
c_naive, c_cv, c_val = np.corrcoef(naive, ytr)[0, 1], np.corrcoef(te_cv, ytr)[0, 1], np.corrcoef(te_val, yva)[0, 1]
out += ["## 1. Target encoding du quartier (nbhd_key)", "",
        f"- corrélation encodage naïf (moyenne du quartier, en échantillon) / cible, train : **{c_naive:.3f}** (optimiste : l'annonce se voit elle-même)",
        f"- corrélation encodage avec CV interne / cible, train : **{c_cv:.3f}**",
        f"- corrélation encodage appliqué à la validation (hôtes jamais vus) / cible : **{c_val:.3f}**",
        "- le TargetEncoder à CV interne (cross-fitting) donne sur le train une corrélation proche de celle observée en validation : pas de fuite visible.", "",
        f"Forme de la matrice de features `core` : {Ztr.shape[1]} colonnes (dont {sum(c.endswith('_missing') for c in Ztr.columns)} indicateurs de manquant).", ""]

# modalité inconnue au transform
Xu = Xva.head(5).copy()
Xu["nbhd_key"] = "paris|QUARTIER_INCONNU"
Xu["property_type"] = "Type inconnu"
assert pp.transform(Xu).notna().all().all(), "NaN après transform sur modalités inconnues"
out += ["- modalités inconnues (nouveau quartier, nouveau type de propriété) : gérées sans erreur (encodage = moyenne globale / catégorie 'infrequent').", ""]

# 2. Pipelines des 3 cibles (échantillon pour la vitesse)
smp = d.sample(40000, random_state=42)
res = []
for target, ttype, mdl in [("is_superhost", "binary", LogisticRegression(max_iter=300)), ("price_tier", "multiclass", LogisticRegression(max_iter=300))]:
    s = smp.dropna(subset=[target])
    s = s[s["in_reg_set"]] if target == "price_tier" else s
    f = feature_set(s.columns, "core")
    p = build_preprocessor(f, s, scale=True, target_type=ttype)
    Z = p.fit_transform(s, s[target].astype(int))
    mdl.fit(Z, s[target].astype(int))
    res.append(f"{target} ({ttype}) : OK, {Z.shape[1]} colonnes, accuracy en échantillon {mdl.score(Z, s[target].astype(int)):.3f} (non évaluée : simple test de fonctionnement)")
out += ["## 2. Pipelines pour les 3 cibles", ""] + [f"- {r}" for r in res] + [f"- régression (log_price_eur) : OK, voir ci-dessous.", ""]

# 3. comparaison indicative des jeux de variables
rows = []
for name in ["core", "core+quote", "core+quote+quality", "full"]:
    f = feature_set(reg.columns, name)
    t0 = time.time()
    pl = build_preprocessor(f, Xtr, scale=True)
    Ztr_l, Zva_l = pl.fit_transform(Xtr, ytr), pl.transform(Xva)
    rid = Ridge(alpha=10, random_state=42).fit(Ztr_l, ytr)
    pr = rid.predict(Zva_l)
    pt = build_preprocessor(f, Xtr, scale=False)
    Ztr_t, Zva_t = pt.fit_transform(Xtr, ytr), pt.transform(Xva)
    gb = lgb.LGBMRegressor(n_estimators=400, learning_rate=0.05, num_leaves=63, random_state=42, verbose=-1).fit(Ztr_t, ytr)
    pg = gb.predict(Zva_t)
    r = lambda p: float(np.sqrt(np.mean((yva - p) ** 2)))
    a = lambda p: float(np.mean(np.abs(yva - p)))
    rows.append([name, len(f), r(pr), a(pr), r(pg), a(pg), time.time() - t0])
    print(rows[-1])
base = [float(np.sqrt(np.mean((yva - np.median(ytr)) ** 2)))]
cmp = pd.DataFrame(rows, columns=["Jeu de variables", "nb variables", "Ridge RMSE", "Ridge MAE", "LightGBM RMSE", "LightGBM MAE", "temps (s)"]).round(4)
out += ["## 3. Comparaison indicative des jeux de variables (cible log_price_eur)", "",
        f"Validation : {len(va):,} annonces (20 % des hôtes du train, jeu in_reg_set) ; entraînement : {len(tr):,}. Référence « médiane globale » : RMSE {base[0]:.4f}.", "",
        cmp.to_markdown(index=False), "",
        "Ces chiffres sont indicatifs (un seul découpage, hyperparamètres par défaut) ; la comparaison rigoureuse en CV est faite en phases 4 à 6. `full` contient des variables d'activité (disponibilité, avis) qui sont en partie des conséquences du prix."]
(ROOT / "reports" / "features_check.md").write_text("\n".join(out) + "\n", encoding="utf-8")
print("\n".join(out))

"""Détection de langue des descriptions d'annonces (langdetect, seed 42), en parallèle.
Fonction déterministe par ligne, sans apprentissage sur les données : pas de fuite.
Usage : python -m src.lang   -> data/processed/listing_lang.parquet
"""
import re

import pandas as pd
from joblib import Parallel, delayed

from src.data import ROOT


def _detect_chunk(texts):
    from langdetect import DetectorFactory, detect

    DetectorFactory.seed = 42
    out = []
    for t in texts:
        t = re.sub(r"<[^>]+>|http\S+|[\d\W_]+", " ", t or "").strip()[:400]
        if len(t) < 20:
            out.append("unk")
            continue
        try:
            out.append(detect(t))
        except Exception:
            out.append("unk")
    return out


def run():
    d = pd.read_parquet(ROOT / "data/processed/listings_all.parquet", columns=["city", "id", "description", "name"])
    txt = d["description"].fillna(d["name"]).fillna("").tolist()
    chunks = [txt[i:i + 5000] for i in range(0, len(txt), 5000)]
    res = Parallel(n_jobs=14)(delayed(_detect_chunk)(c) for c in chunks)
    d["desc_lang"] = [x for r in res for x in r]
    d[["city", "id", "desc_lang"]].to_parquet(ROOT / "data/processed/listing_lang.parquet", index=False)
    print(d["desc_lang"].value_counts(normalize=True).head(10).round(3).to_dict())


if __name__ == "__main__":
    run()

## 6. Variables exclues pour fuite de données

Une **fuite de données** (*data leakage*) survient quand une variable contient une information qu'on n'aurait pas au moment de la prédiction. Le modèle paraît excellent… puis s'effondre en situation réelle. On passe chaque variable suspecte au **détecteur de fuites** : son pouvoir prédictif *seule*.

````python
from sklearn.metrics import roc_auc_score
from scipy.stats import spearmanr
d = feat[~feat.is_test & feat.has_price]
ok = d.estimated_occupancy_l365d > 0
ratio = (d.estimated_revenue_l365d[ok] / (d.price_local[ok] * d.estimated_occupancy_l365d[ok])).round(2)
rows = [
    ("estimated_revenue_l365d", "régression du prix", "revenu = prix × occupation estimée", f"ratio revenu / (prix × occupation) = 1,00 pour {(ratio == 1).mean():.0%} des annonces", "FUITE : exclue"),
    ("price_quote_price_per_night", "régression du prix", "copie du prix", f"Spearman avec le prix (Paris, brut) = {spearmanr(*brut[['price', 'price_quote_price_per_night']].dropna().T.values)[0]:.3f}", "FUITE : supprimée"),
]
for c, lab in [("availability_30", "disponibilité 30 j"), ("availability_365", "disponibilité 365 j"), ("log_n_reviews_ltm", "avis sur 12 mois"), ("log_reviews_per_month", "avis par mois")]:
    x = d[[c, "log_price_eur"]].dropna()
    rows.append((c, "régression du prix", f"{lab} : conséquence du prix (une annonce chère se réserve moins)", f"corrélation de Spearman avec log(prix) = {spearmanr(x[c], x.log_price_eur)[0]:+.2f}", "exclue du jeu de référence"))
s = feat[~feat.is_test].dropna(subset=["is_superhost"])
for c, lab in [("review_scores_rating", "note globale"), ("review_scores_cleanliness", "propreté"), ("log_n_reviews_ltm", "avis sur 12 mois"), ("log_reviews_per_month", "avis par mois")]:
    x = s[[c, "is_superhost"]].dropna(); a = roc_auc_score(x.is_superhost.astype(int), x[c])
    rows.append((c, "Superhost", f"{lab} : critère d'attribution ET conséquence du statut", f"AUC seule = {a:.2f}", "à tester avec / sans (chapitre 14)"))
pd.DataFrame(rows, columns=["variable", "cible concernée", "pourquoi suspecte", "preuve chiffrée", "verdict"])
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #E8A94A;background:#E8A94A14;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#E8A94A">Ce qu'on en retient</span><br><b>Verdict.</b> <code>estimated_revenue_l365d</code> vaut <i>exactement</i> prix × occupation : la prédire, c'est recopier la réponse. Elle est exclue. Les <b>disponibilités et volumes d'avis</b> sont des conséquences du prix (une annonce chère se réserve moins) : elles amélioreraient artificiellement le score, mais ne sont pas utilisables par un hôte qui <i>fixe</i> son prix. Pour le <b>Superhost</b>, le cas est plus subtil : le statut est attribué <i>d'après</i> les notes et l'activité, qui sont donc à la fois cause et conséquence. On les <b>teste avec et sans</b> au chapitre 14.</div>

> **Remarque.** Dans un projet réel, c'est ici que beaucoup de modèles « parfaits » meurent : avec les variables d'activité, le score de prix monte à un RMSE de 0,354 contre 0,374 sans, mais ce gain n'existe pas pour un hôte qui <b>fixe</b> son prix. On le garde en tête : on a préféré un modèle un peu moins précis mais <b>utilisable</b>.

````python
quarantaine = {c: "FUITE_" + c for c in ["estimated_revenue_l365d", "estimated_occupancy_l365d"]}
from src.config_utils import leaky_columns  # liste des colonnes interdites dans le modèle de prix (config.yaml)
print("Colonnes interdites dans la régression de prix (config.yaml > leaky_columns) :", leaky_columns())
````

## 7. Contrôle du nettoyage

Un bon nettoyage **se prouve**. Chaque règle est écrite sous forme d'assertion : si quelqu'un modifie le pipeline et casse une règle, l'exécution s'arrête.

````python
cl = pd.read_parquet(PROC / "listings_clean.parquet")
train_hosts = set(zip(cl.loc[~cl.is_test, "city"], cl.loc[~cl.is_test, "host_id"]))
test_hosts = set(zip(cl.loc[cl.is_test, "city"], cl.loc[cl.is_test, "host_id"]))
med = cl.groupby("city").price_eur.median()
tests = {
    "Une annonce = une ligne : (ville, id) unique": cl.duplicated(["city", "id"]).sum() == 0,
    "Aucune ligne supprimée (485 128 annonces)": len(cl) == 485128,
    "Tous les prix en euros sont strictement positifs": (cl.price_eur.dropna() > 0).all(),
    "Médiane de Londres comparable à celle de Paris (conversion GBP correcte : entre 150 et 260 €)": 150 < med["london"] < 260,
    "Médiane de Tokyo plausible en euros (50 à 250 €)": 50 < med["tokyo"] < 250,
    "Médiane de Mexico plausible en euros (30 à 150 €)": 30 < med["mexico-city"] < 150,
    "AUCUN HÔTE à la fois dans le train et dans le test (pas de fuite par hôte)": len(train_hosts & test_hosts) == 0,
    "Le test final représente 14,3 % des annonces (± 0,5 point)": abs(cl.is_test.mean() - 1 / 7) < 0.005,
    "Chaque ville a des annonces dans le test (stratification)": cl.groupby("city").is_test.sum().min() > 0,
    "Aucune colonne 100 % vide ne subsiste": not any(cl[c].isna().all() for c in cl.columns),
    "Salles de bain : valeurs entre 0 et 50 (quelques hôtels / auberges déclarent plus de 20 : voir la note sous le tableau)": cl.bath_count.dropna().between(0, 50).all(),
    "maximum_nights plafonné à 1125": cl.maximum_nights.max() <= 1125,
    "Le nom de l'hôte (RGPD) n'est plus dans le fichier": "host_name" not in cl.columns,
    "Les 164 équipements du vocabulaire existent comme colonnes": sum(c.startswith("am_") for c in cl.columns) == 164,
    "Les drapeaux d'anomalies existent": {"is_price_outlier", "is_inconsistent", "is_near_duplicate", "is_geo_outlier"} <= set(cl.columns),
}
res = pd.DataFrame({"Règle": list(tests), "Résultat": ["OK" if bool(v) else "ÉCHEC" for v in tests.values()]})
display(res.style.hide(axis="index"))
assert all(bool(v) for v in tests.values()), "Une règle de nettoyage est cassée"
print(f"{sum(bool(v) for v in tests.values())} / {len(tests)} règles validées.")
print(f"Note : {int((cl.bath_count > 20).sum())} annonces déclarent plus de 20 salles de bain (surtout des chambres d'hôtel : valeurs déclarées, rares et non corrigées ; une winsorisation comme pour bedrooms et beds serait à envisager).")
````

## 8. Analyse exploratoire du prix

On travaille maintenant sur les annonces **du train avec prix** (le jeu de test reste mis de côté). Prix en **euros**, tous les chiffres sont calculés dans les cellules ci-dessous.

````python
E = feat[~feat.is_test & feat.has_price].copy()
E["price_rel"] = E.price_eur / E.groupby("city").price_eur.transform("median")
print(f"{fmt(len(E))} annonces avec prix dans le train ; {E.city.nunique()} villes")
````

### 8.1 Le prix : une distribution écrasée, un logarithme qui la redresse

### 8.7 Le podium des variables

On mesure la force du lien entre chaque variable et `log(prix)` : **corrélation de Spearman en valeur absolue** pour les variables quantitatives, **rapport de corrélation η** (part de variance expliquée, en racine) pour les catégories. 0 = aucun lien, 1 = lien parfait.

````python
def eta(cat, y):
    g = y.groupby(cat); ss_b = (g.count() * (g.mean() - y.mean()) ** 2).sum(); return float(np.sqrt(ss_b / ((y - y.mean()) ** 2).sum()))
S_ = E.sample(120000, random_state=SEED)
y_ = S_.log_price_eur
res = {}
for c in ["accommodates", "bedrooms", "beds", "bath_count", "n_amenities", "dist_center_km", "log_min_nights", "quote_nights", "log_quote_lead_days", "review_scores_rating", "review_scores_location",
          "host_tenure_years", "log_host_listings", "desc_len", "latitude", "longitude"]:
    x = S_[[c]].assign(y=y_).dropna(); res[c] = abs(x[c].corr(x.y, method="spearman"))
for c in ["city", "nbhd_key", "room_type", "property_type"]:
    res[c + " (η)"] = eta(S_[c].fillna("na"), y_)
res = pd.Series(res).sort_values()
fig, ax = plt.subplots(figsize=(10, 7)); ax.barh(res.index, res.values, color=[PAL["rose"] if "η" in i else PAL["bleu"] for i in res.index], edgecolor="white", linewidth=0.8)
ax.set(title="Podium : force du lien avec log(prix) (bleu : quantitatif ; rose : catégoriel)", xlabel="|Spearman| ou η"); plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #E07A67;background:#E07A6714;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#E07A67">Ce qu'on en retient</span><br>Le podium confirme trois histoires : <b>la ville et le quartier</b> (η très élevés : ils portent à eux seuls une grande part de la variance), <b>la capacité du logement</b> (<code>accommodates</code>, <code>bedrooms</code>, <code>beds</code>), et <b>la durée du séjour du devis</b>. Les variables de texte (longueur de la description) et d'ancienneté de l'hôte pèsent peu. Attention : η mesure un <b>lien</b>, pas une causalité, et ne dit rien de la manière dont les variables se combinent (voir le chapitre 20).</div>

Usage : python -m src.eda          (Paris d'abord, puis les 15 villes)
Aucune donnée n'est modifiée ; tous les chiffres des insights sont calculés ici.
Les taux de change n'étant pas encore validés, les prix sont comparés entre villes
via l'écart à la médiane de la ville (prix relatif) et le log, pas en devise commune.
"""

matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import pyarrow.parquet as pq

FIG = ROOT / "reports" / "figures"
FIG.mkdir(parents=True, exist_ok=True)

INS: list[str] = []  # insights chiffrés


def save(name):
    plt.tight_layout()
    plt.savefig(FIG / f"eda_{name}.png", dpi=130)
    plt.close()


def haversine(lat1, lon1, lat2, lon2):
    """Distance en km entre deux points (degrés)."""
    p = np.pi / 180
    a = np.sin((lat2 - lat1) * p / 2) ** 2 + np.cos(lat1 * p) * np.cos(lat2 * p) * np.sin((lon2 - lon1) * p / 2) ** 2
    return 12742 * np.arcsin(np.sqrt(a))


def add_insight(txt):
    INS.append(txt)
    print("-", txt)


def run():
    cfg = load_config()
    df = load_listings(cfg)
    df["superhost"] = df["host_is_superhost"].map({"t": 1, "f": 0})
    df["log_price"] = np.log(df["price"])

    empty_all = miss.columns[(miss > 99.9).all()].tolist()

    plt.figure(figsize=(16, 7))

    pm = miss["price"]

    add_insight(f"`price` manque pour {df.price.isna().mean() * 100:.1f} % des annonces (de {pm.min():.1f} % à {pm.idxmin()} à {pm.max():.1f} % à {pm.idxmax()}) : "
                f"{df.price.notna().sum():,} annonces exploitables pour la régression.")

    add_insight(f"Biais de sélection : parmi les annonces SANS prix, {na_av[True]:.0f} % ont availability_365 = 0, contre {na_av[False]:.0f} % parmi celles avec prix "
                "(le prix n'existe que pour les annonces réservables).")
    sparse = (miss.mean() > 30) & ~miss.columns.isin(empty_all)
    add_insight("Colonnes partiellement vides (>30 % en moyenne) : " + ", ".join(f"{c} ({miss[c].mean():.0f} %)" for c in miss.columns[sparse]) + ".")

    d = df[df.price.notna()]

    ax[1].set(title="Paris : log(prix)", xlabel="log(prix en EUR)", ylabel="Nombre d'annonces")
    save("02_price_paris_raw_vs_log")
    sk_raw, sk_log = paris.price.skew(), paris.log_price.skew()
    add_insight(f"Paris : asymétrie du prix = {sk_raw:.1f} (brut) contre {sk_log:.2f} (log) : le log(prix) est quasi symétrique, d'où la cible log(price).")

    fig, ax = plt.subplots(figsize=(14, 6))
    order = d.groupby("city").log_price.median().sort_values().index
    sns.boxplot(data=d, x="city", y="log_price", order=order, fliersize=1, ax=ax, color="#9ecae1")
    ax.set(title="log(prix) par ville, en devise locale (non convertie)", xlabel="Ville", ylabel="log(prix, devise locale)")

    med = d.groupby("city").price.median()

    add_insight("Médianes en devise locale (NON comparables sans conversion) : " + ", ".join(f"{c} {med[c]:,.0f} {cur[c]}" for c in med.sort_values().index) + ".")

    add_insight(f"Prix < 0,1x la médiane de sa ville : {(d.price_rel < 0.1).sum()} annonces (prix aberrants bas possibles : devis par mois ou erreurs).")
    qn = d.groupby("city").price_rel.quantile([0.01, 0.99]).unstack()

    # ------------------------------------------------------------------ 3. Numériques
    num = ["price", "accommodates", "bedrooms", "beds", "minimum_nights", "maximum_nights", "availability_365", "number_of_reviews", "number_of_reviews_ltm",

    fig, ax = plt.subplots(1, 2, figsize=(18, 8))

        sns.heatmap(d[cols].corr(method=m), annot=True, fmt=".2f", cmap="coolwarm", center=0, vmin=-1, vmax=1, annot_kws={"size": 6}, ax=a, cbar=False)
        a.set_title(f"Corrélation de {m.capitalize()}")
    save("05_correlations")
    cs = d[cols].corr("spearman")["log_price"].drop("log_price").sort_values(key=abs, ascending=False)
    add_insight("Corrélations de Spearman avec log(price) : " + ", ".join(f"{k} {v:+.2f}" for k, v in cs.head(5).items()) + ".")
    cm = d[cols].corr("spearman").abs()
    pairs = [(a, b, cm.loc[a, b]) for i, a in enumerate(cols) for b in cols[i + 1:] if cm.loc[a, b] > 0.85 and "log_price" not in (a, b)]
    add_insight("Paires redondantes (|Spearman| > 0,85) : " + ("; ".join(f"{a}~{b} ({v:.2f})" for a, b, v in pairs) or "aucune") + ".")
    sk = st["skew"].abs().sort_values(ascending=False)
    add_insight("Variables les plus asymétriques (à log-transformer) : " + ", ".join(f"{k} ({st.loc[k, 'skew']:.0f})" for k in sk.index[:5]) + ".")

    # ------------------------------------------------------------------ 4. Catégorielles
    cat = ["property_type", "room_type", "neighbourhood_cleansed", "neighbourhood_group_cleansed", "license", "host_location", "source"]

    body = [f"{i}. {t}" for i, t in enumerate(INS, 1)]

if __name__ == "__main__":

## 9. Sélection de variables

On compare des **familles de méthodes** de sélection sur la tâche de référence (régression de `log(prix)`) : filtres, wrappers, méthodes embarquées, méthodes agnostiques (permutation, SHAP), VIF, et la PCA comme alternative par transformation.

**Protocole** : 60 000 annonces du train (échantillon, pour que les méthodes lentes restent praticables), **5 plis groupés par hôte**. Dans chaque pli, le prétraitement **et** la sélection sont ajustés sur le train du pli ; chaque sous-ensemble est évalué avec le même modèle (LightGBM, 200 arbres), Ridge en témoin. La permutation est calculée sur un split interne (jamais sur le pli de validation).

<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5BAA97">Ce qu'on en retient</span><br><b>Aucune méthode ne bat « toutes les variables »</b> (RMSE 0,3854) : LightGBM tolère bien les variables peu utiles. Le bénéfice de la sélection est la <b>simplicité et l'interprétabilité</b>, pas la performance. Un tirage de 60 variables au hasard donne 0,4812, donc la sélection a bien un sens. Le <b>meilleur compromis</b> est SHAP : 81 variables sur 259 pour +0,002 de RMSE ; l'importance LightGBM en garde 39 pour +0,006. Les <b>filtres univariés</b> sont les moins fiables à k = 60 (corrélation 0,429 ; chi² 0,480, proche du hasard) : ils ignorent les interactions. L'<b>information mutuelle</b> (0,397) fait mieux car elle capte les relations non linéaires. Le <b>SFS</b> guidé par Ridge est mauvais pour LightGBM (0,431) et coûteux (87 s) : un sélecteur linéaire ne retient pas ce dont un modèle non linéaire a besoin. Le <b>VIF</b> ne retire que 23 colonnes sur 259 sans changer l'erreur : la multicolinéarité gêne les coefficients linéaires, pas les arbres. La <b>PCA</b> (0,451 avec 191 composantes) est nettement moins bonne : <i>hypothèse non testée</i>, les axes de plus forte variance ne sont pas ceux qui expliquent le prix.</div>

> **Remarque.** <b>Le biais des importances d'arbres.</b> Les variables continues à forte cardinalité sont <b>surévaluées</b> par l'impureté d'une forêt aléatoire : <code>desc_len</code> est au rang <b>29</b> en impureté contre <b>70</b> en permutation, <code>name_len</code> 30 contre 66. Un arbre dispose de beaucoup de seuils candidats sur ces variables et peut y ajuster du bruit. À l'inverse, un indicateur binaire utile comme la climatisation est <b>sous-évalué</b> (rang 43 en impureté, 20 en permutation et en SHAP). Permutation et SHAP, calculées hors échantillon d'entraînement, sont plus fiables. Le quartier encodé reste premier pour toutes les mesures : sa place n'est pas un artefact de cardinalité.

> **Remarque.** <b>Décision.</b> Jeu de référence <code>core+quote+quality</code> ; les 81 variables SHAP (<code>models/selected_features.json</code>) servent de jeu réduit pour les modèles linéaires et l'interprétation ; les modèles d'arbres peuvent utiliser toutes les variables. <b>Limites</b> : échantillon de 60 000 annonces, une seule graine, hyperparamètres fixes, k = 60 non optimisé pour les filtres.

```python
"""Phase 4 : comparaison de familles de sélection de variables (régression sur log_price_eur).

Protocole (anti-fuite) : échantillon du TRAIN (jeu in_reg_set) ; CV 5 plis groupée par hôte (stratifiée par ville).
Dans chaque pli : prétraitement (imputation, encodage, target encoding) ajusté sur le train du pli, puis chaque méthode
sélectionne ses variables sur le train du pli seulement ; évaluation sur le pli de validation avec un modèle commun
(LightGBM, et Ridge en témoin). Le test final n'est jamais lu.
Usage : python -m src.selection   (résultats par pli mis en cache dans models/_sel_cache pour reprise)
"""
import json
import pickle
import time
import warnings
from itertools import combinations

import lightgbm as lgb
import numpy as np
import pandas as pd
import shap
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor
from sklearn.feature_selection import RFECV, SelectKBest, SequentialFeatureSelector, chi2, f_classif, f_regression, mutual_info_regression
from sklearn.inspection import permutation_importance
from sklearn.linear_model import ElasticNetCV, LassoCV, Ridge
from sklearn.model_selection import GroupKFold, GroupShuffleSplit, StratifiedGroupKFold

from src.data import ROOT, load_config
from src.features import build_preprocessor, feature_set

warnings.filterwarnings("ignore")
SEED = 42
K = 60  # nombre de variables conservées par les méthodes à « k fixé »
CACHE = ROOT / "models" / "_sel_cache"
LGB_PARAMS = dict(n_estimators=200, learning_rate=0.1, num_leaves=31, random_state=SEED, n_jobs=16, verbose=-1)


def rmse(a, b):
    return float(np.sqrt(np.mean((a - b) ** 2)))


def mae(a, b):
    return float(np.mean(np.abs(a - b)))


def cum_select(imp: pd.Series, share=0.95) -> list[str]:
    """Variables couvrant `share` de l'importance totale (importances positives)."""
    s = imp.clip(lower=0).sort_values(ascending=False)
    c = s.cumsum() / s.sum()
    return s.index[: int((c < share).sum()) + 1].tolist()


# ------------------------------------------------------------------ méthodes de sélection : (Ztr, ytr, Zs, ctx) -> liste de noms
def m_variance(Z, y, Zs, ctx):
    top = Z.apply(lambda s: s.value_counts(normalize=True).iloc[0])
    return top[top <= 0.995].index.tolist()


def m_corr(Z, y, Zs, ctx):
    sel = SelectKBest(f_regression, k=K).fit(Z, y)
    return Z.columns[sel.get_support()].tolist()


def m_mi(Z, y, Zs, ctx):
    idx = ctx["sub"][:15000]
    disc = (Z.iloc[idx].nunique() <= 2).values
    mi = mutual_info_regression(Z.iloc[idx], y[idx], discrete_features=disc, random_state=SEED, n_jobs=-1)
    ctx["imp"]["mi"] = pd.Series(mi, Z.columns)
    return Z.columns[np.argsort(mi)[::-1][:K]].tolist()


def m_anova(Z, y, Zs, ctx):
    bins = pd.qcut(y, 3, labels=False)
    sel = SelectKBest(f_classif, k=K).fit(Z, bins)
    return Z.columns[sel.get_support()].tolist()


def m_chi2(Z, y, Zs, ctx):
    bins = pd.qcut(y, 3, labels=False)
    mm = (Z - Z.min()) / (Z.max() - Z.min()).replace(0, 1)
    sel = SelectKBest(chi2, k=K).fit(mm, bins)
    return Z.columns[sel.get_support()].tolist()


def m_rfecv(Z, y, Zs, ctx):
    idx = ctx["sub"][:12000]
    g = ctx["groups"][idx]
    cv = list(GroupKFold(3).split(Zs.iloc[idx], groups=g))
    r = RFECV(Ridge(alpha=10), step=8, cv=cv, scoring="neg_root_mean_squared_error", min_features_to_select=10, n_jobs=-1).fit(Zs.iloc[idx], y[idx])
    return Zs.columns[r.support_].tolist()


def m_sfs(Z, y, Zs, ctx):
    idx = ctx["sub"][:8000]
    pool = Zs.columns[SelectKBest(f_regression, k=80).fit(Zs.iloc[idx], y[idx]).get_support()]  # pré-filtre : 80 candidats
    cv = list(GroupKFold(3).split(Zs.iloc[idx], groups=ctx["groups"][idx]))
    s = SequentialFeatureSelector(Ridge(alpha=10), n_features_to_select=25, direction="forward", scoring="neg_root_mean_squared_error", cv=cv, n_jobs=-1)
    s.fit(Zs.iloc[idx][pool], y[idx])
    return pool[s.get_support()].tolist()


def m_lasso(Z, y, Zs, ctx):
    idx = ctx["sub"][:30000]
    m = LassoCV(cv=3, n_alphas=30, random_state=SEED, max_iter=3000, n_jobs=-1).fit(Zs.iloc[idx], y[idx])
    ctx["imp"]["lasso"] = pd.Series(np.abs(m.coef_), Zs.columns)
    return Zs.columns[m.coef_ != 0].tolist()


def m_enet(Z, y, Zs, ctx):
    idx = ctx["sub"][:30000]
    m = ElasticNetCV(l1_ratio=[0.3, 0.7, 0.95], cv=3, n_alphas=30, random_state=SEED, max_iter=3000, n_jobs=-1).fit(Zs.iloc[idx], y[idx])
    return Zs.columns[m.coef_ != 0].tolist()


def m_lgb_gain(Z, y, Zs, ctx):
    m = ctx["lgb_model"]
    imp = pd.Series(m.booster_.feature_importance("gain"), Z.columns)
    ctx["imp"]["lgb_gain"] = imp
    return cum_select(imp)


def m_rf(Z, y, Zs, ctx):
    idx = ctx["sub"][:20000]
    m = RandomForestRegressor(n_estimators=100, max_features=0.3, min_samples_leaf=5, n_jobs=-1, random_state=SEED).fit(Z.iloc[idx], y[idx])
    imp = pd.Series(m.feature_importances_, Z.columns)
    ctx["imp"]["rf_impurity"] = imp
    return cum_select(imp)


def m_perm(Z, y, Zs, ctx):
    # split interne du train du pli (par hôtes) : on n'évalue jamais l'importance sur le pli de validation
    itr, iva = next(GroupShuffleSplit(1, test_size=0.2, random_state=SEED).split(Z, groups=ctx["groups"]))
    m = lgb.LGBMRegressor(**LGB_PARAMS).fit(Z.iloc[itr], y[itr])
    iva = iva[:8000]
    r = permutation_importance(m, Z.iloc[iva], y[iva], n_repeats=3, random_state=SEED, scoring="neg_root_mean_squared_error", n_jobs=1)
    ctx["imp"]["permutation"] = pd.Series(r.importances_mean, Z.columns)
    return Z.columns[(r.importances_mean - r.importances_std) > 0].tolist()


def m_shap(Z, y, Zs, ctx):
    idx = ctx["sub"][:5000]
    sv = shap.TreeExplainer(ctx["lgb_model"]).shap_values(Z.iloc[idx])
    imp = pd.Series(np.abs(sv).mean(0), Z.columns)
    ctx["imp"]["shap"] = imp
    return cum_select(imp)


def m_vif(Z, y, Zs, ctx):
    idx = ctx["sub"][:15000]
    X = Zs.iloc[idx]
    X = X.loc[:, X.std() > 0]
    cols = list(X.columns)
    C = np.corrcoef(X.values, rowvar=False)
    keep = np.arange(len(cols))
    while True:
        sub = C[np.ix_(keep, keep)]
        vif = np.diag(np.linalg.inv(sub + 1e-6 * np.eye(len(keep))))
        if vif.max() <= 10 or len(keep) <= 5:
            break
        keep = np.delete(keep, int(vif.argmax()))
    return [cols[i] for i in keep]


def m_random(Z, y, Zs, ctx):
    return list(np.random.RandomState(SEED + ctx["fold"]).choice(Z.columns, K, replace=False))


SELECTORS = [
    ("Toutes les variables (référence)", "témoin", lambda Z, y, Zs, c: Z.columns.tolist()),
    ("60 variables aléatoires (témoin)", "témoin", m_random),
    ("Variance quasi nulle (>99,5 % identique)", "filtre", m_variance),
    ("Corrélation (f_regression, k=60)", "filtre", m_corr),
    ("Information mutuelle (k=60)", "filtre", m_mi),
    ("ANOVA F sur tercile de prix (k=60)", "filtre", m_anova),
    ("Chi² sur tercile de prix (k=60)", "filtre", m_chi2),
    ("RFECV (Ridge)", "wrapper", m_rfecv),
    ("SFS avant (Ridge, 25 parmi 80)", "wrapper", m_sfs),
    ("Lasso (LassoCV)", "embedded", m_lasso),
    ("Elastic Net (CV)", "embedded", m_enet),
    ("LightGBM gain (95 % cumulé)", "embedded", m_lgb_gain),
    ("Random Forest impurité (95 % cumulé)", "embedded", m_rf),
    ("Permutation (LightGBM, split interne)", "agnostique", m_perm),
    ("SHAP |moyenne| (95 % cumulé)", "agnostique", m_shap),
    ("VIF itératif (seuil 10)", "colinéarité", m_vif),
]


def load_data(cfg):
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    d = d[~d["is_test"] & d["in_reg_set"]]
    n = cfg["sampling"].get("selection_n", 60000)
    d = d.sample(n, random_state=SEED).reset_index(drop=True)
    return d


def run_fold(k, tr, va, d, feats, y):
    f = CACHE / f"fold{k}.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    Xtr, Xva = d.iloc[tr], d.iloc[va]
    ytr, yva = y[tr], y[va]
    pp = build_preprocessor(feats, Xtr, scale=False)
    Z = pp.fit_transform(Xtr, ytr)
    Zv = pp.transform(Xva)
    mu, sd = Z.mean(), Z.std().replace(0, 1)
    Zs, Zvs = (Z - mu) / sd, (Zv - mu) / sd
    groups = (Xtr["city"] + "_" + Xtr["host_id"].astype(str)).values
    rng = np.random.RandomState(SEED + k)
    ctx = {"sub": rng.permutation(len(Z)), "groups": groups, "imp": {}, "fold": k}
    t0 = time.time()
    ctx["lgb_model"] = lgb.LGBMRegressor(**LGB_PARAMS).fit(Z, ytr)
    t_lgb_full = time.time() - t0
    out = {"fold": k, "n_features_total": Z.shape[1], "methods": {}, "importances": {}}
    for name, fam, fn in SELECTORS:
        t0 = time.time()
        S = fn(Z, ytr, Zs, ctx)
        t_sel = time.time() - t0
        if name.startswith(("LightGBM gain", "SHAP")):
            t_sel += t_lgb_full  # on compte aussi l'entraînement du modèle utilisé
        t0 = time.time()
        g = lgb.LGBMRegressor(**LGB_PARAMS).fit(Z[S], ytr)
        t_fit = time.time() - t0
        pg = g.predict(Zv[S])
        rd = Ridge(alpha=10).fit(Zs[S], ytr)
        pr = rd.predict(Zvs[S])
        out["methods"][name] = dict(family=fam, n=len(S), selected=S, lgb_rmse=rmse(yva, pg), lgb_mae=mae(yva, pg), ridge_rmse=rmse(yva, pr), ridge_mae=mae(yva, pr),
                                    t_select=t_sel, t_fit=t_fit)
        print(f"fold {k} | {name:45s} n={len(S):3d} lgb_rmse={rmse(yva, pg):.4f} t_sel={t_sel:.1f}s", flush=True)
    # PCA (transformation, pas sélection)
    for nm, nc in [("PCA (95 % de variance)", 0.95), ("PCA (60 composantes)", 60)]:
        t0 = time.time()
        p = PCA(n_components=nc, random_state=SEED).fit(Zs)
        P, Pv = p.transform(Zs), p.transform(Zvs)
        t_sel = time.time() - t0
        t0 = time.time()
        g = lgb.LGBMRegressor(**LGB_PARAMS).fit(P, ytr)
        t_fit = time.time() - t0
        pg = g.predict(Pv)
        pr = Ridge(alpha=10).fit(P, ytr).predict(Pv)
        out["methods"][nm] = dict(family="transformation", n=P.shape[1], selected=None, lgb_rmse=rmse(yva, pg), lgb_mae=mae(yva, pg), ridge_rmse=rmse(yva, pr),
                                  ridge_mae=mae(yva, pr), t_select=t_sel, t_fit=t_fit)
        print(f"fold {k} | {nm:45s} n={P.shape[1]:3d} lgb_rmse={rmse(yva, pg):.4f}", flush=True)
    out["importances"] = {k_: v.to_dict() for k_, v in ctx["imp"].items()}
    CACHE.mkdir(parents=True, exist_ok=True)
    pickle.dump(out, open(f, "wb"))
    return out


def jaccard(sets):
    js = [len(a & b) / len(a | b) for a, b in combinations([set(s) for s in sets], 2)]
    return float(np.mean(js))


def run():
    cfg = load_config()
    CACHE.mkdir(parents=True, exist_ok=True)
    d = load_data(cfg)
    feats = feature_set(d.columns, "core+quote+quality")
    y = d["log_price_eur"].values
    groups = d["city"] + "_" + d["host_id"].astype(str)
    cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=SEED)
    folds = [run_fold(k, tr, va, d, feats, y) for k, (tr, va) in enumerate(cv.split(d, d["city"], groups))]

    rows = []
    for name in folds[0]["methods"]:
        ms = [f["methods"][name] for f in folds]
        sel = [m["selected"] for m in ms]
        rows.append(dict(méthode=name, famille=ms[0]["family"], n_variables=np.mean([m["n"] for m in ms]), lgb_rmse=np.mean([m["lgb_rmse"] for m in ms]),
                         lgb_rmse_std=np.std([m["lgb_rmse"] for m in ms]), lgb_mae=np.mean([m["lgb_mae"] for m in ms]), ridge_rmse=np.mean([m["ridge_rmse"] for m in ms]),
                         ridge_mae=np.mean([m["ridge_mae"] for m in ms]), t_selection_s=np.mean([m["t_select"] for m in ms]), t_entrainement_s=np.mean([m["t_fit"] for m in ms]),
                         jaccard=jaccard(sel) if sel[0] is not None else np.nan))
    res = pd.DataFrame(rows)
    res.to_csv(ROOT / "reports" / "selection_results.csv", index=False)

    # fréquence de sélection par variable et méthode (sur les 5 plis)
    freq = {}
    for name in folds[0]["methods"]:
        if folds[0]["methods"][name]["selected"] is None:
            continue
        c = pd.Series(0.0, index=sorted({v for f in folds for v in f["methods"][name]["selected"]}))
        for f in folds:
            c[f["methods"][name]["selected"]] += 1 / len(folds)
        freq[name] = c
    F = pd.DataFrame(freq).fillna(0)
    F.to_csv(ROOT / "reports" / "selection_frequency.csv")
    imps = {}
    for key in folds[0]["importances"]:
        imps[key] = pd.DataFrame([f["importances"][key] for f in folds]).mean()
    pd.DataFrame(imps).to_csv(ROOT / "reports" / "selection_importances.csv")
    print(res.round(4).to_string())


if __name__ == "__main__":
    run()
```

## 10. Stratégie de validation

**Question** : un modèle de prix entraîné sur des villes est-il bon (A) sur ces mêmes villes, (B) sur une **ville jamais vue**, (C) sur d'autres **dates de séjour** ?

| Schéma | Principe |
|---|---|
| **A** | 5 plis, stratifiés par ville, **groupés par hôte** (performance « intra-villes ») |
| **B** | **Leave-one-city-out** : 15 plis, entraînement sur 14 villes, test sur la 15e |
| **C** | Variante temporelle : entraînement sur les séjours de **septembre 2026**, test sur des hôtes **disjoints** avec check-in **à partir d'octobre** |

Trois variantes de variables mesurent l'effet ville : **V1** toutes ; **V2** sans identifiants de ville ni de quartier ; **V3** sans non plus la latitude et la longitude absolues (la distance au centre est conservée). Entraînement sur les 309 412 annonces du train.

```python
"""Métriques et schémas de validation réutilisables (Phase 5)."""
import numpy as np
import pandas as pd
from sklearn.model_selection import StratifiedGroupKFold

SEED = 42


def rmse(y, p):
    return float(np.sqrt(np.mean((np.asarray(y) - np.asarray(p)) ** 2)))


def mae(y, p):
    return float(np.mean(np.abs(np.asarray(y) - np.asarray(p))))


def r2(y, p):
    y, p = np.asarray(y), np.asarray(p)
    return float(1 - ((y - p) ** 2).sum() / ((y - y.mean()) ** 2).sum())


def mae_eur(y_log, p_log):
    """MAE en euros après retransformation exp() simple (la correction de biais est traitée en phase 6)."""
    return float(np.mean(np.abs(np.exp(np.asarray(y_log)) - np.exp(np.asarray(p_log)))))


def metrics(y, p) -> dict:
    return dict(rmse=rmse(y, p), mae=mae(y, p), r2=r2(y, p), mae_eur=mae_eur(y, p), n=len(y))


def group_key(df: pd.DataFrame) -> pd.Series:
    """Groupe de validation = hôte (un hôte est propre à une ville)."""
    return df["city"].astype(str) + "_" + df["host_id"].astype(str)


def scheme_a(df: pd.DataFrame, k: int = 5):
    """A : K-fold stratifié par ville, groupé par hôte (intra-villes)."""
    cv = StratifiedGroupKFold(n_splits=k, shuffle=True, random_state=SEED)
    yield from cv.split(df, df["city"], group_key(df))


def scheme_b(df: pd.DataFrame):
    """B : leave-one-city-out. Rend (idx_train, idx_val, ville)."""
    cities = df["city"].values
    for c in sorted(df["city"].unique()):
        yield np.where(cities != c)[0], np.where(cities == c)[0], c


def scheme_c(df: pd.DataFrame, cutoff="2026-10-01"):
    """C (temporel via la date de check-in du devis) : hôtes disjoints ; entraînement = hôtes du groupe 1 avec check-in < cutoff ;
    validation « tardive » = hôtes du groupe 2 avec check-in >= cutoff ; contrôle « précoce » = hôtes du groupe 2 avec check-in < cutoff."""
    ci = pd.to_datetime(df["scrape_date"]) + pd.to_timedelta(df["quote_lead_days"], unit="D")
    hosts = group_key(df).drop_duplicates()
    rng = np.random.RandomState(SEED)
    g1 = set(hosts[rng.rand(len(hosts)) < 0.5])
    in1 = group_key(df).isin(g1).values
    early = (ci < pd.Timestamp(cutoff)).values
    known = ci.notna().values
    return (np.where(in1 & early & known)[0], np.where(~in1 & ~early & known)[0], np.where(~in1 & early & known)[0])
```

<summary><code>src/validation.py</code> (115 lignes) : Comparaison des schémas A / B / C et des variantes de variables</summary>

```python
"""Phase 5 : comparaison des schémas de validation A / B / C (régression log_price_eur, LightGBM).
Train uniquement (jeu in_reg_set) ; le test final n'est pas lu. Prétraitement ajusté dans chaque pli sur le train du pli.
Variantes de variables :
  V1 toutes ; V2 sans city_* ni nbhd_key ; V3 = V2 sans latitude/longitude absolues (distance au centre conservée).
Usage : python -m src.validation   -> reports/validation_*.csv
"""
import time
import warnings

import lightgbm as lgb
import numpy as np
import pandas as pd

from src.data import ROOT
from src.evaluate import metrics, scheme_a, scheme_b, scheme_c
from src.features import build_preprocessor, feature_set

warnings.filterwarnings("ignore")
SEED = 42
PARAMS = dict(n_estimators=300, learning_rate=0.08, num_leaves=63, random_state=SEED, n_jobs=16, verbose=-1)


def variants(cols):
    v1 = list(cols)
    v2 = [c for c in v1 if not c.startswith("city_") and c != "nbhd_key"]
    v3 = [c for c in v2 if c not in ("latitude", "longitude")]
    return {"V1 toutes": v1, "V2 sans ville/quartier": v2, "V3 agnostique ville": v3}


def fit_eval(Z, y, tr, va, cols):
    m = lgb.LGBMRegressor(**PARAMS).fit(Z.iloc[tr][cols], y[tr])
    return m.predict(Z.iloc[va][cols])


def run():
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    d = d[~d["is_test"] & d["in_reg_set"]].reset_index(drop=True)
    feats = feature_set(d.columns, "core+quote+quality")
    y = d["log_price_eur"].values
    print(f"{len(d):,} annonces train/régression", flush=True)
    rows_city, rows_sum = [], []

    # ---------------- A : intra-villes
    oof = {v: np.zeros(len(d)) for v in variants(feats)}
    for k, (tr, va) in enumerate(scheme_a(d)):
        t0 = time.time()
        pp = build_preprocessor(feats, d.iloc[tr], scale=False)
        Ztr = pp.fit_transform(d.iloc[tr], y[tr])
        Zva = pp.transform(d.iloc[va])
        Z = pd.concat([Ztr, Zva])
        idx_tr, idx_va = np.arange(len(tr)), np.arange(len(tr), len(tr) + len(va))
        for name, cols in variants(Ztr.columns).items():
            oof[name][va] = fit_eval(Z, np.concatenate([y[tr], y[va]]), idx_tr, idx_va, cols)
        print(f"A pli {k} : {time.time() - t0:.0f}s", flush=True)
    for name, p in oof.items():
        m = metrics(y, p)
        rows_sum.append(dict(schema="A intra-villes (5 plis, hôtes groupés)", variante=name, **m))
        for c in sorted(d.city.unique()):
            i = (d.city == c).values
            rows_city.append(dict(schema="A", variante=name, city=c, bias=float((p[i] - y[i]).mean()), **metrics(y[i], p[i])))

    # ---------------- B : leave-one-city-out (+ recalibration par 200 annonces de la ville cible)
    rng = np.random.RandomState(SEED)
    oofB = {v: np.zeros(len(d)) for v in variants(feats)}
    for tr, va, c in scheme_b(d):
        t0 = time.time()
        pp = build_preprocessor(feats, d.iloc[tr], scale=False)
        Ztr = pp.fit_transform(d.iloc[tr], y[tr])
        Zva = pp.transform(d.iloc[va])
        Z = pd.concat([Ztr, Zva])
        yy = np.concatenate([y[tr], y[va]])
        idx_tr, idx_va = np.arange(len(tr)), np.arange(len(tr), len(tr) + len(va))
        calib = rng.choice(len(va), min(200, len(va)), replace=False)
        for name, cols in variants(Ztr.columns).items():
            p = fit_eval(Z, yy, idx_tr, idx_va, cols)
            oofB[name][va] = p
            res = y[va] - p
            shift = res[calib].mean()  # recalibration du niveau de prix avec 200 annonces de la nouvelle ville
            mm = metrics(y[va], p)
            rows_city.append(dict(schema="B", variante=name, city=c, bias=float(-res.mean()), rmse_recal200=float(np.sqrt(np.mean((res - shift) ** 2))),
                                  rmse_recal_oracle=float(res.std()), corr_within=float(np.corrcoef(y[va], p)[0, 1]), **mm))
        print(f"B {c} : {time.time() - t0:.0f}s", flush=True)
    for name, p in oofB.items():
        rows_sum.append(dict(schema="B leave-one-city-out", variante=name, **metrics(y, p)))

    # ---------------- C : temporel (date de check-in du devis), hôtes disjoints
    tr, late, early = scheme_c(d)
    print(f"C : train {len(tr):,} (check-in < 1er oct. 2026), validation tardive {len(late):,}, contrôle précoce {len(early):,}", flush=True)
    pp = build_preprocessor(feats, d.iloc[tr], scale=False)
    Ztr = pp.fit_transform(d.iloc[tr], y[tr])
    Z = pd.concat([Ztr, pp.transform(d.iloc[late]), pp.transform(d.iloc[early])])
    yy = np.concatenate([y[tr], y[late], y[early]])
    i_tr = np.arange(len(tr))
    i_late = np.arange(len(tr), len(tr) + len(late))
    i_early = np.arange(len(tr) + len(late), len(Z))
    date_cols = {"log_quote_lead_days", "quote_month_sin", "quote_month_cos", "quote_is_weekend"}
    for name, cols in [("toutes les variables", list(Ztr.columns)), ("sans variables de date du devis", [c for c in Ztr.columns if c not in date_cols])]:
        m = lgb.LGBMRegressor(**PARAMS).fit(Z.iloc[i_tr][cols], yy[i_tr])
        for lab, ii in [("C contrôle : hôtes disjoints, même période (check-in sept.)", i_early), ("C temporel : hôtes disjoints, check-in >= oct. 2026", i_late)]:
            p = m.predict(Z.iloc[ii][cols])
            rows_sum.append(dict(schema=lab, variante=name, **metrics(yy[ii], p)))
            if "temporel" in lab:
                lt = d.iloc[late].assign(p=p, y=yy[ii], ci=lambda x: pd.to_datetime(x.scrape_date) + pd.to_timedelta(x.quote_lead_days, unit="D"))
                lt["mois"] = lt.ci.dt.to_period("M").astype(str)
                g = lt.groupby("mois").apply(lambda x: pd.Series(dict(n=len(x), rmse=float(np.sqrt(np.mean((x.y - x.p) ** 2))), bias=float((x.p - x.y).mean()))), include_groups=False)
                g.insert(0, "variante", name)
                g.to_csv(ROOT / "reports" / f"validation_C_by_month_{'all' if name.startswith('toutes') else 'nodate'}.csv")

    pd.DataFrame(rows_sum).to_csv(ROOT / "reports" / "validation_summary.csv", index=False)
    pd.DataFrame(rows_city).to_csv(ROOT / "reports" / "validation_by_city.csv", index=False)
    print(pd.DataFrame(rows_sum).round(4).to_string())


if __name__ == "__main__":
    run()
```

## 11. Comparaison des modèles de régression

Onze modèles, des plus simples aux plus sophistiqués, plus deux références naïves, s'affrontent avec **leurs réglages par défaut** en **validation croisée à 5 plis groupés par hôte** (schéma A). Tous courent sur **les mêmes 80 000 annonces du train** et les mêmes plis : un échantillon, pour que le kNN, la forêt aléatoire et le Gradient Boosting restent praticables (ils seront réentraînés sur tout le train ensuite). Le prétraitement (imputation par ville, one-hot, target encoding du quartier) est **ajusté dans chaque pli** ; les modèles linéaires et le kNN reçoivent des variables standardisées.

Le boosting corrige ses erreurs étape par étape et capte les interactions (quartier × capacité × durée du séjour) que ne voit pas un modèle linéaire. <b>Les modèles linéaires</b> (0,428 pour les quatre) donnent tous le même score : Lasso et ElasticNet ne retirent presque rien, Ridge suffit. Ils expliquent déjà 72 % de la variance avec une formule lisible. <b>Le kNN</b> est le pire modèle (0,518) : en 259 dimensions, les « voisins » ne sont plus proches (malédiction de la dimension). <b>L'arbre seul</b> (0,472) sur-apprend ou sous-apprend selon la profondeur. <b>La médiane de la ville</b> donne 0,747 : c'est la barre à franchir, et <b>tous les modèles la battent largement</b> (R² de 0,15 contre 0,79 pour le meilleur). Enfin l'<b>empilement</b> (méta-modèle Ridge sur 5 modèles) gagne encore un peu (0,371), mais il pèse presque uniquement XGBoost (0,50) et LightGBM (0,43).</div>

> **Remarque.** <b>Une erreur de 0,37 en log, ça veut dire quoi ?</b> Une erreur de 0,37 correspond à se tromper d'un facteur d'environ ×1,45 ou ×0,69 : pour un logement à 150 €, cela fait de 100 € à 220 €. L'erreur absolue moyenne est de <b>70 € par nuit</b> (MAE sur tout l'échantillon, prix moyen élevé par quelques logements chers) et le MAPE de <b>31 %</b>. C'est utile pour situer un prix, pas pour le fixer au centime : d'où les <b>intervalles de prédiction</b> du chapitre 13.

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br>Prédire <code>log(prix)</code> puis appliquer <code>exp()</code> donne l'estimation <b>médiane</b> du prix, qui est <b>sous-estimée en moyenne</b> (E[exp(erreur)] > 1). La correction de Duan (« smearing », facteur ≈ 1,10) corrige ce biais <b>de la moyenne</b>. Mais ce n'est <b>pas</b> ce qui minimise l'erreur absolue en euros : la MAE est minimisée par la médiane. Résultat contre-intuitif mais cohérent : <b>la MAE augmente de 70 à 74 € avec la correction</b>. Décision : le <b>prix recommandé à un hôte est l'estimation médiane</b> <code>exp(prédiction)</code> ; le prix moyen corrigé est conservé à part (utile pour des totaux de marché, pas pour un prix individuel).</div>

```python
"""Catalogue de modèles (Phases 6, 7, 11) et utilitaires communs de validation croisée sur matrices prétraitées."""
import numpy as np
import pandas as pd
from sklearn.ensemble import GradientBoostingRegressor, HistGradientBoostingRegressor, RandomForestRegressor
from sklearn.linear_model import ElasticNet, Lasso, LinearRegression, Ridge
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor

SEED = 42


class CityMedianBaseline:
    """Baseline : médiane de log(prix) de la ville (repli : médiane globale pour une ville inconnue)."""

    def fit(self, cities, y):
        s = pd.Series(np.asarray(y)).groupby(np.asarray(cities)).median()
        self.med_, self.glob_ = s, float(np.median(y))
        return self

    def predict(self, cities):
        return pd.Series(np.asarray(cities)).map(self.med_).fillna(self.glob_).values


def regression_zoo(n_jobs=16):
    """nom -> (estimateur, a_besoin_de_scaling). Hyperparamètres « par défaut raisonnables », identiques pour tous les plis."""
    import lightgbm as lgb
    import xgboost as xgb

    return {
        "Régression linéaire": (LinearRegression(), True),
        "Ridge": (Ridge(alpha=10, random_state=SEED), True),
        "Lasso": (Lasso(alpha=0.001, max_iter=5000, random_state=SEED), True),
        "ElasticNet": (ElasticNet(alpha=0.001, l1_ratio=0.5, max_iter=5000, random_state=SEED), True),
        "kNN (k=15)": (KNeighborsRegressor(n_neighbors=15, weights="distance", n_jobs=n_jobs), True),
        "Arbre de décision": (DecisionTreeRegressor(max_depth=12, min_samples_leaf=20, random_state=SEED), False),
        "Random Forest": (RandomForestRegressor(n_estimators=200, min_samples_leaf=3, max_features=0.33, n_jobs=n_jobs, random_state=SEED), False),
        "Gradient Boosting (sklearn)": (GradientBoostingRegressor(n_estimators=200, max_depth=5, learning_rate=0.1, subsample=0.8, random_state=SEED), False),
        "HistGradientBoosting": (HistGradientBoostingRegressor(max_iter=300, learning_rate=0.08, random_state=SEED), False),
        "XGBoost": (xgb.XGBRegressor(n_estimators=400, learning_rate=0.08, max_depth=7, subsample=0.8, colsample_bytree=0.8, tree_method="hist", n_jobs=n_jobs,
                                     random_state=SEED, verbosity=0), False),
        "LightGBM": (lgb.LGBMRegressor(n_estimators=300, learning_rate=0.08, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8,
                                       n_jobs=n_jobs, random_state=SEED, verbose=-1), False),
    }


def fold_matrices(d, feats, tr, va, y, build_preprocessor):
    """Prétraitement ajusté sur le train du pli ; renvoie (Z_tr, Z_va, Zs_tr, Zs_va) (Zs = standardisé avec les stats du train du pli)."""
    pp = build_preprocessor(feats, d.iloc[tr], scale=False)
    Z = pp.fit_transform(d.iloc[tr], y[tr])
    Zv = pp.transform(d.iloc[va])
    mu, sd = Z.mean(), Z.std().replace(0, 1)
    return Z, Zv, (Z - mu) / sd, (Zv - mu) / sd
```

<summary><code>src/regression.py</code> (216 lignes) : Comparaison des modèles, réglage Optuna, prédictions hors échantillon, leave-one-city-out (chapitres 11 à 13)</summary>

```python
"""Phase 6 : régression du prix (log_price_eur).

Étapes (chaque étape met ses résultats en cache dans models/_reg_cache) :
  course   : 11 modèles + baselines, CV A (5 plis groupés par hôte), échantillon de 80 000 annonces du train
  tune     : Optuna (LightGBM, XGBoost) dans la CV (3 plis groupés par hôte), même échantillon
  final    : modèle réglé sur tout le train (jeu in_reg_set) : prédictions hors échantillon (A), leave-one-city-out (B),
             courbe d'apprentissage, résidus, intervalles de prédiction
Le test final n'est jamais lu. Usage : python -m src.regression [course|tune|final|all]
"""
import json
import pickle
import sys
import time
import warnings

import lightgbm as lgb
import numpy as np
import optuna
import pandas as pd
import xgboost as xgb
from sklearn.linear_model import RidgeCV
from sklearn.model_selection import GroupKFold, StratifiedGroupKFold, cross_val_predict

from src.data import ROOT
from src.evaluate import group_key, mae, mae_eur, r2, rmse
from src.features import build_preprocessor, feature_set
from src.models import CityMedianBaseline, fold_matrices, regression_zoo

warnings.filterwarnings("ignore")
optuna.logging.set_verbosity(optuna.logging.WARNING)
SEED = 42
N_COURSE = 80000
CACHE = ROOT / "models" / "_reg_cache"
CACHE.mkdir(parents=True, exist_ok=True)
FEATS_NAME = "core+quote+quality"


def load_train():
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    return d[~d["is_test"] & d["in_reg_set"]].reset_index(drop=True)


def smear_factor(y, p):
    """Duan smearing : facteur multiplicatif qui corrige le biais de la retransformation exp() (E[exp(e)] > 1)."""
    return float(np.mean(np.exp(np.asarray(y) - np.asarray(p))))


def full_metrics(y, p, smear=None):
    pe, ye = np.exp(p), np.exp(y)
    s = smear if smear is not None else smear_factor(y, p)
    pc = pe * s
    return dict(rmse=rmse(y, p), mae=mae(y, p), r2=r2(y, p), mae_eur=float(np.mean(np.abs(ye - pe))), mae_eur_corr=float(np.mean(np.abs(ye - pc))),
                mape=float(np.mean(np.abs(ye - pc) / ye)), smear=s)


# ===================================================================================== COURSE
def course():
    f = CACHE / "course.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    d = load_train()
    S = d.sample(N_COURSE, random_state=SEED).reset_index(drop=True)
    feats = feature_set(S.columns, FEATS_NAME)
    y = S["log_price_eur"].values
    cv = StratifiedGroupKFold(5, shuffle=True, random_state=SEED)
    folds = list(cv.split(S, S["city"], group_key(S)))
    zoo = regression_zoo()
    oof = {k: np.zeros(len(S)) for k in list(zoo) + ["Médiane de la ville"]}
    times = {k: 0.0 for k in oof}
    for i, (tr, va) in enumerate(folds):
        t0 = time.time()
        Z, Zv, Zs, Zvs = fold_matrices(S, feats, tr, va, y, build_preprocessor)
        b = CityMedianBaseline().fit(S.iloc[tr]["city"], y[tr])
        oof["Médiane de la ville"][va] = b.predict(S.iloc[va]["city"])
        for name, (m, scale) in zoo.items():
            t1 = time.time()
            m.fit(Zs if scale else Z, y[tr])
            oof[name][va] = m.predict(Zvs if scale else Zv)
            times[name] += time.time() - t1
        print(f"course pli {i} : {time.time() - t0:.0f}s", flush=True)
    rows = []
    for k, p in oof.items():
        m = full_metrics(y, p)
        # écart-type entre plis
        fr = [rmse(y[va], p[va]) for _, va in folds]
        rows.append(dict(modèle=k, rmse_std=float(np.std(fr)), temps_s=times[k] / 5, **m))
    # stacking : méta-modèle Ridge sur les prédictions hors échantillon des meilleurs modèles (évalué par CV groupée du méta-modèle)
    base = [k for k in ["Ridge", "Random Forest", "XGBoost", "LightGBM", "HistGradientBoosting"] if k in oof]
    M = np.column_stack([oof[k] for k in base])
    meta = RidgeCV(alphas=[0.01, 0.1, 1, 10])
    ps = cross_val_predict(meta, M, y, cv=GroupKFold(5), groups=group_key(S))
    oof["Stacking (Ridge sur 5 modèles)"] = ps
    rows.append(dict(modèle="Stacking (Ridge sur 5 modèles)", rmse_std=float(np.std([rmse(y[va], ps[va]) for _, va in folds])), temps_s=np.nan, **full_metrics(y, ps)))
    meta.fit(M, y)
    res = dict(table=pd.DataFrame(rows).sort_values("rmse"), oof=oof, y=y, S_index=S[["city", "id", "host_id"]], meta_weights=dict(zip(base, meta.coef_.round(3))), n=len(S))
    pickle.dump(res, open(f, "wb"))
    res["table"].to_csv(ROOT / "reports" / "regression_course.csv", index=False)
    return res


# ===================================================================================== RÉGLAGE
def tune(n_trials_lgb=40, n_trials_xgb=25):
    f = CACHE / "tune.json"
    if f.exists():
        return json.load(open(f))
    d = load_train()
    S = d.sample(N_COURSE, random_state=SEED).reset_index(drop=True)
    feats = feature_set(S.columns, FEATS_NAME)
    y = S["log_price_eur"].values
    folds = list(GroupKFold(3).split(S, groups=group_key(S)))
    mats = [fold_matrices(S, feats, tr, va, y, build_preprocessor)[:2] for tr, va in folds]
    out = {}

    def obj_lgb(t):
        p = dict(n_estimators=t.suggest_int("n_estimators", 200, 900), learning_rate=t.suggest_float("learning_rate", 0.02, 0.15, log=True),
                 num_leaves=t.suggest_int("num_leaves", 15, 255, log=True), min_child_samples=t.suggest_int("min_child_samples", 5, 100, log=True),
                 colsample_bytree=t.suggest_float("colsample_bytree", 0.3, 1.0), subsample=t.suggest_float("subsample", 0.5, 1.0), subsample_freq=1,
                 reg_lambda=t.suggest_float("reg_lambda", 1e-3, 30, log=True), random_state=SEED, n_jobs=16, verbose=-1)
        sc = []
        for (tr, va), (Z, Zv) in zip(folds, mats):
            m = lgb.LGBMRegressor(**p).fit(Z, y[tr])
            sc.append(rmse(y[va], m.predict(Zv)))
            t.report(float(np.mean(sc)), len(sc))
            if t.should_prune():
                raise optuna.TrialPruned()
        return float(np.mean(sc))

    def obj_xgb(t):
        p = dict(n_estimators=t.suggest_int("n_estimators", 200, 900), learning_rate=t.suggest_float("learning_rate", 0.02, 0.15, log=True),
                 max_depth=t.suggest_int("max_depth", 4, 11), min_child_weight=t.suggest_float("min_child_weight", 1, 30, log=True),
                 colsample_bytree=t.suggest_float("colsample_bytree", 0.3, 1.0), subsample=t.suggest_float("subsample", 0.5, 1.0),
                 reg_lambda=t.suggest_float("reg_lambda", 1e-3, 30, log=True), tree_method="hist", n_jobs=16, random_state=SEED, verbosity=0)
        sc = []
        for (tr, va), (Z, Zv) in zip(folds, mats):
            m = xgb.XGBRegressor(**p).fit(Z, y[tr])
            sc.append(rmse(y[va], m.predict(Zv)))
            t.report(float(np.mean(sc)), len(sc))
            if t.should_prune():
                raise optuna.TrialPruned()
        return float(np.mean(sc))

    for name, obj, n in [("LightGBM", obj_lgb, n_trials_lgb), ("XGBoost", obj_xgb, n_trials_xgb)]:
        st = optuna.create_study(direction="minimize", sampler=optuna.samplers.TPESampler(seed=SEED), pruner=optuna.pruners.MedianPruner(n_startup_trials=6))
        t0 = time.time()
        st.optimize(obj, n_trials=n)
        hist = [dict(trial=t.number, value=t.value, state=str(t.state)) for t in st.trials]
        out[name] = dict(best_params=st.best_params, best_rmse=st.best_value, seconds=time.time() - t0, history=hist,
                         importances={k: float(v) for k, v in optuna.importance.get_param_importances(st).items()})
        print(name, "meilleur RMSE CV", round(st.best_value, 4), st.best_params, flush=True)
    json.dump(out, open(f, "w"), indent=1)
    return out


# ===================================================================================== FINAL (OOF sur tout le train, LOCO, intervalles)
def final():
    f = CACHE / "final.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    tuned = tune()
    pl = dict(tuned["LightGBM"]["best_params"], subsample_freq=1, random_state=SEED, n_jobs=16, verbose=-1)
    d = load_train()
    feats = feature_set(d.columns, FEATS_NAME)
    y = d["log_price_eur"].values
    folds = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(d, d["city"], group_key(d)))
    oof, oof_lo, oof_hi, oof_default = (np.zeros(len(d)) for _ in range(4))
    fold_id = np.zeros(len(d), int)
    default_p = dict(n_estimators=300, learning_rate=0.08, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, random_state=SEED, n_jobs=16, verbose=-1)
    for i, (tr, va) in enumerate(folds):
        t0 = time.time()
        Z, Zv, _, _ = fold_matrices(d, feats, tr, va, y, build_preprocessor)
        oof[va] = lgb.LGBMRegressor(**pl).fit(Z, y[tr]).predict(Zv)
        oof_default[va] = lgb.LGBMRegressor(**default_p).fit(Z, y[tr]).predict(Zv)
        qp = dict(pl, objective="quantile")
        oof_lo[va] = lgb.LGBMRegressor(**qp, alpha=0.05).fit(Z, y[tr]).predict(Zv)
        oof_hi[va] = lgb.LGBMRegressor(**qp, alpha=0.95).fit(Z, y[tr]).predict(Zv)
        fold_id[va] = i
        print(f"final OOF pli {i} : {time.time() - t0:.0f}s", flush=True)
    # courbe d'apprentissage : taille d'entraînement vs erreur (hôtes disjoints), pli 0 comme validation
    tr0, va0 = folds[0]
    lc = []
    rng = np.random.RandomState(SEED)
    for frac in [0.05, 0.1, 0.2, 0.4, 0.7, 1.0]:
        sub = rng.choice(tr0, int(len(tr0) * frac), replace=False)
        Z, Zv, _, _ = fold_matrices(d, feats, sub, va0, y, build_preprocessor)
        m = lgb.LGBMRegressor(**pl).fit(Z, y[sub])
        lc.append(dict(n_train=len(sub), rmse_val=rmse(y[va0], m.predict(Zv)), rmse_train=rmse(y[sub], m.predict(Z))))
        print("courbe d'apprentissage", lc[-1], flush=True)
    # leave-one-city-out avec le modèle réglé (V1 et V3)
    loco = []
    oofB = {"V1": np.zeros(len(d)), "V3": np.zeros(len(d))}
    rng = np.random.RandomState(SEED)
    for c in sorted(d.city.unique()):
        tr, va = np.where(d.city != c)[0], np.where(d.city == c)[0]
        Z, Zv, _, _ = fold_matrices(d, feats, tr, va, y, build_preprocessor)
        drop3 = [k for k in Z.columns if k.startswith("city_") or k == "nbhd_key" or k in ("latitude", "longitude")]
        cal = rng.choice(len(va), min(200, len(va)), replace=False)
        for v, cols in [("V1", list(Z.columns)), ("V3", [k for k in Z.columns if k not in drop3])]:
            p = lgb.LGBMRegressor(**pl).fit(Z[cols], y[tr]).predict(Zv[cols])
            oofB[v][va] = p
            res = y[va] - p
            loco.append(dict(city=c, variante=v, rmse=rmse(y[va], p), bias=float(-res.mean()), rmse_recal200=float(np.sqrt(np.mean((res - res[cal].mean()) ** 2))), n=len(va)))
        print("LOCO", c, flush=True)
    out = dict(d_index=d[["city", "id", "host_id", "neighbourhood_cleansed", "room_type", "price_eur", "price_tier"]], y=y, oof=oof, oof_default=oof_default, oof_lo=oof_lo,
               oof_hi=oof_hi, fold_id=fold_id, learning_curve=lc, loco=pd.DataFrame(loco), oofB=oofB, params=pl)
    pickle.dump(out, open(f, "wb"))
    return out


if __name__ == "__main__":
    step = sys.argv[1] if len(sys.argv) > 1 else "all"
    if step in ("course", "all"):
        course()
    if step in ("tune", "all"):
        tune()
    if step in ("final", "all"):
        final()
```

## 12. Réglage des hyperparamètres

Un modèle a deux sortes de réglages : les **paramètres**, appris sur les données (les coupures d'un arbre), et les **hyperparamètres**, fixés avant (profondeur, taux d'apprentissage). On les cherche avec **Optuna** (recherche bayésienne TPE, avec arrêt précoce des essais médiocres) **à l'intérieur de la validation croisée** : 3 plis groupés par hôte sur l'échantillon de 80 000 annonces, **40 essais pour LightGBM et 25 pour XGBoost**. Le budget est annoncé : une recherche plus longue améliorerait un peu, jamais beaucoup.

````python
tune = json.load(open(MOD / "_reg_cache" / "tune.json"))
rows = []
for m in ["LightGBM", "XGBoost"]:
    h = pd.DataFrame(tune[m]["history"]); rows.append({"modèle": m, "essais": len(h), "essais abandonnés": int((h.state.str.contains("PRUNED")).sum()), "meilleur RMSE (CV 3 plis)": tune[m]["best_rmse"], "durée (s)": tune[m]["seconds"]})
display(pd.DataFrame(rows).style.format({"meilleur RMSE (CV 3 plis)": "{:.4f}", "durée (s)": "{:.0f}"}).hide(axis="index"))
bp = pd.DataFrame({m: tune[m]["best_params"] for m in tune}); display(bp.style.format(lambda v: f"{v:.4g}", na_rep="— (sans objet)"))
````

````python
fig, axes = plt.subplots(2, 2, figsize=(16, 8))
for j, m in enumerate(["LightGBM", "XGBoost"]):
    h = pd.DataFrame(tune[m]["history"]); done = h[h.state.str.contains("COMPLETE")]
    ax = axes[0, j]
    ax.scatter(done.trial, done.value, color=PAL["gris"], edgecolor="white", linewidth=0.8, s=45, zorder=3, label="essai mené à terme")
    pr = h[~h.state.str.contains("COMPLETE")]; ax.scatter(pr.trial, [done.value.max()] * len(pr), marker="x", color=PAL["rose"], label="essai abandonné (pruning)")
    ax.step(done.trial, done.value.cummin(), where="post", color=PAL["noir"], lw=2, label="meilleur jusqu'ici")
    ax.set(title=f"{m} : historique de la recherche", xlabel="numéro d'essai", ylabel="RMSE (CV)"); ax.legend(fontsize=8)
    imp = pd.Series(tune[m]["importances"]).sort_values()
    axes[1, j].barh(imp.index, imp.values, color=PAL["jaune"], edgecolor="white", linewidth=0.8); axes[1, j].set(title=f"{m} : quels réglages comptent ?", xlabel="importance (fANOVA)")
plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br>Chaque point gris est un essai mené jusqu'au bout ; la ligne noire en escalier est le meilleur score atteint jusque-là ; les croix roses sont les essais <b>abandonnés après quelques plis</b> car ils étaient déjà moins bons que la médiane (économie de calcul). Le meilleur LightGBM (RMSE 0,3699 en CV à 3 plis) a <b>873 arbres, un taux d'apprentissage de 0,04, 149 feuilles</b> et ne regarde que 53 % des variables par arbre : un modèle plus profond et plus lent que celui par défaut, régularisé par l'échantillonnage. Selon l'analyse d'importance des réglages (fANOVA), les écarts entre essais s'expliquent surtout par <code>subsample</code> (45 %), le nombre d'arbres (28 %) et le nombre de feuilles (14 %) pour LightGBM ; par <code>subsample</code>, <code>reg_lambda</code> et <code>colsample_bytree</code> pour XGBoost. <b>Prudence</b> : avec 25 à 40 essais, cette décomposition est indicative, et la surface est assez plate (voir l'écart entre le meilleur et le pire essai ci-dessous).</div>

````python
for m in ["LightGBM", "XGBoost"]:
    v = pd.DataFrame(tune[m]["history"]).query("state.str.contains('COMPLETE')").value
    print(f"{m} : {len(v)} essais menés à terme, RMSE de {v.min():.4f} (meilleur) à {v.max():.4f} (pire), médiane {v.median():.4f}")
````

````python
fin = pickle.load(open(MOD / "_reg_cache" / "final.pkl", "rb"))
yv, idx = fin["y"], fin["d_index"]
def rm(p): return float(np.sqrt(np.mean((yv - p) ** 2)))
av = pd.DataFrame({"LightGBM par défaut": [rm(fin["oof_default"]), np.mean(np.abs(yv - fin["oof_default"]))], "LightGBM réglé (Optuna)": [rm(fin["oof"]), np.mean(np.abs(yv - fin["oof"]))]}, index=["RMSE (log)", "MAE (log)"]).T
av["gain RMSE"] = (av.loc["LightGBM par défaut", "RMSE (log)"] - av["RMSE (log)"]) if False else av["RMSE (log)"].iloc[0] - av["RMSE (log)"]
display(av.style.format("{:.4f}")); print(f"Validation : {fmt(len(yv))} annonces du train, 5 plis A groupés par hôte (hors échantillon).")
````

Le gros du gain vient des variables (le contexte du devis du chapitre 4 fait passer le RMSE de 0,396 à 0,374) et du choix de famille (boosting contre linéaire : 0,375 contre 0,428). Le modèle réglé atteint un <b>R² de 0,81</b>. <b>La courbe d'apprentissage</b> : l'erreur d'entraînement monte (0,05 → 0,27) quand on ajoute des données, l'erreur de validation (hôtes jamais vus) descend de 0,406 avec 12 000 annonces à 0,364 avec 248 000. > **Remarque.** <b>Le champion.</b> Le <b>LightGBM réglé</b> est retenu pour sa simplicité, sa rapidité de service et sa compatibilité directe avec SHAP (chapitre 20) et avec le calibrage conforme des intervalles (chapitre 13). L'empilement de la course (RMSE 0,371 sur 80 000 annonces) n'a pas été reproduit sur tout le train réglé : on ne peut donc pas affirmer qu'il ferait mieux, ni l'écart exact ; son gain sur LightGBM par défaut était de l'ordre de 0,004 sur l'échantillon de la course.

## 13. Analyse des erreurs et intervalles de prédiction

Une métrique moyenne cache les endroits où le modèle est mauvais. On analyse les **prédictions hors échantillon** du modèle réglé (5 plis A, 309 412 annonces, chaque annonce prédite par un modèle qui ne l'a jamais vue ni son hôte). Puis on construit des **intervalles de prédiction** pour donner à l'hôte une fourchette plutôt qu'un point.

````python
smear = float(np.mean(np.exp(yv - fin["oof"])))
pe, ye = np.exp(fin["oof"]), np.exp(yv)
rel = np.abs(pe - ye) / ye
m = pd.Series({"RMSE (log)": rm(fin["oof"]), "MAE (log)": np.mean(np.abs(yv - fin["oof"])), "R²": 1 - ((yv - fin["oof"]) ** 2).sum() / ((yv - yv.mean()) ** 2).sum(),
               "MAE (EUR)": np.mean(np.abs(pe - ye)), "Erreur absolue médiane (EUR)": np.median(np.abs(pe - ye)), "MAPE médian": np.median(rel),
               "prix réel à ±10 % du prix prédit": np.mean(rel <= 0.10), "à ±20 %": np.mean(rel <= 0.20), "à ±30 %": np.mean(rel <= 0.30)})
display(m.to_frame("modèle réglé, hors échantillon").style.format({"RMSE (log)": "{:.3f}", "MAE (log)": "{:.3f}", "R²": "{:.3f}", "MAE (EUR)": "{:.1f}", "Erreur absolue médiane (EUR)": "{:.1f}", "MAPE médian": "{:.1%}",
                                                                   "prix réel à ±10 % du prix prédit": "{:.1%}", "à ±20 %": "{:.1%}", "à ±30 %": "{:.1%}"}))
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br>Traduit en décisions : le prix réel est à <b>±20 % du prix prédit pour 53 % des annonces</b>, à ±30 % pour 70 % et à ±40 % pour 82 %. L'erreur absolue <b>médiane</b> (27 €) est bien plus basse que la moyenne (66 €) : quelques logements très chers tirent la moyenne, on y revient plus bas. Erreur relative médiane : 19 % du prix.</div>

### 13.1 Les erreurs par ville, type de logement et tranche de prix

````python
r = idx.assign(res=yv - fin["oof"], pe=pe, ye=ye, ae=np.abs(pe - ye), ape=rel)
bycity = r.groupby("city").agg(n=("res", "size"), RMSE=("res", lambda s: np.sqrt((s ** 2).mean())), biais=("res", "mean"), MAE_EUR=("ae", "mean"), MAPE_médian=("ape", "median"), prix_médian=("ye", "median")).sort_values("MAE_EUR")
byroom = r.groupby("room_type").agg(n=("res", "size"), RMSE=("res", lambda s: np.sqrt((s ** 2).mean())), biais=("res", "mean"), MAE_EUR=("ae", "mean")).query("n > 300")
bytier = r.groupby("price_tier").agg(n=("res", "size"), RMSE=("res", lambda s: np.sqrt((s ** 2).mean())), biais=("res", "mean"), MAE_EUR=("ae", "mean"), MAPE_médian=("ape", "median"))
bytier.index = ["tranche basse", "tranche moyenne", "tranche haute"]
fig, ax = plt.subplots(1, 3, figsize=(19, 5.2))
ax[0].barh(bycity.index, bycity.MAE_EUR, color=PAL["jaune"], edgecolor="white", linewidth=0.8); ax[0].set(title="MAE en euros par ville", xlabel="EUR / nuit")
ax[1].barh(bycity.index, bycity.MAPE_médian * 100, color=PAL["rose"], edgecolor="white", linewidth=0.8); ax[1].set(title="Erreur relative médiane par ville", xlabel="% du prix")
ax[2].bar(bytier.index, bytier.MAPE_médian * 100, color=[PAL["vert"], PAL["jaune"], PAL["rose"]], edgecolor="white", linewidth=0.8); ax[2].set(title="Erreur relative médiane par tranche de prix (dans la ville)", ylabel="% du prix")
plt.tight_layout(); plt.show()
display(pd.concat({"par ville": bycity.round(3), "par type de logement": byroom.round(3), "par tranche de prix": bytier.round(3)}).fillna(""))
````

### 13.2 Résidus et sous-estimation des logements chers

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Régression vers la moyenne.</b> Les logements du <b>décile de prix réel le plus bas</b> sont en moyenne <b>surestimés de 21 %</b> et ceux du <b>plus haut sous-estimés de 47 %</b> ; la prédiction est « tirée » vers le centre. C'est un effet attendu de toute régression quand l'information est incomplète : un logement très cher l'est souvent pour des raisons que les données ne contiennent pas (vue, rénovation, photos, standing). <b>Hétéroscédasticité.</b> La dispersion des résidus est à peu près constante (écart-type de 0,26 à 0,32) sauf pour le <b>dernier décile de prix, où elle double (0,56)</b> ; le test de Breusch-Pagan rejette formellement l'homoscédasticité (p < 10⁻⁶) mais sa statistique est <b>modeste</b> (27 pour 309 000 observations) : une régression linéaire sur la prédiction capte mal un effet concentré dans la queue. Un intervalle de <b>largeur fixe</b> serait donc trop étroit pour les logements chers, d'où les intervalles <b>adaptatifs</b> ci-dessous.</div>

````python
nb = r[r.city == "paris"].groupby("neighbourhood_cleansed").agg(n=("res", "size"), biais=("res", "mean"), RMSE=("res", lambda s: np.sqrt((s ** 2).mean()))).query("n > 150").sort_values("biais")
fig, ax = plt.subplots(1, 2, figsize=(15, 5.5))
ax[0].barh(nb.index, np.expm1(nb.biais) * 100, color=[PAL["bleu"] if v > 0 else PAL["rose"] for v in nb.biais], edgecolor="white", linewidth=0.8); ax[0].axvline(0, color=PAL["noir"]); ax[0].set(title="Paris : écart moyen réel − prédit par arrondissement", xlabel="% du prix prédit")
ax[1].barh(nb.sort_values("RMSE").index, nb.sort_values("RMSE").RMSE, color=PAL["jaune"], edgecolor="white", linewidth=0.8); ax[1].set(title="Paris : RMSE par arrondissement", xlabel="RMSE (log)")
plt.tight_layout(); plt.show()
````

Un écart positif signifie que <b>les annonces du quartier sont en moyenne plus chères que ce que le modèle attend</b> : c'est la matière première de l'analyse « quartiers sous-évalués / sur-évalués » du chapitre 22.</div>

### 13.3 Des intervalles de prédiction pour donner une fourchette

Trois méthodes sont comparées pour un objectif de **90 % de couverture** :
1. **Régression quantile** (deux LightGBM qui prédisent les quantiles 5 % et 95 %).
2. **Conformal global** : on ajoute et retranche aux prédictions le quantile 90 % des résidus absolus (calibrés sur les autres plis, avec la correction de taille finie).
3. **Conformal adaptatif (Mondrian)** : même idée, mais **par tranche (décile) de prix prédit**, pour tenir compte de l'hétéroscédasticité.

Le calibrage de chaque pli utilise **uniquement les résidus des autres plis** (les hôtes du pli évalué ne servent jamais à fixer sa marge).

````python
alpha = 0.1; fid = fin["fold_id"]; res_abs = np.abs(yv - fin["oof"]); lp = fin["oof"]
glob_lo, glob_hi, mon_lo, mon_hi = (np.zeros(len(yv)) for _ in range(4))
for k in range(5):
    cal, ev = fid != k, fid == k
    r_cal = res_abs[cal]; q = np.quantile(r_cal, min(1, np.ceil((len(r_cal) + 1) * (1 - alpha)) / len(r_cal)))
    glob_lo[ev], glob_hi[ev] = lp[ev] - q, lp[ev] + q
    edges = np.quantile(lp[cal], np.linspace(0, 1, 11)); edges[0], edges[-1] = -np.inf, np.inf
    bc, be = np.clip(np.searchsorted(edges, lp[cal], side="right") - 1, 0, 9), np.clip(np.searchsorted(edges, lp[ev], side="right") - 1, 0, 9)
    qb = np.array([np.quantile(r_cal[bc == b], min(1, np.ceil(((bc == b).sum() + 1) * (1 - alpha)) / (bc == b).sum())) for b in range(10)])
    mon_lo[ev], mon_hi[ev] = lp[ev] - qb[be], lp[ev] + qb[be]
meth = {"Régression quantile (5 %–95 %)": (fin["oof_lo"], fin["oof_hi"]), "Conformal global": (glob_lo, glob_hi), "Conformal adaptatif (Mondrian)": (mon_lo, mon_hi)}
rows = []
for name, (lo_, hi_) in meth.items():
    inside = (yv >= lo_) & (yv <= hi_)
    rows.append({"méthode": name, "couverture globale": inside.mean(), "largeur moyenne (EUR)": np.mean(np.exp(hi_) - np.exp(lo_)), "largeur relative médiane": np.median((np.exp(hi_) - np.exp(lo_)) / np.exp(lp)),
                 "couverture min. par ville": pd.Series(inside).groupby(idx.city.values).mean().min(), "couverture min. par décile de prix prédit": pd.Series(inside).groupby(pd.qcut(lp, 10, labels=False)).mean().min()})
IV = pd.DataFrame(rows).set_index("méthode")
display(IV.style.format({"couverture globale": "{:.1%}", "largeur moyenne (EUR)": "{:.0f}", "largeur relative médiane": "{:.2f}", "couverture min. par ville": "{:.1%}", "couverture min. par décile de prix prédit": "{:.1%}"}))
fig, ax = plt.subplots(1, 2, figsize=(16, 5))
decp = pd.qcut(lp, 10, labels=False)
for (name, (lo_, hi_)), c in zip(meth.items(), [PAL["gris"], PAL["bleu"], PAL["rose"]]):
    ins = ((yv >= lo_) & (yv <= hi_)); ax[0].plot(range(1, 11), pd.Series(ins).groupby(decp).mean() * 100, "o-", color=c, mec="white", label=name)
ax[0].axhline(90, color=PAL["noir"], ls="--"); ax[0].set(title="Couverture selon le décile de prix PRÉDIT (cible 90 %)", xlabel="décile de prix prédit", ylabel="% de prix réels dans la fourchette"); ax[0].legend(fontsize=8)
cv = pd.DataFrame({m_: pd.Series((yv >= lo_) & (yv <= hi_)).groupby(idx.city.values).mean() * 100 for m_, (lo_, hi_) in meth.items()})
cv.plot.barh(ax=ax[1], color=[PAL["gris"], PAL["bleu"], PAL["rose"]], edgecolor="white", linewidth=0.8, width=0.8); ax[1].axvline(90, color=PAL["noir"], ls="--"); ax[1].set(title="Couverture par ville", xlabel="%", xlim=(60, 100)); ax[1].legend(fontsize=7, loc="lower left")
plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Les trois méthodes ne se valent pas.</b> La <b>régression quantile</b> (5 %-95 %) ne couvre que <b>82 %</b> des prix réels au lieu des 90 % visés : un modèle de quantiles entraîné sur les mêmes annonces que celles qu'il prédit est trop optimiste sur des hôtes nouveaux. Le <b>conformal global</b> atteint 90 % <b>en moyenne</b> mais ne tient pas pour les logements chers : la couverture tombe à <b>83,5 % dans le décile de prix prédit le plus élevé</b> (90 à 92 % ailleurs), c'est l'hétéroscédasticité du graphique précédent. Le <b>conformal adaptatif (Mondrian)</b> couvre <b>90 % dans chacun des dix déciles</b> (entre 89,9 % et 90,1 %), pour une largeur moyenne de 243 € contre 229 € (+6 %). <b>C'est la méthode retenue pour l'outil de prix</b> (chapitre 21). <b>Limite</b> : la couverture par ville n'est pas uniforme (minimum 86 % pour les deux méthodes conformes) : certaines villes sont moins prévisibles que d'autres.</div>

> **Remarque.** <b>Ce que signifie « 90 % »</b> : sur des hôtes comparables à ceux déjà vus, <b>9 fois sur 10</b> le prix réel du devis tombera dans la fourchette annoncée. Cette garantie suppose que l'hôte nouveau ressemble aux hôtes de la calibration ; pour une <b>nouvelle ville</b> elle ne tient plus sans recalibrage (chapitre 22).

## 14. Classification : Superhost et tranche de prix

Deux tâches de **classification**. **Tâche 1** : l'hôte est-il **Superhost** ? Classes **déséquilibrées** (30 % de positifs). **Tâche 2** : l'annonce est-elle dans la tranche de prix **basse, moyenne ou haute** de sa ville (3 classes équilibrées, terciles par ville) ? Même protocole que pour la régression : 5 plis groupés par hôte sur un **échantillon de 60 000 annonces du train**, prétraitement ajusté dans chaque pli, test mis de côté.

````python
from sklearn.metrics import roc_curve, precision_recall_curve, roc_auc_score, average_precision_score, confusion_matrix, f1_score, brier_score_loss
from sklearn.calibration import calibration_curve
clfS = pickle.load(open(MOD / "_clf_cache" / "superhost.pkl", "rb")); clfT = pickle.load(open(MOD / "_clf_cache" / "tier.pkl", "rb"))
yS = clfS["y"]; print(f"Superhost : {fmt(len(yS))} annonces, {yS.mean():.1%} de positifs | Tranche de prix : {fmt(len(clfT['y']))} annonces, classes {np.bincount(clfT['y']) / len(clfT['y'])}")
````

### 14.1 Superhost : la course des modèles (variables = caractéristiques seules, sans notes ni avis)

````python
tab = clfS["table"]; A = tab[tab.jeu.str.startswith("A.")].set_index("modèle")
show = A[["auc", "auc_std", "pr_auc", "f1", "f1_macro", "accuracy", "precision", "recall", "temps_s"]].rename(columns={"auc": "ROC-AUC", "auc_std": "± plis", "pr_auc": "PR-AUC", "f1": "F1 (Superhost)", "f1_macro": "F1 macro", "accuracy": "accuracy", "precision": "précision", "recall": "rappel", "temps_s": "temps/pli (s)"})
display(show.sort_values("ROC-AUC", ascending=False).style.format("{:.3f}").background_gradient(subset=["ROC-AUC"], cmap="RdYlGn"))
print("Référence : un classifieur aléatoire a une ROC-AUC de 0,50 et une PR-AUC égale au taux de positifs (%.2f)." % yS.mean())
oofA = clfS["oof"][[k for k in clfS["oof"] if k.startswith("A.")][0]]
fig, ax = plt.subplots(1, 2, figsize=(15, 5.5))
cols = {"Régression logistique": PAL["bleu"], "kNN (k=25)": "#8E7DBE", "SVM (RBF, 12 000 ex.)": "#E8833A", "Arbre de décision": PAL["jaune"], "Random Forest": PAL["vert"], "HistGradientBoosting": PAL["gris"], "LightGBM": PAL["rose"]}
for m, p in oofA.items():
    f_, t_, _ = roc_curve(yS, p); ax[0].plot(f_, t_, color=cols[m], lw=2, label=f"{m} ({roc_auc_score(yS, p):.3f})")
    pr_, rc_, _ = precision_recall_curve(yS, p); ax[1].plot(rc_, pr_, color=cols[m], lw=2)
ax[0].plot([0, 1], [0, 1], color=PAL["noir"], ls="--"); ax[0].set(title="Courbes ROC (hors échantillon)", xlabel="taux de faux positifs", ylabel="taux de vrais positifs"); ax[0].legend(fontsize=8, loc="lower right")
ax[1].axhline(yS.mean(), color=PAL["noir"], ls="--"); ax[1].set(title="Courbes précision-rappel", xlabel="rappel", ylabel="précision"); plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Le boosting l'emporte encore</b> (LightGBM : ROC-AUC 0,817, HistGradientBoosting 0,813), devant la forêt aléatoire (0,800), la régression logistique (0,778) et le SVM (0,772) ; le kNN est le moins bon (0,749). Le <b>SVM</b> coûte très cher pour un résultat moyen (<b>179 s par pli</b> sur seulement 12 000 exemples d'entraînement, contre 4 s pour LightGBM sur 48 000). Attention à l'<b>accuracy</b> : avec 30 % de positifs, un modèle qui répondrait toujours « non Superhost » aurait 70 % d'accuracy ; les 77 % obtenus ne valent donc que 7 points de mieux que le silence. Les métriques utiles ici sont la <b>PR-AUC</b> (0,64 contre 0,30 pour le hasard) et le <b>F1</b> de la classe rare (0,56).</div>

### 14.2 Le test de la fuite : avec et sans les notes et l'activité des avis

````python
t2 = tab[tab.modèle.isin(["Régression logistique", "LightGBM"])].copy(); t2["jeu"] = t2.jeu.str[:2].map({"A.": "A · caractéristiques seules", "B.": "B · + notes d'avis", "C.": "C · + notes et activité des avis"})
pv = t2.pivot(index="jeu", columns="modèle", values="auc")
fig, ax = plt.subplots(figsize=(10, 4.8)); w = 0.36; x = np.arange(len(pv))
ax.bar(x - w / 2, pv["Régression logistique"], w, color=PAL["bleu"], edgecolor="white", linewidth=0.8, label="régression logistique"); ax.bar(x + w / 2, pv["LightGBM"], w, color=PAL["rose"], edgecolor="white", linewidth=0.8, label="LightGBM")
for i, (a, b) in enumerate(zip(pv["Régression logistique"], pv["LightGBM"])): ax.text(i - w / 2, a + 0.006, f"{a:.3f}", ha="center", fontsize=9); ax.text(i + w / 2, b + 0.006, f"{b:.3f}", ha="center", fontsize=9)
ax.set_xticks(x, pv.index, rotation=0, fontsize=9); ax.set(ylim=(0.65, 0.95), title="ROC-AUC du statut Superhost selon les variables autorisées", ylabel="ROC-AUC"); ax.legend(); plt.tight_layout(); plt.show()
pv.round(3)
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Les notes d'avis font gagner 6 points d'AUC</b> (0,817 → 0,875) <b>et l'activité en gagne encore 3</b> (→ 0,902). Ces gains sont <b>trop beaux</b> : le statut Superhost est attribué <i>d'après</i> la note, la réactivité et le nombre de séjours. Prédire le statut avec ces variables, c'est en partie <b>lire les critères d'attribution</b> ; ce n'est pas expliquer pourquoi un hôte devient Superhost. <b>Deux usages, deux jeux de variables.</b> (A) <i>Comprendre ce qui, dans le logement et l'hôte, accompagne le statut</i>, sans circularité : AUC 0,82, utile pour la plateforme pour repérer des hôtes à fort potentiel avant qu'ils n'aient des avis. (B) <i>Reconstituer le statut à partir de tout ce qui est connu</i>, y compris les notes : AUC 0,88, mais une grande partie du signal vient des critères eux-mêmes. Le jeu C est une fuite pour tout usage prédictif (volume d'avis sur 12 mois, qui est un critère d'attribution) et n'est <b>pas retenu</b>.</div>

### 14.3 Le déséquilibre : poids de classe, SMOTE, seuil de décision

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Les trois traitements du déséquilibre ne changent presque pas l'AUC</b> (0,875 à 0,876) : ils ne rendent pas le modèle meilleur à <i>classer</i> les hôtes, ils déplacent seulement le <b>compromis précision / rappel</b>. <b>SMOTE</b> (génération de faux exemples dans l'espace des variables, uniquement sur le train du pli) n'apporte rien : les variables mêlent comptages, binaires et encodages, et interpoler entre elles produit des profils irréalistes. <b>Le poids de classe</b> et <b>le seuil optimisé</b> font la même chose : on passe d'un rappel de 62 % (précision 72 %) à un rappel d'environ 76-78 % (précision 63 %). Le bon choix <b>dépend du coût des erreurs</b> : si la plateforme veut <i>ne rater aucun candidat Superhost</i> (campagne d'accompagnement peu coûteuse), on prend le seuil bas ; si elle veut des étiquettes fiables, on garde 0,5.</div>

#### Code source du chapitre 14

<details>
<summary><code>src/classification.py</code> (210 lignes) : Classification : Superhost (avec et sans variables suspectes) et tranche de prix</summary>

```python
"""Phase 7 : classification.
  Tâche 1 : is_superhost (déséquilibré, ~30 % de positifs) ; test avec / sans les variables suspectes de fuite (notes, activité des avis)
  Tâche 2 : price_tier (bas / moyen / haut, terciles de prix PAR ville)
Validation : CV A (5 plis groupés par hôte) sur un échantillon du train ; schéma B (leave-one-city-out) pour le meilleur modèle.
Le test final n'est jamais lu. Usage : python -m src.classification [test]   (« test » = version rapide)
"""
import pickle
import sys
import time
import warnings

import lightgbm as lgb
import numpy as np
import pandas as pd
from imblearn.over_sampling import SMOTE
from sklearn.ensemble import HistGradientBoostingClassifier, RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (accuracy_score, average_precision_score, balanced_accuracy_score, confusion_matrix, f1_score, precision_score, recall_score,
                             roc_auc_score)
from sklearn.model_selection import GroupShuffleSplit, StratifiedGroupKFold
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier

from src.data import ROOT
from src.evaluate import group_key
from src.features import build_preprocessor, feature_groups, feature_set
from src.models import fold_matrices

warnings.filterwarnings("ignore")
SEED = 42
CACHE = ROOT / "models" / "_clf_cache"
CACHE.mkdir(parents=True, exist_ok=True)
FAST = len(sys.argv) > 1 and sys.argv[1] == "test"
N = 15000 if FAST else 60000


def sets_superhost(columns):
    g = feature_groups(columns)
    core = g["property"] + g["amenities"] + g["location"] + g["host"] + g["text"]
    return {"A. sans avis (caractéristiques seules)": core,
            "B. + notes d'avis": core + g["quality"],
            "C. + notes et activité des avis (suspect de fuite)": core + g["quality"] + [c for c in g["activity"] if c not in ("is_stale",)]}


def zoo(multiclass=False, n_jobs=16):
    return {
        "Régression logistique": (LogisticRegression(max_iter=500, C=0.5), True),
        "kNN (k=25)": (KNeighborsClassifier(n_neighbors=25, n_jobs=n_jobs), True),
        "SVM (RBF, 12 000 ex.)": (SVC(C=1.0, probability=True, random_state=SEED), True),
        "Arbre de décision": (DecisionTreeClassifier(max_depth=10, min_samples_leaf=30, random_state=SEED), False),
        "Random Forest": (RandomForestClassifier(n_estimators=250, min_samples_leaf=3, max_features="sqrt", n_jobs=n_jobs, random_state=SEED), False),
        "HistGradientBoosting": (HistGradientBoostingClassifier(max_iter=300, learning_rate=0.08, random_state=SEED), False),
        "LightGBM": (lgb.LGBMClassifier(n_estimators=300, learning_rate=0.06, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, n_jobs=n_jobs,
                                        random_state=SEED, verbose=-1), False),
    }


def best_f1_threshold(y, p):
    ts = np.linspace(0.05, 0.95, 91)
    f = [f1_score(y, p >= t) for t in ts]
    return float(ts[int(np.argmax(f))])


def bin_metrics(y, p, thr=0.5):
    yh = p >= thr
    return dict(auc=roc_auc_score(y, p), pr_auc=average_precision_score(y, p), accuracy=accuracy_score(y, yh), precision=precision_score(y, yh), recall=recall_score(y, yh),
                f1=f1_score(y, yh), f1_macro=f1_score(y, yh, average="macro"), thr=thr)


def multi_metrics(y, P):
    yh = P.argmax(1)
    return dict(accuracy=accuracy_score(y, yh), bal_acc=balanced_accuracy_score(y, yh), f1_macro=f1_score(y, yh, average="macro"), auc_ovr=roc_auc_score(y, P, multi_class="ovr"))


def load(task):
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    d = d[~d["is_test"]]
    if task == "superhost":
        d = d.dropna(subset=["is_superhost"])
    else:
        d = d[d["in_reg_set"]].dropna(subset=["price_tier"])
    return d.sample(min(N, len(d)), random_state=SEED).reset_index(drop=True)


# ----------------------------------------------------------------------------------------- Tâche 1
def superhost():
    f = CACHE / ("superhost_fast.pkl" if FAST else "superhost.pkl")
    if f.exists():
        return pickle.load(open(f, "rb"))
    S = load("superhost")
    y = S["is_superhost"].astype(int).values
    folds = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(S, y, group_key(S)))
    rows, oofs = [], {}
    # (1) comparaison des modèles sur le jeu A
    # (2) test avec / sans variables suspectes avec LightGBM et régression logistique
    feat_sets = sets_superhost(S.columns)
    for set_name, feats in feat_sets.items():
        models = zoo() if set_name.startswith("A.") else {k: v for k, v in zoo().items() if k in ("Régression logistique", "LightGBM")}
        oof = {k: np.zeros(len(S)) for k in models}
        tm = {k: 0.0 for k in models}
        for i, (tr, va) in enumerate(folds):
            Z, Zv, Zs, Zvs = fold_matrices(S, feats, tr, va, y, lambda ft, X, scale: build_preprocessor(ft, X, scale=scale, target_type="binary"))
            for name, (m, scale) in models.items():
                t0 = time.time()
                Xtr, Xva, ytr = (Zs, Zvs, y[tr]) if scale else (Z, Zv, y[tr])
                if name.startswith("SVM"):
                    sub = np.random.RandomState(SEED).choice(len(Xtr), min(12000 if not FAST else 4000, len(Xtr)), replace=False)
                    m.fit(Xtr.iloc[sub], ytr[sub])
                else:
                    m.fit(Xtr, ytr)
                oof[name][va] = m.predict_proba(Xva)[:, 1]
                tm[name] += time.time() - t0
            print(f"superhost {set_name[:2]} pli {i}", flush=True)
        for k, p in oof.items():
            m = bin_metrics(y, p, 0.5)
            m.update(modèle=k, jeu=set_name, temps_s=tm[k] / 5, auc_std=float(np.std([roc_auc_score(y[va], p[va]) for _, va in folds])))
            rows.append(m)
        oofs[set_name] = oof
    table = pd.DataFrame(rows)

    # (3) déséquilibre : class_weight, SMOTE (dans la CV, train du pli seulement), seuil optimisé (choisi sur un split interne du train du pli)
    feats = feat_sets["B. + notes d'avis"]
    variants = {"LightGBM standard": None, "LightGBM class_weight='balanced'": "cw", "LightGBM + SMOTE": "smote", "LightGBM seuil optimisé (F1)": "thr"}
    oofv = {k: np.zeros(len(S)) for k in variants}
    thr_used = {k: [] for k in variants}
    for i, (tr, va) in enumerate(folds):
        Z, Zv, _, _ = fold_matrices(S, feats, tr, va, y, lambda ft, X, scale: build_preprocessor(ft, X, scale=scale, target_type="binary"))
        P = dict(n_estimators=300, learning_rate=0.06, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, n_jobs=16, random_state=SEED, verbose=-1)
        for k, mode in variants.items():
            if mode is None:
                oofv[k][va] = lgb.LGBMClassifier(**P).fit(Z, y[tr]).predict_proba(Zv)[:, 1]
            elif mode == "cw":
                oofv[k][va] = lgb.LGBMClassifier(**P, class_weight="balanced").fit(Z, y[tr]).predict_proba(Zv)[:, 1]
            elif mode == "smote":
                Xs, ys = SMOTE(random_state=SEED).fit_resample(Z, y[tr])
                oofv[k][va] = lgb.LGBMClassifier(**P).fit(Xs, ys).predict_proba(Zv)[:, 1]
            else:
                a, b = next(GroupShuffleSplit(1, test_size=0.2, random_state=SEED).split(Z, groups=group_key(S).values[tr]))
                mm = lgb.LGBMClassifier(**P).fit(Z.iloc[a], y[tr][a])
                t = best_f1_threshold(y[tr][b], mm.predict_proba(Z.iloc[b])[:, 1])
                thr_used[k].append(t)
                oofv[k][va] = lgb.LGBMClassifier(**P).fit(Z, y[tr]).predict_proba(Zv)[:, 1]
        print(f"superhost déséquilibre pli {i}", flush=True)
    imb = []
    for k, p in oofv.items():
        thr = float(np.mean(thr_used[k])) if thr_used[k] else 0.5
        imb.append(dict(variante=k, **bin_metrics(y, p, thr)))
    imb = pd.DataFrame(imb)

    # (4) schéma B (leave-one-city-out) avec LightGBM, jeu B
    lo = []
    cities = S["city"].values
    for c in sorted(S.city.unique()):
        tr, va = np.where(cities != c)[0], np.where(cities == c)[0]
        Z, Zv, _, _ = fold_matrices(S, feats, tr, va, y, lambda ft, X, scale: build_preprocessor(ft, X, scale=scale, target_type="binary"))
        p = lgb.LGBMClassifier(**P).fit(Z, y[tr]).predict_proba(Zv)[:, 1]
        lo.append(dict(city=c, auc=roc_auc_score(y[va], p), pr_auc=average_precision_score(y[va], p), base_rate=y[va].mean(), n=len(va)))
    loco = pd.DataFrame(lo)
    res = dict(table=table, imbalance=imb, oof=oofs, oofv=oofv, y=y, loco=loco, S_index=S[["city", "id", "host_id"]], n=len(S))
    pickle.dump(res, open(f, "wb"))
    return res


# ----------------------------------------------------------------------------------------- Tâche 2
def price_tier():
    f = CACHE / ("tier_fast.pkl" if FAST else "tier.pkl")
    if f.exists():
        return pickle.load(open(f, "rb"))
    S = load("tier")
    y = S["price_tier"].astype(int).values
    feats = feature_set(S.columns, "core+quote+quality")
    folds = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(S, y, group_key(S)))
    models = zoo(multiclass=True)
    oof = {k: np.zeros((len(S), 3)) for k in models}
    tm = {k: 0.0 for k in models}
    for i, (tr, va) in enumerate(folds):
        Z, Zv, Zs, Zvs = fold_matrices(S, feats, tr, va, y, lambda ft, X, scale: build_preprocessor(ft, X, scale=scale, target_type="multiclass"))
        for name, (m, scale) in models.items():
            t0 = time.time()
            Xtr, Xva = (Zs, Zvs) if scale else (Z, Zv)
            if name.startswith("SVM"):
                sub = np.random.RandomState(SEED).choice(len(Xtr), min(12000 if not FAST else 4000, len(Xtr)), replace=False)
                m.fit(Xtr.iloc[sub], y[tr][sub])
            else:
                m.fit(Xtr, y[tr])
            oof[name][va] = m.predict_proba(Xva)
            tm[name] += time.time() - t0
        print(f"tier pli {i}", flush=True)
    table = pd.DataFrame([dict(modèle=k, temps_s=tm[k] / 5, **multi_metrics(y, P)) for k, P in oof.items()])
    # schéma B pour LightGBM : tranches relatives à la ville -> généralisation à une ville inconnue
    lo = []
    cities = S["city"].values
    for c in sorted(S.city.unique()):
        tr, va = np.where(cities != c)[0], np.where(cities == c)[0]
        Z, Zv, _, _ = fold_matrices(S, feats, tr, va, y, lambda ft, X, scale: build_preprocessor(ft, X, scale=scale, target_type="multiclass"))
        m = lgb.LGBMClassifier(n_estimators=300, learning_rate=0.06, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, n_jobs=16, random_state=SEED, verbose=-1)
        P = m.fit(Z, y[tr]).predict_proba(Zv)
        lo.append(dict(city=c, n=len(va), **multi_metrics(y[va], P)))
    res = dict(table=table, oof=oof, y=y, loco=pd.DataFrame(lo), S_index=S[["city", "id", "host_id"]], n=len(S))
    pickle.dump(res, open(f, "wb"))
    return res


if __name__ == "__main__":
    a = superhost()
    print(a["table"].round(4).to_string())
    print(a["imbalance"].round(4).to_string())
    b = price_tier()
    print(b["table"].round(4).to_string())

## 15. Clustering et détection d'anomalies

Sans cible à prédire, on laisse les données se regrouper. Pour que les groupes ne reproduisent pas simplement les villes, on utilise des variables **relatives à la ville** : prix **relatif** à la médiane de la ville, **rang** de distance au centre dans la ville. Échantillon : **60 000 annonces du train avec prix** (10 000 pour le hiérarchique, 20 000 pour UMAP), 12 variables standardisées.

````python
clu = json.load(open(REP / "clust_summary.json")); ksel = pd.read_csv(REP / "clust_k_selection.csv"); gbic = pd.read_csv(REP / "clust_gmm_bic.csv")
print({k: (round(v, 3) if isinstance(v, float) else v) for k, v in clu.items()})
````

### 15.1 Combien de groupes ? Coude, silhouette, Davies-Bouldin

````python
fig, ax = plt.subplots(1, 3, figsize=(17, 4.6))
ax[0].plot(ksel.k, ksel.inertie, "o-", color=PAL["bleu"], mec="white"); ax[0].set(title="Coude : inertie intra-cluster", xlabel="k", ylabel="inertie")
ax[1].plot(ksel.k, ksel.silhouette, "o-", color=PAL["vert"], mec="white"); ax[1].axhline(0.25, color=PAL["gris"], ls="--"); ax[1].text(7.2, 0.255, "seuil usuel d'une structure nette (0,25)", fontsize=8); ax[1].set(title="Silhouette (plus haut = mieux)", xlabel="k")
ax[2].plot(ksel.k, ksel.davies_bouldin, "o-", color=PAL["rose"], mec="white"); ax[2].set(title="Davies-Bouldin (plus bas = mieux)", xlabel="k")
for a in ax: a.axvline(clu["k_best"], color=PAL["noir"], ls=":")
plt.tight_layout(); plt.show()
ksel.round(3).set_index("k").T
````

Les plus extrêmes sont des villas de 16 personnes et 9 chambres à plus de 3 000 € la nuit ; plusieurs appartiennent à <b>un même hôte barcelonais qui gère 519 annonces</b>. Atypique ne veut pas dire suspect : c'est une <b>liste d'annonces à examiner</b>, pas une accusation. Le croisement avec les écarts de prix du modèle (chapitre 22) permettra de repérer les annonces <i>à la fois</i> atypiques et sur-évaluées.</div>

> **Remarque.** <b>À retenir.</b> Les annonces forment un continuum plus qu'un archipel (silhouette de 0,12). Les cinq profils retenus sont des <b>outils de lecture</b>, utiles pour parler de types d'offre, mais ils ne doivent pas être sur-interprétés. En revanche, la PCA et l'UMAP confirment l'essentiel : <b>la taille et le standing</b> d'un côté, <b>la maturité de l'annonce</b> de l'autre, et un clivage net entre chambres et logements entiers.

```python
"""Phase 8 : apprentissage non supervisé (clustering, réduction de dimension, anomalies).
Variables RELATIVES à la ville (prix relatif, rang de distance au centre) pour que les clusters ne reproduisent pas simplement les villes.
Échantillon : 60 000 annonces avec prix (train, jeu in_reg_set) pour k-means / PCA / anomalies ; 10 000 pour le hiérarchique ; 20 000 pour UMAP.
Usage : python -m src.clustering
"""
import warnings

import matplotlib

matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from scipy.cluster.hierarchy import dendrogram, fcluster, linkage
from sklearn.cluster import DBSCAN, KMeans
from sklearn.decomposition import PCA
from sklearn.ensemble import IsolationForest
from sklearn.metrics import adjusted_rand_score, davies_bouldin_score, silhouette_score
from sklearn.mixture import GaussianMixture
from sklearn.preprocessing import StandardScaler

from src.data import ROOT

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", context="notebook")
FIG = ROOT / "reports" / "figures"
SEED = 42
FEATS = {"log_price_rel": "prix relatif (log, vs médiane ville)", "accommodates": "capacité", "bedrooms": "chambres", "bath_count": "salles de bain",
         "n_amenities": "nb équipements", "dist_rank": "éloignement du centre (rang dans la ville)", "log_host_listings": "log annonces de l'hôte",
         "is_entire": "logement entier", "review_scores_rating": "note globale", "log_n_reviews": "log nb d'avis", "availability_365": "jours disponibles/an",
         "host_tenure_years": "ancienneté hôte"}


def load():
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    d = d[~d["is_test"] & d["in_reg_set"]].copy()
    d["log_price_rel"] = d["log_price_eur"] - d.groupby("city")["log_price_eur"].transform("median")
    d["dist_rank"] = d.groupby("city")["dist_center_km"].rank(pct=True)
    d["is_entire"] = (d["room_type"] == "Entire home/apt").astype(float)
    return d


LABELS = {"log_price_rel": ("prix élevés", "prix bas"), "accommodates": ("grande capacité", "petits logements"), "n_amenities": ("bien équipés", "peu équipés"),
          "dist_rank": ("périphérie", "centre"), "log_host_listings": ("hôtes multi-annonces", "hôtes particuliers"), "is_entire": ("logements entiers", "chambres chez l'habitant"),
          "review_scores_rating": ("très bien notés", "mal notés"), "log_n_reviews": ("beaucoup d'avis", "peu d'avis"), "availability_365": ("très disponibles", "très demandés"),
          "host_tenure_years": ("hôtes anciens", "hôtes récents")}


def name_cluster(z: pd.Series, entire_share: float) -> str:
    """Nom parlant : les 3 caractéristiques les plus écartées de la médiane globale (|z| décroissant), avec leur sens."""
    z = z[list(LABELS)].copy()
    z["is_entire"] = (entire_share - 0.8) * 4  # part de logements entiers (médiane 1 : on mesure l'écart à ~80 %)
    top = z.abs().sort_values(ascending=False).index[:3]
    return " · ".join(LABELS[c][0] if z[c] > 0 else LABELS[c][1] for c in top)


def run():
    d = load()
    S = d.sample(60000, random_state=SEED).reset_index(drop=True)
    cols = list(FEATS)
    X = S[cols].copy()
    X = X.fillna(X.median())
    sc = StandardScaler().fit(X)
    Z = sc.transform(X)
    out = {}

    # ---------------- choix de k (k-means)
    ks = range(2, 13)
    inertia, sil, dbi = [], [], []
    rs = np.random.RandomState(SEED).choice(len(Z), 10000, replace=False)
    for k in ks:
        km = KMeans(k, n_init=5, random_state=SEED).fit(Z)
        inertia.append(km.inertia_)
        sil.append(silhouette_score(Z[rs], km.labels_[rs]))
        dbi.append(davies_bouldin_score(Z, km.labels_))
    sel = pd.DataFrame(dict(k=list(ks), inertie=inertia, silhouette=sil, davies_bouldin=dbi))
    sel.to_csv(ROOT / "reports" / "clust_k_selection.csv", index=False)
    fig, ax = plt.subplots(1, 3, figsize=(16, 4.5))
    ax[0].plot(sel.k, sel.inertie, "o-"); ax[0].set(title="Coude (inertie)", xlabel="k", ylabel="Inertie intra-cluster")
    ax[1].plot(sel.k, sel.silhouette, "o-", color="#54a24b"); ax[1].set(title="Silhouette (plus haut = mieux)", xlabel="k", ylabel="Silhouette moyenne")
    ax[2].plot(sel.k, sel.davies_bouldin, "o-", color="#e45756"); ax[2].set(title="Davies-Bouldin (plus bas = mieux)", xlabel="k", ylabel="Indice")
    plt.tight_layout(); plt.savefig(FIG / "clu_01_choice_of_k.png", dpi=130); plt.close()
    # choix : k maximisant la silhouette parmi k >= 4 (au moins 4 profils pour une lecture métier)
    k_best = int(sel[sel.k >= 4].sort_values("silhouette", ascending=False).k.iloc[0])
    km = KMeans(k_best, n_init=10, random_state=SEED).fit(Z)
    S["cluster"] = km.labels_

    # ---------------- profils et noms
    med = S.groupby("cluster")[cols].median()
    glob_std = X.std()
    zprof = (med - X.median()) / glob_std
    ent = S.groupby("cluster")["is_entire"].mean()
    names = {c: name_cluster(zprof.loc[c], ent[c]) for c in zprof.index}
    # unicité des noms
    seen = {}
    for c, n in names.items():
        seen[n] = seen.get(n, 0) + 1
        if seen[n] > 1:
            names[c] = f"{n} ({seen[n]})"
    S["cluster_name"] = S["cluster"].map(names)
    prof = S.groupby("cluster_name").agg(annonces=("id", "size"), part_pct=("id", lambda s: len(s) / len(S) * 100), prix_median_eur=("price_eur", "median"),
                                         prix_relatif=("log_price_rel", lambda s: float(np.exp(s.median()))), capacité=("accommodates", "median"), chambres=("bedrooms", "median"),
                                         logements_entiers_pct=("is_entire", lambda s: s.mean() * 100), dist_centre_rang=("dist_rank", "median"),
                                         annonces_hôte=("calculated_host_listings_count", "median"), note=("review_scores_rating", "median"),
                                         jours_dispo=("availability_365", "median"), superhosts_pct=("is_superhost", lambda s: s.mean() * 100)).round(2)
    prof.sort_values("prix_median_eur", ascending=False).to_csv(ROOT / "reports" / "clust_profiles.csv")
    comp = pd.crosstab(S["cluster_name"], S["city"], normalize="columns") * 100
    comp.round(1).to_csv(ROOT / "reports" / "clust_city_composition.csv")
    fig, ax = plt.subplots(figsize=(14, 6))
    sns.heatmap(zprof.rename(index=names).rename(columns=FEATS), annot=True, fmt=".1f", cmap="RdBu_r", center=0, ax=ax, cbar_kws={"label": "écart à la médiane globale (en écarts-types)"})
    ax.set(title=f"Profils des {k_best} clusters d'annonces (k-means)", xlabel="Variable", ylabel="Cluster")
    plt.xticks(rotation=40, ha="right")
    plt.tight_layout(); plt.savefig(FIG / "clu_02_profiles.png", dpi=130); plt.close()
    fig, ax = plt.subplots(figsize=(14, 6))
    comp.T.plot(kind="bar", stacked=True, ax=ax, colormap="tab20")
    ax.set(title="Composition des villes par cluster (chaque barre = 100 % des annonces de la ville)", xlabel="Ville", ylabel="% des annonces")
    ax.legend(fontsize=7, bbox_to_anchor=(1.01, 1))
    plt.tight_layout(); plt.savefig(FIG / "clu_03_city_composition.png", dpi=130); plt.close()
    out["k_best"] = k_best

    # ---------------- hiérarchique (Ward) sur 10 000 annonces + accord avec k-means ; GMM (BIC)
    h = np.random.RandomState(SEED).choice(len(Z), 10000, replace=False)
    L = linkage(Z[h], method="ward")
    hl = fcluster(L, k_best, criterion="maxclust") - 1
    out["ari_ward_kmeans"] = float(adjusted_rand_score(S.loc[h, "cluster"], hl))
    out["sil_ward"] = float(silhouette_score(Z[h], hl))
    fig, ax = plt.subplots(figsize=(12, 5))
    dendrogram(L, truncate_mode="lastp", p=30, ax=ax, color_threshold=0)
    ax.set(title="Dendrogramme (Ward, 10 000 annonces, 30 dernières fusions)", xlabel="Groupes (taille entre parenthèses)", ylabel="Distance de Ward")
    plt.tight_layout(); plt.savefig(FIG / "clu_04_dendrogram.png", dpi=130); plt.close()
    pcs = PCA(8, random_state=SEED).fit_transform(Z)
    bic = []
    for k in range(2, 11):
        g = GaussianMixture(k, covariance_type="full", random_state=SEED, n_init=2).fit(pcs[:30000])
        bic.append(dict(k=k, bic=g.bic(pcs[:30000])))
    bic = pd.DataFrame(bic)
    bic.to_csv(ROOT / "reports" / "clust_gmm_bic.csv", index=False)
    out["gmm_k_bic"] = int(bic.sort_values("bic").k.iloc[0])
    gm = GaussianMixture(k_best, random_state=SEED, n_init=2).fit(pcs[:30000])
    out["ari_gmm_kmeans"] = float(adjusted_rand_score(S["cluster"], gm.predict(pcs)))

    # ---------------- PCA : variance expliquée et composantes
    pca = PCA(random_state=SEED).fit(Z)
    ev = pd.Series(pca.explained_variance_ratio_, index=[f"PC{i + 1}" for i in range(len(cols))])
    loadings = pd.DataFrame(pca.components_[:4].T, index=[FEATS[c] for c in cols], columns=[f"PC{i + 1}" for i in range(4)])
    loadings.to_csv(ROOT / "reports" / "clust_pca_loadings.csv")
    ev.to_csv(ROOT / "reports" / "clust_pca_variance.csv")
    fig, ax = plt.subplots(1, 2, figsize=(15, 5))
    ax[0].bar(ev.index, ev.values * 100, color="#4c78a8"); ax[0].plot(ev.index, ev.cumsum().values * 100, "o-", color="#e45756")
    ax[0].set(title="PCA : variance expliquée (barres) et cumulée (ligne)", xlabel="Composante", ylabel="% de variance")
    sns.heatmap(loadings, annot=True, fmt=".2f", cmap="RdBu_r", center=0, ax=ax[1], cbar=False)
    ax[1].set(title="PCA : poids des variables sur les 4 premières composantes", xlabel="Composante", ylabel="Variable")
    plt.tight_layout(); plt.savefig(FIG / "clu_05_pca.png", dpi=130); plt.close()
    out["pca_var_top2"] = float(ev.iloc[:2].sum()); out["pca_n_for_80"] = int((ev.cumsum() < 0.8).sum() + 1)

    # ---------------- UMAP (20 000) coloré par ville, prix, cluster
    import umap

    u = np.random.RandomState(SEED).choice(len(Z), 20000, replace=False)
    emb = umap.UMAP(n_neighbors=30, min_dist=0.3, random_state=SEED).fit_transform(Z[u])
    fig, ax = plt.subplots(1, 3, figsize=(20, 6))
    for a, (col, t, cm) in zip(ax, [(S.loc[u, "city"].astype("category").cat.codes, "ville", "tab20"), (S.loc[u, "log_price_rel"], "log prix relatif à la ville", "viridis"),
                                    (S.loc[u, "cluster"], "cluster k-means", "tab10")]):
        sc_ = a.scatter(emb[:, 0], emb[:, 1], c=col, s=2, cmap=cm, alpha=0.6)
        a.set(title=f"UMAP coloré par {t}", xlabel="UMAP 1", ylabel="UMAP 2")
        if t == "log prix relatif à la ville":
            plt.colorbar(sc_, ax=a, label="log(prix / médiane ville)")
    plt.tight_layout(); plt.savefig(FIG / "clu_06_umap.png", dpi=120); plt.close()

    # ---------------- DBSCAN géographique (Paris) : zones denses
    P = d[d.city == "paris"][["latitude", "longitude", "log_price_rel"]].dropna()
    coords = np.radians(P[["latitude", "longitude"]].values)
    db = DBSCAN(eps=0.15 / 6371, min_samples=40, metric="haversine", algorithm="ball_tree").fit(coords)
    P["zone"] = db.labels_
    n_zones = int((P.zone >= 0).sum() and P.zone.max() + 1)
    out.update(db_zones=n_zones, db_noise_pct=float((P.zone == -1).mean() * 100), db_n=len(P))
    fig, ax = plt.subplots(figsize=(9, 8))
    ax.scatter(P.longitude[P.zone == -1], P.latitude[P.zone == -1], s=1, c="lightgrey", label="isolées (bruit)")
    m = P.zone >= 0
    ax.scatter(P.longitude[m], P.latitude[m], s=2, c=P.zone[m] % 20, cmap="tab20", alpha=0.7)
    ax.set(title=f"Paris : zones denses détectées par DBSCAN (rayon 150 m, >= 40 annonces) : {n_zones} zones", xlabel="Longitude (°)", ylabel="Latitude (°)")
    ax.legend(markerscale=6)
    plt.tight_layout(); plt.savefig(FIG / "clu_07_dbscan_paris.png", dpi=130); plt.close()
    top_zones = P[P.zone >= 0].groupby("zone").agg(annonces=("zone", "size"), prix_rel=("log_price_rel", lambda s: float(np.exp(s.median())))).sort_values("annonces", ascending=False).head(8)
    top_zones.to_csv(ROOT / "reports" / "clust_dbscan_paris_top.csv")

    # ---------------- clustering des quartiers
    q = d.groupby(["city", "neighbourhood_cleansed"]).agg(n=("id", "size"), prix_rel=("log_price_rel", "median"), dist=("dist_rank", "median"), entier=("is_entire", "mean"),
                                                         pro=("host_is_pro", "mean"), note=("review_scores_rating", "median"), dispo=("availability_365", "median"),
                                                         capacité=("accommodates", "median")).reset_index()
    q = q[q.n >= 30].dropna()
    Zq = StandardScaler().fit_transform(q[["prix_rel", "dist", "entier", "pro", "note", "dispo", "capacité"]])
    kq = KMeans(5, n_init=10, random_state=SEED).fit(Zq)
    q["groupe"] = kq.labels_
    qp = q.groupby("groupe").agg(quartiers=("n", "size"), annonces=("n", "sum"), prix_rel=("prix_rel", lambda s: float(np.exp(s.median()))), dist=("dist", "median"), entier=("entier", "median"),
                                 pro=("pro", "median"), dispo=("dispo", "median")).round(2)
    qn = {}
    for g, r in qp.iterrows():
        qn[g] = ("chers" if r.prix_rel > 1.15 else "bon marché" if r.prix_rel < 0.85 else "prix moyens") + (" · centraux" if r.dist < 0.4 else " · périphériques" if r.dist > 0.6 else "") + (" · très pros" if r.pro > 0.5 else "")
    q["groupe_nom"] = q["groupe"].map(qn)
    qp["nom"] = qp.index.map(qn)
    q.to_csv(ROOT / "reports" / "clust_neighbourhoods.csv", index=False)
    qp.to_csv(ROOT / "reports" / "clust_neighbourhood_groups.csv")
    out["n_quartiers"] = len(q)

    # ---------------- anomalies (Isolation Forest)
    iso = IsolationForest(n_estimators=300, contamination=0.01, random_state=SEED, n_jobs=-1).fit(Z)
    S["anomaly_score"] = -iso.score_samples(Z)
    S["is_anomaly"] = iso.predict(Z) == -1
    an = S[S.is_anomaly]
    cmp = pd.DataFrame({"anomalies (1 %)": an[cols].median(), "ensemble": S[cols].median()})
    cmp.index = [FEATS[c] for c in cmp.index]
    cmp.round(2).to_csv(ROOT / "reports" / "clust_anomalies_profile.csv")
    an.sort_values("anomaly_score", ascending=False)[["city", "id", "price_eur", "accommodates", "bedrooms", "calculated_host_listings_count", "availability_365", "anomaly_score"]].head(15).to_csv(
        ROOT / "reports" / "clust_anomalies_top.csv", index=False)
    fig, ax = plt.subplots(figsize=(9, 5))
    ax.hist(S.anomaly_score, bins=80, color="#4c78a8")
    ax.axvline(S.loc[S.is_anomaly, "anomaly_score"].min(), color="#e45756", ls="--", label="seuil (1 % les plus atypiques)")
    ax.set(title="Isolation Forest : score d'atypie des annonces", xlabel="Score d'anomalie (plus haut = plus atypique)", ylabel="Nombre d'annonces")
    ax.legend()
    plt.tight_layout(); plt.savefig(FIG / "clu_08_anomalies.png", dpi=130); plt.close()
    S[["city", "id", "cluster", "cluster_name", "anomaly_score", "is_anomaly"]].to_parquet(ROOT / "data/processed/clusters_sample.parquet", index=False)
    pd.Series(out).to_json(ROOT / "reports" / "clust_summary.json", indent=1)
    print(out)
    print(prof.sort_values("prix_median_eur", ascending=False).to_string())
    print(qp.to_string())


if __name__ == "__main__":
    run()
```

## 16. Analyse des avis (NLP)

Les avis sont le seul texte libre rédigé par les clients. On cherche trois choses : **ce que ressentent les voyageurs** (sentiment), **de quoi ils parlent** (thèmes), **ce qui les gêne** (points de friction), puis on teste si ce signal **améliore la prédiction du prix et du statut Superhost** (même validation, avec et sans texte).

**Échantillonnage (explicite)** : 21 millions d'avis sont trop nombreux pour un portable. On prend **~8 000 annonces du train** (jeu de régression, ≥ 8 avis, 540 par ville au plus) et leurs **8 avis les plus récents**, soit **64 528 avis**. Ce choix **favorise les annonces bien documentées** : les résultats ne valent pas pour les annonces sans avis. Tout ce qui apprend sur le texte (TF-IDF, LDA, PCA des embeddings) est ajusté **dans chaque pli, sur les annonces d'entraînement seulement**.

````python
nlp = pickle.load(open(MOD / "_nlp_cache" / "analysis.pkl", "rb"))
rev = pd.read_parquet(MOD / "_nlp_cache" / "reviews_sample.parquet")
print(f"{fmt(nlp['n_reviews'])} avis, {fmt(nlp['n_listings'])} annonces ; langues : {nlp['langs']}")
rev[["city", "lang", "comments"]].sample(5, random_state=SEED).assign(comments=lambda d: d.comments.str[:130] + "…")
````

### 16.1 Détection de langue et nettoyage

### 16.2 Le sentiment, avec un modèle multilingue pré-entraîné

**Modèle** : `lxyuan/distilbert-base-multilingual-cased-sentiments-student` (DistilBERT multilingue distillé, sorties positif / neutre / négatif), appliqué à **tous** les avis de l'échantillon, tronqués à 128 tokens. Le **sentiment de l'avis** = P(positif) − P(négatif) ; un avis est **négatif** si P(négatif) > 0,5. Agrégation par annonce : moyenne, part d'avis négatifs.

### 16.5 TF-IDF ou embeddings : quelle représentation du texte ?

Pour comparer les deux représentations **à armes égales**, on les utilise **seules** (aucune variable tabulaire), avec un modèle linéaire, sur trois tâches posées au niveau de l'annonce (texte de ses 8 avis) : **note ≥ 4,8**, **Superhost**, **log(prix)**. TF-IDF : 1-2 grammes, 60 000 variables max. Embeddings : `paraphrase-multilingual-MiniLM-L12-v2` (384 dimensions), moyenne par annonce. Validation : 5 plis groupés par hôte.

````python
rc = nlp["rep_compare"]; tab = rc.groupby(["tâche", "représentation"]).score.agg(["mean", "std"]).round(4)
display(tab.style.format("{:.4f}"))
fig, ax = plt.subplots(1, 3, figsize=(17, 4.6))
for a, t in zip(ax, ["note ≥ 4,8", "Superhost", "log prix"]):
    d_ = rc[rc.tâche == t].groupby("représentation").score.agg(["mean", "std"])
    a.bar(range(len(d_)), d_["mean"], yerr=d_["std"], color=[PAL["jaune"], PAL["rose"]], edgecolor="white", linewidth=0.8, capsize=4); a.set_xticks(range(len(d_)), [i.split(" (")[0] for i in d_.index], fontsize=9)
    a.set(title=f"{t} · " + ("RMSE (plus bas = mieux)" if t == "log prix" else "ROC-AUC (plus haut = mieux)"))
    for i, v in enumerate(d_["mean"]): a.text(i, v, f"{v:.3f}", ha="center", va="bottom", fontsize=10)
plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>À armes égales, le choix de représentation dépend de la tâche.</b> Pour prédire <b>une note ≥ 4,8</b> et <b>le Superhost</b>, TF-IDF et embeddings sont <b>équivalents</b> (AUC 0,835 contre 0,837 ; 0,702 contre 0,702). Pour le <b>prix</b>, <b>TF-IDF est nettement meilleur</b> (RMSE 0,480 contre 0,513) : les mots précis (noms de quartiers, de monuments) portent de l'information géographique que la <b>moyenne d'un embedding sémantique lisse</b> (<i>explication plausible, non testée</i>). Les embeddings multilingues ont l'avantage de traiter toutes les langues d'un coup, mais perdent les mots-clés locaux. Le <b>texte seul</b> prédit le prix avec un RMSE de 0,48, à comparer aux 0,34 des variables tabulaires sur cet échantillon.</div>

### 16.6 Le texte améliore-t-il la prédiction ? Valeur ajoutée sur le prix et le Superhost

Même validation (5 plis groupés par hôte, sur les ~8 000 annonces de l'échantillon), même modèle (LightGBM). On ajoute au jeu de variables tabulaires : le **sentiment**, les **points de friction**, les **thèmes LDA**, les **embeddings (PCA 16)**, ou **tout le texte**. Deux jeux de base pour le Superhost : **A** (sans les notes) et **B** (avec les notes).

````python
va = nlp["value_added"]; piv = va.pivot_table(index=["tâche", "variante"], columns="pli", values="score")
ref = piv.xs("sans texte (référence)", level="variante")
rows = []
for (t, v), r in piv.iterrows():
    if v.startswith("sans"): continue
    d_ = r - ref.loc[t]; rows.append({"tâche": t, "ajout": v, "score référence": ref.loc[t].mean(), "score avec": r.mean(), "gain moyen": d_.mean(), "écart-type du gain (plis)": d_.std(), "plis améliorés (sur 5)": int((d_ < 0).sum()) if t.startswith("prix") else int((d_ > 0).sum())})
VA = pd.DataFrame(rows)
display(VA.style.format({"score référence": "{:.4f}", "score avec": "{:.4f}", "gain moyen": "{:+.4f}", "écart-type du gain (plis)": "{:.4f}"}).hide(axis="index").background_gradient(subset=["gain moyen"], cmap="RdYlGn"))
fig, ax = plt.subplots(1, 3, figsize=(19, 4.8))
for a, t in zip(ax, VA.tâche.unique()):
    d_ = VA[VA.tâche == t]; g = d_["gain moyen"] * (-1 if t.startswith("prix") else 1)
    a.barh(d_.ajout[::-1], g[::-1], xerr=d_["écart-type du gain (plis)"][::-1], color=[PAL["vert"] if v > 0 else PAL["rose"] for v in g[::-1]], edgecolor="white", linewidth=0.8, capsize=3); a.axvline(0, color=PAL["noir"])
    a.set(title=t, xlabel="amélioration (RMSE évité ou AUC gagnée)")
plt.tight_layout(); plt.show()
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #5B8EC2;background:#5B8EC214;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#5B8EC2">Ce qu'on en retient</span><br><b>Résultat net : le texte n'améliore pas la prédiction du prix.</b> Le RMSE de référence est de 0,3387 ; avec tout le texte il vaut 0,3382 (gain de 0,0005, <b>bien inférieur à l'écart-type entre plis de 0,004</b>) ; le sentiment seul <b>dégrade</b> légèrement (+0,0027, 4 plis sur 5). Les avis décrivent l'<b>expérience du séjour</b>, pas les caractéristiques qui fixent le prix (déjà dans les variables tabulaires). <b>Pour le Superhost sans les notes (jeu A), le texte apporte beaucoup</b> : l'AUC passe de <b>0,735 à 0,774</b> avec tout le texte (<b>+0,039</b>, 5 plis sur 5), dont <b>+0,036 avec les embeddings</b>, <b>+0,024 avec le sentiment</b> et +0,020 avec les thèmes LDA ; les points de friction n'apportent presque rien (+0,007). <b>Mais avec les notes (jeu B), le texte n'apporte plus rien</b> : 0,8289 → 0,8291 (+0,0002 ± 0,0055). Interprétation : les avis <i>contiennent déjà</i> l'information des notes ; le texte joue le rôle de <b>substitut aux notes</b>, pas de signal indépendant. C'est cohérent avec la corrélation de 0,50 entre sentiment et note, et avec le chapitre 14 : le statut Superhost se lit largement dans la satisfaction exprimée.</div>

> **Remarque.** <b>Limites du NLP</b> : (1) échantillon de 8 100 annonces <i>bien documentées</i> (≥ 8 avis) : les conclusions ne valent pas pour les annonces récentes ou sans avis ; (2) langues minoritaires mal couvertes (60 % d'anglais, 8 langues seulement au-dessus de 1 %) ; (3) <b>biais de positivité</b> (6 % d'avis négatifs) : les problèmes sont rares et ressortent mal ; (4) avis génériques (« great stay ! ») sans information ; (5) le lexique de friction ne couvre que l'anglais et reste une liste de mots ; (6) le sentiment est mesuré par un modèle dont la calibration varie selon la langue.

#### Code source du chapitre 16

<details>
<summary><code>src/nlp.py</code> (135 lignes) : Échantillonnage des avis, détection de langue, sentiment et embeddings multilingues</summary>

```python
"""Phase 9 : NLP sur les avis (et descriptions).

Étapes (résultats en cache dans models/_nlp_cache) :
  1. échantillon d'avis (stratifié par ville), détection de langue, nettoyage
  2. sentiment : modèle multilingue pré-entraîné (lxyuan/distilbert-base-multilingual-cased-sentiments-student), agrégé par annonce
  3. représentations : TF-IDF + modèle linéaire ; embeddings de phrases multilingues (paraphrase-multilingual-MiniLM-L12-v2) ;
     tâche : prédire la note de l'annonce (note < médiane) et comparer
  4. topics : LDA sur avis en anglais (thèmes appréciés / reprochés), + lexique de friction (propreté, bruit, localisation, accueil)
  5. valeur ajoutée : le sentiment / les thèmes améliorent-ils la prédiction du prix et du Superhost ? (même validation, avec / sans)
Échantillonnage explicite : ~8 000 annonces du train (jeu de régression, >= 8 avis, 540 par ville max), leurs 8 avis les plus récents (~60 000 avis). Usage : python -m src.nlp [etape]
"""
import pickle
import re
import sys
import time
import warnings

import lightgbm as lgb
import numpy as np
import pandas as pd
import pyarrow.parquet as pq
from sklearn.decomposition import LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedGroupKFold

from src.data import ROOT
from src.evaluate import group_key

warnings.filterwarnings("ignore")
SEED = 42
CACHE = ROOT / "models" / "_nlp_cache"
CACHE.mkdir(parents=True, exist_ok=True)
N_REVIEWS = 60000
SENT_MODEL = "lxyuan/distilbert-base-multilingual-cased-sentiments-student"
EMB_MODEL = "paraphrase-multilingual-MiniLM-L12-v2"


def clean(t: str) -> str:
    t = re.sub(r"<[^>]+>|http\S+", " ", str(t))
    return re.sub(r"\s+", " ", t).strip()


def sample_listings(per_city=540, min_reviews=8):
    """Annonces du TRAIN avec prix (jeu de régression) et >= 8 avis, échantillonnées par ville (échantillonnage explicite, voir en-tête)."""
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet", columns=["city", "id", "host_id", "number_of_reviews", "is_test", "in_reg_set"])
    d = d[~d.is_test & d.in_reg_set & (d.number_of_reviews >= min_reviews)]
    return pd.concat([g.sample(min(per_city, len(g)), random_state=SEED) for _, g in d.groupby("city")])[["city", "id"]]


def sample_reviews(k_per_listing=8):
    """Les k avis les plus récents de chaque annonce échantillonnée, nettoyés, avec langue détectée."""
    f = CACHE / "reviews_sample.parquet"
    if f.exists():
        return pd.read_parquet(f)
    L = sample_listings()
    ids = set(zip(L.city, L.id))
    pf = pq.ParquetFile(ROOT / "data/processed/reviews_all.parquet")
    parts = []
    for i in range(pf.num_row_groups):
        t = pf.read_row_group(i, columns=["city", "listing_id", "date", "comments"]).to_pandas()
        t = t[[(c, l) in ids for c, l in zip(t.city, t.listing_id)]]
        parts.append(t.dropna(subset=["comments"]))
    r = pd.concat(parts, ignore_index=True)
    r["comments"] = r["comments"].map(clean)
    r = r[r["comments"].str.len().between(30, 1200)]
    r["date"] = pd.to_datetime(r["date"], errors="coerce")
    r = r.sort_values("date", ascending=False).groupby(["city", "listing_id"]).head(k_per_listing).reset_index(drop=True)
    from langdetect import DetectorFactory, detect

    DetectorFactory.seed = SEED

    def det(s):
        try:
            return detect(s[:300])
        except Exception:
            return "unk"

    r["lang"] = r["comments"].map(det)
    r.to_parquet(f, index=False)
    return r


def sentiment(r):
    f = CACHE / "sentiment.parquet"
    if f.exists():
        return pd.read_parquet(f)
    import torch
    from transformers import AutoModelForSequenceClassification, AutoTokenizer

    torch.set_num_threads(16)
    tok = AutoTokenizer.from_pretrained(SENT_MODEL)
    mdl = AutoModelForSequenceClassification.from_pretrained(SENT_MODEL).eval()
    labels = [mdl.config.id2label[i].lower() for i in range(mdl.config.num_labels)]
    out = []
    texts = r["comments"].tolist()
    t0 = time.time()
    order = np.argsort([len(t) for t in texts])
    probs = np.zeros((len(texts), len(labels)), dtype="float32")
    with torch.no_grad():
        for s in range(0, len(texts), 64):
            idx = order[s:s + 64]
            b = tok([texts[i] for i in idx], padding=True, truncation=True, max_length=128, return_tensors="pt")
            probs[idx] = torch.softmax(mdl(**b).logits, -1).numpy()
            if (s // 64) % 100 == 0:
                print(f"sentiment {s}/{len(texts)} {time.time() - t0:.0f}s", flush=True)
    df = pd.DataFrame(probs, columns=[f"p_{l}" for l in labels])
    df["sentiment"] = df.get("p_positive", 0) - df.get("p_negative", 0)
    df.to_parquet(f, index=False)
    return df


def embeddings(r):
    f = CACHE / "embeddings.npy"
    if f.exists():
        return np.load(f)
    import torch
    from sentence_transformers import SentenceTransformer

    torch.set_num_threads(16)
    m = SentenceTransformer(EMB_MODEL, device="cpu")
    E = m.encode(r["comments"].tolist(), batch_size=128, show_progress_bar=False, normalize_embeddings=True)
    np.save(f, E.astype("float32"))
    return E


if __name__ == "__main__":
    step = sys.argv[1] if len(sys.argv) > 1 else "all"
    r = sample_reviews()
    print(len(r), "avis ;", r.lang.value_counts(normalize=True).head(6).round(3).to_dict(), flush=True)
    if step in ("sentiment", "all"):
        sentiment(r)
    if step in ("emb", "all"):
        embeddings(r)
```

```python
"""Phase 9 (suite) : analyses NLP à partir des avis échantillonnés, du sentiment et des embeddings (src/nlp.py).
Tout ce qui apprend sur le texte (TF-IDF, LDA, PCA des embeddings) est ajusté dans chaque pli sur les annonces d'entraînement seulement.
Usage : python -m src.nlp_analysis   -> models/_nlp_cache/analysis.pkl + reports/nlp_*.csv
"""
import pickle
import re
import warnings

import lightgbm as lgb
import numpy as np
import pandas as pd
from scipy.stats import spearmanr
from sklearn.decomposition import PCA, LatentDirichletAllocation
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression, Ridge
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import StratifiedGroupKFold

from src.data import ROOT
from src.evaluate import group_key, rmse
from src.features import build_preprocessor, feature_set
from src.models import fold_matrices
from src.nlp import CACHE, embeddings, sample_reviews, sentiment

warnings.filterwarnings("ignore")
SEED = 42
FRICTION = {
    "propreté": r"\b(dirty|filthy|stain\w*|smell\w*|stink\w*|dusty|mold\w*|mould\w*|cockroach\w*|bugs?|bedbugs?|hair on|not clean|unclean|grim\w*)\b",
    "bruit": r"\b(noise|noisy|loud|thin walls?|music|party|construction|street noise|couldn't sleep|could not sleep|barking)\b",
    "accueil / communication": r"\b(rude|unresponsive|no response|didn't answer|did not answer|never answered|unhelpful|unfriendly|hard to reach|cancel\w*)\b",
    "literie / confort": r"\b(uncomfortable|hard mattress|old mattress|sagging|pillows? (were|was)|tiny|cramped|too small|no air|stuffy)\b",
    "équipements en panne": r"\b(no hot water|broken|not working|didn't work|did not work|no heating|no wifi|wifi (was )?(down|bad|slow)|no air conditioning|no a/?c|leak\w*|mice|ants)\b",
    "localisation / sécurité": r"\b(far from|unsafe|sketchy|dangerous|isolated|hard to find|inconvenient|steep stairs|no elevator)\b",
    "conformité à l'annonce": r"\b(misleading|not as described|not as advertised|different from (the )?(photos|pictures)|photos? (are|were) (old|misleading)|false advertis\w*|disappointed)\b",
    "arrivée / clés": r"\b(lockbox|key|keys|check-?in|check in|late arrival|waited|waiting|couldn't get in|could not get in|code)\b.*\b(problem|issue|trouble|wait\w*|late|difficult|confus\w*)\b",
}


def build():
    f = CACHE / "analysis.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    r = sample_reviews()
    s = sentiment(r)
    E = embeddings(r)
    r = pd.concat([r.reset_index(drop=True), s.reset_index(drop=True)], axis=1)
    r["neg"] = (r["p_negative"] > 0.5).astype(int)
    en = r["lang"] == "en"
    txt = r["comments"].str.lower()
    for k, rx in FRICTION.items():
        r["fr_" + k] = (en & txt.str.contains(rx, regex=True)).astype(int)
    out = {"n_reviews": len(r), "langs": r.lang.value_counts(normalize=True).head(8).round(3).to_dict()}

    # ---- sentiment : distribution par langue et ville ; validité externe (corrélation avec la note de l'annonce)
    out["sent_by_lang"] = r.groupby("lang").agg(n=("sentiment", "size"), sentiment_moyen=("sentiment", "mean"), part_negatifs=("neg", "mean")).query("n >= 800").sort_values("n", ascending=False).round(3)
    out["sent_by_city"] = r.groupby("city").agg(sentiment_moyen=("sentiment", "mean"), part_negatifs=("neg", "mean"), n=("sentiment", "size")).round(3)
    L = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    L = L[~L.is_test & L.in_reg_set]
    key = r.groupby(["city", "listing_id"]).agg(sent_mean=("sentiment", "mean"), p_pos=("p_positive", "mean"), p_neg=("p_negative", "mean"), neg_share=("neg", "mean"), n_rev=("neg", "size"),
                                                 **{"fr_" + k: ("fr_" + k, "mean") for k in FRICTION}).reset_index().rename(columns={"listing_id": "id"})
    D = key.merge(L, on=["city", "id"], how="inner").reset_index(drop=True)
    out["n_listings"] = len(D)
    out["sent_vs_rating_spearman"] = float(spearmanr(D["sent_mean"], D["review_scores_rating"], nan_policy="omit")[0])
    out["sent_vs_price_spearman"] = float(spearmanr(D["sent_mean"], D["log_price_eur"])[0])

    # ---- lexique de friction (avis en anglais) : fréquence dans les avis négatifs vs positifs, par ville
    e = r[en]
    fr = pd.DataFrame({k: [e.loc[e.neg == 1, "fr_" + k].mean() * 100, e.loc[e.neg == 0, "fr_" + k].mean() * 100] for k in FRICTION}, index=["avis négatifs (%)", "avis non négatifs (%)"]).T
    fr["rapport"] = fr["avis négatifs (%)"] / fr["avis non négatifs (%)"].replace(0, np.nan)
    out["friction"] = fr.sort_values("avis négatifs (%)", ascending=False).round(2)
    out["friction_by_city"] = (e[e.neg == 1].groupby("city")[["fr_" + k for k in FRICTION]].mean() * 100).rename(columns=lambda c: c[3:]).round(1)
    out["n_neg_en"] = int((e.neg == 1).sum())
    out["share_neg"] = float(r.neg.mean())

    # ---- topics LDA (descriptif, avis en anglais)
    cv = CountVectorizer(stop_words="english", min_df=30, max_df=0.4, max_features=8000, token_pattern=r"(?u)\b[a-zA-Z]{3,}\b")
    Xc = cv.fit_transform(e["comments"])
    lda = LatentDirichletAllocation(8, learning_method="batch", max_iter=25, random_state=SEED, n_jobs=-1).fit(Xc)
    vocab = np.array(cv.get_feature_names_out())
    out["topics"] = pd.DataFrame({f"thème {i + 1}": vocab[np.argsort(c)[::-1][:12]] for i, c in enumerate(lda.components_)})
    tn = lda.transform(Xc)
    out["topic_sentiment"] = pd.DataFrame({"thème dominant": tn.argmax(1) + 1, "sentiment": e["sentiment"].values, "neg": e["neg"].values}).groupby("thème dominant").agg(
        part_des_avis=("sentiment", lambda s: len(s) / len(e)), sentiment_moyen=("sentiment", "mean"), part_negatifs=("neg", "mean")).round(3)

    # ---- texte par annonce : TF-IDF / embeddings (comparaison) puis valeur ajoutée (même validation)
    r = r.merge(D[["city", "id"]].rename(columns={"id": "listing_id"}), on=["city", "listing_id"])
    docs = r.groupby(["city", "listing_id"])["comments"].apply(lambda x: " ".join(x)).reset_index().rename(columns={"listing_id": "id"})
    D = D.merge(docs, on=["city", "id"])
    idx_map = {(c, l): i for i, (c, l) in enumerate(zip(r.city, r.listing_id))}
    Er = E[[i for i, (c, l) in enumerate(zip(sample_reviews().city, sample_reviews().listing_id)) if (c, l) in idx_map]] if False else None
    base_r = sample_reviews().reset_index(drop=True)
    pos = base_r.groupby(["city", "listing_id"]).indices
    Emb = np.vstack([E[pos[(c, i)]].mean(0) for c, i in zip(D.city, D.id)])
    y_price = D["log_price_eur"].values
    y_sh = D["is_superhost"].astype(int).values
    y_hi = (D["review_scores_rating"] >= 4.8).astype(float).values.copy()
    y_hi[D["review_scores_rating"].isna().values] = np.nan
    folds = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(D, D["city"], group_key(D)))
    feats_price = feature_set(D.columns, "core+quote+quality")
    # ATTENTION : pour prédire le Superhost, le jeu ne doit PAS contenir host_is_superhost (la cible). On réutilise les jeux de la classification (phase 7).
    from src.classification import sets_superhost

    sh_sets = sets_superhost(D.columns)
    feats_sh = {"A. sans notes": sh_sets["A. sans avis (caractéristiques seules)"], "B. avec notes": sh_sets["B. + notes d'avis"]}
    assert all("host_is_superhost" not in v for v in feats_sh.values())
    tfi_rows, val_rows = [], []
    fr_cols = ["fr_" + k for k in FRICTION]
    sent_cols = ["sent_mean", "p_pos", "p_neg", "neg_share"]
    P = dict(n_estimators=300, learning_rate=0.05, num_leaves=31, subsample=0.8, subsample_freq=1, colsample_bytree=0.7, n_jobs=16, random_state=SEED, verbose=-1)
    for i, (tr, va) in enumerate(folds):
        Dtr, Dva = D.iloc[tr], D.iloc[va]
        # TF-IDF vs embeddings (texte seul) : note élevée, Superhost, prix
        tf = TfidfVectorizer(ngram_range=(1, 2), min_df=5, max_features=60000, sublinear_tf=True).fit(Dtr["comments"])
        Ttr, Tva = tf.transform(Dtr["comments"]), tf.transform(Dva["comments"])
        ok_tr, ok_va = ~np.isnan(y_hi[tr]), ~np.isnan(y_hi[va])
        for task, y, kind in [("note ≥ 4,8", y_hi, "cls"), ("Superhost", y_sh, "cls"), ("log prix", y_price, "reg")]:
            ytr, yva = y[tr], y[va]
            mtr, mva = (~np.isnan(ytr), ~np.isnan(yva))
            for rep, (A, B) in {"TF-IDF (1-2 grammes)": (Ttr, Tva), "Embeddings multilingues (moyenne)": (Emb[tr], Emb[va])}.items():
                if kind == "cls":
                    m = LogisticRegression(C=4 if "TF" in rep else 2, max_iter=1000).fit(A[mtr], ytr[mtr].astype(int))
                    sc = roc_auc_score(yva[mva], m.predict_proba(B[mva])[:, 1])
                else:
                    m = Ridge(alpha=1.0 if "TF" in rep else 0.3).fit(A, ytr)
                    sc = rmse(yva, m.predict(B))
                tfi_rows.append(dict(pli=i, tâche=task, représentation=rep, score=sc))
        # valeur ajoutée sur les variables tabulaires
        pcs = PCA(16, random_state=SEED).fit(Emb[tr])
        Ptr, Pva = pcs.transform(Emb[tr]), pcs.transform(Emb[va])
        cvf = CountVectorizer(stop_words="english", min_df=20, max_df=0.4, max_features=5000, token_pattern=r"(?u)\b[a-zA-Z]{3,}\b")
        en_tr = r[(r.lang == "en")].merge(Dtr[["city", "id"]].rename(columns={"id": "listing_id"}), on=["city", "listing_id"])
        Xl = cvf.fit_transform(en_tr["comments"])
        ldaf = LatentDirichletAllocation(8, learning_method="batch", max_iter=15, random_state=SEED, n_jobs=-1).fit(Xl)

        def topic_feats(Dsub):
            rr = r[r.lang == "en"].merge(Dsub[["city", "id"]].rename(columns={"id": "listing_id"}), on=["city", "listing_id"])
            T = pd.DataFrame(ldaf.transform(cvf.transform(rr["comments"])), columns=[f"topic{j + 1}" for j in range(8)])
            T["city"], T["id"] = rr["city"].values, rr["listing_id"].values
            T = T.groupby(["city", "id"]).mean().reset_index()
            return Dsub[["city", "id"]].merge(T, on=["city", "id"], how="left").drop(columns=["city", "id"]).values

        Ttop_tr, Ttop_va = topic_feats(Dtr), topic_feats(Dva)
        extra = {"sentiment": (Dtr[sent_cols].values, Dva[sent_cols].values), "friction": (Dtr[fr_cols].values, Dva[fr_cols].values),
                 "thèmes LDA": (Ttop_tr, Ttop_va), "embeddings (PCA 16)": (Ptr, Pva),
                 "tout le texte": (np.hstack([Dtr[sent_cols].values, Dtr[fr_cols].values, Ttop_tr, Ptr]), np.hstack([Dva[sent_cols].values, Dva[fr_cols].values, Ttop_va, Pva]))}
        tasks = [("prix (log)", feats_price, y_price, "reg", None)] + [(f"Superhost {k}", v, y_sh, "cls", k) for k, v in feats_sh.items()]
        for tname, ft, y, kind, _ in tasks:
            Z, Zv, _, _ = fold_matrices(D, ft, tr, va, y, lambda f_, X, scale: build_preprocessor(f_, X, scale=scale, target_type="continuous" if kind == "reg" else "binary"))
            variants = {"sans texte (référence)": (Z.values, Zv.values)}
            for k, (a, b) in extra.items():
                variants["+ " + k] = (np.hstack([Z.values, a]), np.hstack([Zv.values, b]))
            for vn, (A, B) in variants.items():
                if kind == "reg":
                    p = lgb.LGBMRegressor(**P).fit(A, y[tr]).predict(B)
                    sc = rmse(y[va], p)
                else:
                    p = lgb.LGBMClassifier(**P).fit(A, y[tr]).predict_proba(B)[:, 1]
                    sc = roc_auc_score(y[va], p)
                val_rows.append(dict(pli=i, tâche=tname, variante=vn, score=sc))
        print(f"NLP pli {i} terminé", flush=True)
    out["rep_compare"] = pd.DataFrame(tfi_rows)
    out["value_added"] = pd.DataFrame(val_rows)
    pickle.dump(out, open(CACHE / "analysis.pkl", "wb"))
    return out


if __name__ == "__main__":
    o = build()
    print(o["n_reviews"], o["n_listings"], o["langs"])
    print(o["sent_by_lang"])
    print(o["friction"])
    print(o["topics"])
    print(o["rep_compare"].groupby(["tâche", "représentation"]).score.mean().round(4))
    v = o["value_added"].pivot_table(index=["tâche", "variante"], values="score", aggfunc=["mean", "std"]).round(4)
    print(v)
    print("sent~rating", o["sent_vs_rating_spearman"], "sent~price", o["sent_vs_price_spearman"])
```

## 17. Série temporelle du calendrier de disponibilité

Jusqu'ici, chaque ville n'était observée qu'à **une date**. Le fichier **calendar** donne, pour chaque annonce, la disponibilité de chacune des **365 prochaines nuits** : on peut donc étudier la **saisonnalité**. Nous l'avons téléchargé pour les 15 villes (≈ 400 Mo, `data/raw/*/calendar.csv.gz`).

**Deux limites à lire avant tout** : (1) le calendar 2026 **ne contient pas de prix** (seulement `available`, `minimum_nights`, `maximum_nights`) : la série étudiée est la **part de nuits indisponibles** (réservées *ou* bloquées par l'hôte), un proxy de la demande, pas un prix ; (2) c'est un **instantané unique** : l'effet de l'horizon (les réservations s'accumulent à l'approche de la date) est mélangé à la saisonnalité annuelle, qu'on ne peut donc pas isoler.

````python
daily = pd.read_parquet(PROC / "daily_availability.parquet"); city = daily[daily.segment == "ville"]
act = city.groupby("city").agg(annonces=("n_listings_all", "first"), actives=("n_listings_active", "first")); act["part active (%)"] = act.actives / act.annonces * 100
print(f"{fmt(len(daily))} lignes (ville × segment × jour). Annonces « actives » = au moins une nuit disponible dans l'année ; les autres sont bloquées en permanence (fantômes).")
act.sort_values("part active (%)").style.format({"annonces": "{:,.0f}", "actives": "{:,.0f}", "part active (%)": "{:.0f}"})
````

### 17.4 Prévoir la disponibilité à 28 jours

**Protocole.** Fenêtre glissante à origine mobile (`TimeSeriesSplit` à la main) : 5 origines successives, prévision des **28 jours suivants**, pour **chaque ville** (15 séries). Modèles : **baseline naïve saisonnière** (on répète la dernière semaine observée), **ETS** (tendance amortie + saisonnalité hebdomadaire), **SARIMA(1,1,1)(1,0,1,7)**, **LightGBM global** (toutes villes ensemble ; variables : jour de la semaine, mois, jours fériés du pays, distance au prochain férié, période de Noël, horizon, niveau récent). Métriques : **MAE** (en points de %) et **sMAPE**.

````python
fs = pd.read_csv(REP / "ts_forecast_summary.csv").set_index("modèle"); sc = pd.read_csv(REP / "ts_forecast_scores.csv")
display(fs.rename(columns={"mae": "MAE (pts de %)", "smape": "sMAPE (%)", "gain_mae_vs_naïf_%": "gain de MAE vs naïf (%)"}).style.format("{:.2f}").background_gradient(subset=["MAE (pts de %)"], cmap="RdYlGn_r"))
bc = sc.groupby(["city", "modèle"]).mae.mean().unstack()
vs = (bc.drop(columns="Naïf saisonnier (semaine précédente)").sub(bc["Naïf saisonnier (semaine précédente)"], axis=0) < 0).sum()
print("Nombre de villes (sur 15) où le modèle bat la baseline naïve :", vs.to_dict())
figcard("ts_04_forecast_paris.png", 900)
````

Les modèles paramétriques <b>extrapolent la tendance</b> à l'horizon de 28 jours et se trompent souvent davantage : SARIMA atteint 4,4 points de MAE à Londres (1,5 pour le naïf), l'ETS 5,2 à Mexico (1,1 pour le naïf). <b>LightGBM bat le naïf dans 4 villes sur 15</b> (Amsterdam, Lisbonne, Londres, Tokyo), l'ETS dans 3 (Amsterdam, Londres, Paris), SARIMA dans 1 (Barcelone) : pas assez pour compenser leurs erreurs ailleurs. Pour battre le naïf, il faudrait <b>plusieurs instantanés</b> (voir l'évolution des mêmes dates dans le temps) : avec un seul, l'apport du calendrier se limite à décrire.</div>

> **Remarque.** <b>Apport pour un hôte.</b> (1) Les <b>nuits de vendredi et samedi</b> sont les plus demandées dans presque toutes les villes (sauf Paris : jeudi) : c'est là qu'un prix plus élevé a le plus de chances de tenir. (2) Les <b>pics de fin d'année</b> (29-31 décembre) et d'événements locaux (Thanksgiving, Jour des morts) se voient dans la demande : ce sont les dates où ajuster les prix <i>à la hausse</i>. (3) À l'inverse, les <b>creux</b> (début de semaine, ponts de la Toussaint à Paris) justifient des remises ou des séjours minimum plus courts. <b>Attention</b> : ces conclusions portent sur la <i>disponibilité</i>, pas sur le prix payé (non fourni) ; une nuit « indisponible » peut être bloquée par l'hôte.

#### Code source du chapitre 17

<details>
<summary><code>src/timeseries.py</code> (239 lignes) : Calendrier : agrégation, décomposition STL, événements, modèles de prévision</summary>

```python
"""Phase 10 : séries temporelles sur le calendar (disponibilité par jour).

ATTENTION aux limites des données (documentées dans le rapport) :
  - le calendar 2026 ne contient PAS de prix (colonnes : listing_id, date, available, minimum_nights, maximum_nights) : on modélise la
    disponibilité (part de nuits indisponibles = réservées OU bloquées par l'hôte), proxy de la demande ;
  - c'est UN instantané pris à la date de collecte, qui regarde 365 jours vers l'avenir : la part indisponible dépend de l'horizon
    (forte à court terme : réservations + blocages ; minimale vers 6 mois ; qui remonte ensuite : calendriers pas encore ouverts)
    -> l'effet « horizon » est mélangé à la saisonnalité annuelle, que l'on ne peut donc pas isoler avec un seul instantané.
Usage : python -m src.timeseries   -> data/processed/daily_availability.parquet, reports/ts_*.csv, figures ts_*.png
"""
import time
import warnings

import holidays
import lightgbm as lgb
import matplotlib

matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from statsmodels.tsa.holtwinters import ExponentialSmoothing
from statsmodels.tsa.seasonal import STL
from statsmodels.tsa.statespace.sarimax import SARIMAX

from src.data import ROOT

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid", context="notebook")
FIG = ROOT / "reports" / "figures"
SEED = 42
COUNTRY = {"amsterdam": "NL", "barcelona": "ES", "berlin": "DE", "lisbon": "PT", "london": "GB", "los-angeles": "US", "lyon": "FR", "madrid": "ES",
           "mexico-city": "MX", "new-york-city": "US", "paris": "FR", "rome": "IT", "sydney": "AU", "tokyo": "JP", "toronto": "CA"}


# ------------------------------------------------------------------------------------------------ agrégation
def aggregate():
    out = ROOT / "data/processed/daily_availability.parquet"
    if out.exists():
        return pd.read_parquet(out)
    feat = pd.read_parquet(ROOT / "data/processed/listings_features.parquet", columns=["city", "id", "room_type", "neighbourhood_cleansed", "scrape_date"])
    rows = []
    for city in sorted(COUNTRY):
        f = next((ROOT / "data/raw" / city).glob("*/calendar.csv.gz"))
        t0 = time.time()
        c = pd.read_csv(f, usecols=["listing_id", "date", "available", "minimum_nights"], dtype={"available": "string"})
        c["unavail"] = (c["available"] == "f").astype("int8")
        # annonces « actives » : au moins une nuit disponible dans l'année (les autres sont bloquées en permanence = fantômes)
        act = c.groupby("listing_id")["unavail"].min().eq(0)
        n_all, n_act = len(act), int(act.sum())
        c = c[c["listing_id"].isin(act[act].index)]
        meta = feat[feat.city == city].set_index("id")[["room_type", "neighbourhood_cleansed"]]
        c = c.join(meta, on="listing_id")
        scrape = pd.to_datetime(feat.loc[feat.city == city, "scrape_date"].iloc[0])
        for seg_name, keys in [("ville", []), ("room_type", ["room_type"])] + ([("quartier", ["neighbourhood_cleansed"])] if city == "paris" else []):
            g = c.groupby(keys + ["date"]).agg(n_active=("unavail", "size"), share_unavail=("unavail", "mean"), min_nights_mean=("minimum_nights", "mean")).reset_index()
            g["segment"] = seg_name + ("" if not keys else ":" + g[keys[0]].astype(str))
            g = g.drop(columns=keys)
            g["city"], g["scrape_date"], g["n_listings_all"], g["n_listings_active"] = city, scrape, n_all, n_act
            rows.append(g)
        print(f"{city}: {n_act:,}/{n_all:,} annonces actives, {time.time() - t0:.0f}s", flush=True)
    df = pd.concat(rows, ignore_index=True)
    df["date"] = pd.to_datetime(df["date"])
    df["days_ahead"] = (df["date"] - df["scrape_date"]).dt.days
    df.to_parquet(out, index=False)
    return df


def calendar_features(dates, city, scrape):
    h = holidays.country_holidays(COUNTRY[city], years=sorted({d.year for d in dates}))
    hd = pd.Series(sorted(pd.Timestamp(k) for k in h.keys()))
    x = pd.DataFrame({"date": dates})
    x["dow"] = x.date.dt.dayofweek
    x["is_weekend_night"] = x.dow.isin([4, 5]).astype(int)  # nuit de vendredi ou samedi
    x["month"] = x.date.dt.month
    x["doy_sin"] = np.sin(2 * np.pi * x.date.dt.dayofyear / 365.25)
    x["doy_cos"] = np.cos(2 * np.pi * x.date.dt.dayofyear / 365.25)
    x["is_holiday"] = x.date.isin(hd).astype(int)
    nxt = [(hd[hd >= d].iloc[0] - d).days if (hd >= d).any() else 60 for d in x.date]
    x["days_to_holiday"] = np.clip(nxt, 0, 60)
    x["xmas_ny"] = (((x.date.dt.month == 12) & (x.date.dt.day >= 22)) | ((x.date.dt.month == 1) & (x.date.dt.day <= 3))).astype(int)
    x["days_ahead"] = (x.date - scrape).dt.days
    return x


# ------------------------------------------------------------------------------------------------ prévision
def smape(y, p):
    y, p = np.asarray(y), np.asarray(p)
    return float(np.mean(2 * np.abs(y - p) / (np.abs(y) + np.abs(p) + 1e-9)) * 100)


def forecast_eval(daily, horizon=28, n_splits=5):
    ser = {c: g.sort_values("date").set_index("date") for c, g in daily[daily.segment == "ville"].groupby("city")}
    T = min(len(s) for s in ser.values())
    cuts = [T - horizon * (n_splits - i) for i in range(n_splits)]  # origines successives (fenêtre extensible)
    rows, example = [], {}
    for k, cut in enumerate(cuts):
        # LightGBM global (toutes les villes) : variables calendaires + niveau récent connu à l'origine
        tr_parts, te_parts = [], []
        for c, s in ser.items():
            x = calendar_features(s.index, c, s.scrape_date.iloc[0])
            x["city"] = c
            x["y"] = s.share_unavail.values
            x["lvl7"] = s.share_unavail.iloc[cut - 7:cut].mean()
            x["lvl28"] = s.share_unavail.iloc[max(cut - 28, 0):cut].mean()
            x["slope"] = s.share_unavail.iloc[cut - 14:cut].mean() - s.share_unavail.iloc[cut - 28:cut - 14].mean()
            tr_parts.append(x.iloc[:cut].assign(lvl7=lambda z: z.y.rolling(7, min_periods=1).mean().shift(1).bfill(), lvl28=lambda z: z.y.rolling(28, min_periods=1).mean().shift(1).bfill(),
                                                slope=0.0))
            te_parts.append(x.iloc[cut:cut + horizon])
        tr, te = pd.concat(tr_parts), pd.concat(te_parts)
        cats = pd.CategoricalDtype(sorted(tr["city"].unique()))
        tr["city"], te["city"] = tr["city"].astype(cats), te["city"].astype(cats)
        feats = ["dow", "is_weekend_night", "month", "doy_sin", "doy_cos", "is_holiday", "days_to_holiday", "xmas_ny", "days_ahead", "lvl7", "lvl28", "city"]
        m = lgb.LGBMRegressor(n_estimators=300, learning_rate=0.05, num_leaves=15, min_child_samples=10, random_state=SEED, verbose=-1, n_jobs=8).fit(tr[feats], tr["y"])
        te["p_lgb"] = m.predict(te[feats])
        for c, s in ser.items():
            y = s.share_unavail.values
            trn, tst = y[:cut], y[cut:cut + horizon]
            preds = {}
            preds["Naïf saisonnier (semaine précédente)"] = np.tile(trn[-7:], horizon // 7 + 1)[:horizon]
            try:
                ets = ExponentialSmoothing(trn, trend="add", damped_trend=True, seasonal="add", seasonal_periods=7).fit()
                preds["ETS (tendance amortie + saison. hebdo)"] = ets.forecast(horizon)
            except Exception:
                preds["ETS (tendance amortie + saison. hebdo)"] = preds["Naïf saisonnier (semaine précédente)"]
            try:
                sar = SARIMAX(trn, order=(1, 1, 1), seasonal_order=(1, 0, 1, 7)).fit(disp=False, maxiter=60)
                preds["SARIMA(1,1,1)(1,0,1,7)"] = sar.forecast(horizon)
            except Exception:
                preds["SARIMA(1,1,1)(1,0,1,7)"] = preds["Naïf saisonnier (semaine précédente)"]
            preds["LightGBM global (calendrier + niveau récent)"] = te[te.city == c]["p_lgb"].values
            for name, p in preds.items():
                rows.append(dict(origine=k, city=c, modèle=name, mae=float(np.mean(np.abs(tst - p))) * 100, smape=smape(tst, p)))
            if c == "paris" and k == n_splits - 1:
                example = dict(dates=s.index[:cut + horizon], y=y[:cut + horizon], cut=cut, preds=preds)
        print(f"origine {k + 1}/{n_splits} évaluée", flush=True)
    return pd.DataFrame(rows), example


def run():
    daily = aggregate()
    city = daily[daily.segment == "ville"].copy()
    # ---- figure 1 : courbes par ville
    fig, axes = plt.subplots(3, 5, figsize=(20, 9), sharey=True)
    for ax, (c, g) in zip(axes.ravel(), city.groupby("city")):
        g = g.sort_values("date")
        ax.plot(g.date, g.share_unavail * 100, lw=0.8, color="#4c78a8")
        ax.plot(g.date, g.share_unavail.rolling(7, center=True).mean() * 100, lw=1.6, color="#e45756")
        ax.set_title(f"{c} ({g.n_listings_active.iloc[0]:,} annonces actives)", fontsize=9)
        ax.tick_params(axis="x", rotation=45, labelsize=7)
    fig.suptitle("Part de nuits indisponibles par jour (annonces actives) : instantané du calendar, 365 jours vers l'avenir (bleu : jour ; rouge : moyenne mobile 7 j)", fontsize=12)
    fig.supylabel("% de nuits indisponibles")
    plt.tight_layout()
    plt.savefig(FIG / "ts_01_unavailability_by_city.png", dpi=120)
    plt.close()

    # ---- décomposition STL (Paris) + profil hebdomadaire par ville (résidu après retrait de la tendance d'horizon)
    p = city[city.city == "paris"].sort_values("date").set_index("date")["share_unavail"] * 100
    stl = STL(p, period=7, robust=True).fit()
    fig, ax = plt.subplots(4, 1, figsize=(13, 9), sharex=True)
    for a, (s, t) in zip(ax, [(p, "Série observée"), (stl.trend, "Tendance (effet d'horizon + saison longue)"), (stl.seasonal, "Saisonnalité hebdomadaire"), (stl.resid, "Résidu (événements)")]):
        a.plot(s, lw=1)
        a.set_title(t, fontsize=10)
        a.set_ylabel("points de %")
    fig.suptitle("Paris : décomposition STL de la part de nuits indisponibles (période 7 jours)", y=1.0)
    plt.tight_layout()
    plt.savefig(FIG / "ts_02_stl_paris.png", dpi=120)
    plt.close()
    prof = {}
    for c, g in city.groupby("city"):
        g = g.sort_values("date").set_index("date")["share_unavail"] * 100
        s = STL(g, period=7, robust=True).fit()
        d = (s.seasonal + s.resid).groupby(g.index.dayofweek).mean()
        prof[c] = d
    prof = pd.DataFrame(prof).T
    prof.columns = ["lun", "mar", "mer", "jeu", "ven", "sam", "dim"]
    prof.to_csv(ROOT / "reports" / "ts_weekday_profile.csv")
    fig, ax = plt.subplots(figsize=(9, 7))
    sns.heatmap(prof, annot=True, fmt=".1f", cmap="RdBu_r", center=0, cbar_kws={"label": "écart à la tendance, points de %"}, ax=ax)
    ax.set(title="Profil hebdomadaire : nuits plus (rouge) ou moins (bleu) indisponibles que la tendance", xlabel="Jour de la nuit", ylabel="Ville")
    plt.tight_layout()
    plt.savefig(FIG / "ts_03_weekday_profile.png", dpi=130)
    plt.close()

    # ---- événements : jours dont le résidu STL est le plus fort / faible (par ville)
    ev = []
    for c, g in city.groupby("city"):
        g = g.sort_values("date").set_index("date")["share_unavail"] * 100
        r = STL(g, period=7, robust=True).fit().resid
        r = r.iloc[2:-21]  # on écarte les bords de la fenêtre (1er jour : annonces vues le jour de collecte ; 3 dernières semaines : calendriers non ouverts)
        z = (r - r.median()) / (r.abs().median() * 1.4826)
        hol = holidays.country_holidays(COUNTRY[c], years=[2026, 2027])
        for d, v in z.sort_values(ascending=False).head(4).items():
            ev.append(dict(city=c, date=d.date(), type="pic de demande", z=float(v), residu_pts=float(r[d]), jour_férié=hol.get(d.date(), "")))
        for d, v in z.sort_values().head(3).items():
            ev.append(dict(city=c, date=d.date(), type="creux", z=float(v), residu_pts=float(r[d]), jour_férié=hol.get(d.date(), "")))
    pd.DataFrame(ev).to_csv(ROOT / "reports" / "ts_events.csv", index=False)

    # ---- prévision : fenêtre glissante, 5 origines de 28 jours
    res, ex = forecast_eval(daily)
    res.to_csv(ROOT / "reports" / "ts_forecast_scores.csv", index=False)

Le réseau accentue aussi la <b>régression vers la moyenne</b> : écart moyen de −0,34 au premier décile de prix (−0,25 pour LightGBM) et de +0,47 au dernier (+0,40).</div>

> **Remarque.** <b>Budget comparé.</b> LightGBM : 40 essais en <b>20 minutes</b>. Réseau : 20 essais en <b>82 minutes</b> (4 fois plus long pour deux fois moins d'essais). Le coût de calcul ajoute un second argument en faveur du boosting. <b>Le comparatif n'est pas parfaitement symétrique</b> (20 essais contre 40, un réseau limité à 3 couches cachées), mais l'écart de 8 % dépasse largement la variation observée entre essais du réseau.

**Pourquoi les arbres gagnent souvent sur le tabulaire** (explications générales de la littérature, notamment Grinsztajn, Oyallon et Varoquaux, NeurIPS 2022 ; *non testées une à une ici*) :
1. **Les variables sont hétérogènes** (comptages, indicatrices, encodages, ratios) et les relations ont des **seuils nets** (capacité, nombre de nuits) : un arbre les capte par coupures, un réseau lisse.
2. **Beaucoup de variables peu informatives** (164 indicateurs d'équipements, la plupart à effet minime) : un arbre les ignore, un réseau doit apprendre à les ignorer.
3. **Invariance** : les arbres ne dépendent pas de l'échelle ni de la rotation des variables, les réseaux oui.
4. **Taille des données** : 80 000 lignes est modeste pour un réseau ; les réseaux profitent davantage des très grands jeux et de plus de réglage.
5. **Coût** : réglage 4 fois plus long ici.

**Pistes non réalisées** : un modèle de texte (embeddings des avis + tête de régression) ou multimodal. Le chapitre 16 montre que le texte n'améliore pas le prix, ce qui rend cette piste peu prometteuse pour la régression.

```python
"""Phase 11 : réseau de neurones dense (PyTorch, CPU) sur les données tabulaires vs boosting.

- catégories (type de logement, type de propriété, ville, quartier) -> couches d'embeddings (modalités rares regroupées)
- numériques : imputation médiane par ville + indicateurs de manquant + standardisation (ajustés sur le train de chaque pli)
- dropout, BatchNorm, AdamW, early stopping sur 12 % des hôtes du train du pli (jamais sur le pli de validation)
- comparaison HONNÊTE avec LightGBM : mêmes 80 000 annonces, mêmes plis A (5 plis groupés par hôte), réglage Optuna des deux (budgets annoncés)
Usage : python -m src.deep
"""
import json
import pickle
import time
import warnings

import lightgbm as lgb
import numpy as np
import optuna
import pandas as pd
import torch
import torch.nn as nn
from sklearn.model_selection import GroupKFold, GroupShuffleSplit, StratifiedGroupKFold
from sklearn.preprocessing import StandardScaler

from src.data import ROOT
from src.evaluate import group_key, rmse, mae
from src.features import CityMedianImputer, feature_set
from src.regression import N_COURSE, load_train, tune
from src.models import fold_matrices
from src.features import build_preprocessor

warnings.filterwarnings("ignore")
optuna.logging.set_verbosity(optuna.logging.WARNING)
SEED = 42
CACHE = ROOT / "models" / "_deep_cache"
CACHE.mkdir(parents=True, exist_ok=True)
torch.set_num_threads(16)
CATS = ["room_type", "property_type", "city", "nbhd_key"]


class Net(nn.Module):
    def __init__(self, n_num, cat_sizes, emb_dims, hidden, dropout):
        super().__init__()
        self.embs = nn.ModuleList([nn.Embedding(s, e) for s, e in zip(cat_sizes, emb_dims)])
        self.emb_drop = nn.Dropout(dropout / 2)
        d_in = n_num + sum(emb_dims)
        layers = []
        for h in hidden:
            layers += [nn.Linear(d_in, h), nn.BatchNorm1d(h), nn.ReLU(), nn.Dropout(dropout)]
            d_in = h
        layers.append(nn.Linear(d_in, 1))
        self.mlp = nn.Sequential(*layers)

    def forward(self, xn, xc):
        e = torch.cat([emb(xc[:, i]) for i, emb in enumerate(self.embs)], 1)
        return self.mlp(torch.cat([xn, self.emb_drop(e)], 1)).squeeze(1)


class TabPrep:
    """Prétraitement NN ajusté sur le train : numériques (imputation par ville + standardisation) et indices de catégories."""

    def __init__(self, feats, min_count=20):
        self.cat = [c for c in CATS if c in feats]
        self.num = [c for c in feats if c not in self.cat]
        self.min_count = min_count

    def fit(self, X):
        self.imp = CityMedianImputer(self.num).fit(X[self.num + ["city"]] if "city" not in self.num else X[self.num])
        Xi = self.imp.transform(X[self.num + ["city"]])
        self.cols = [c for c in Xi.columns if c != "city"]
        self.sc = StandardScaler().fit(Xi[self.cols].astype(float))
        self.maps = {}
        for c in self.cat:
            vc = X[c].astype(str).value_counts()
            self.maps[c] = {v: i + 1 for i, v in enumerate(vc[vc >= self.min_count].index)}  # 0 = rare / inconnu
        return self

    def transform(self, X):
        Xi = self.imp.transform(X[self.num + ["city"]])
        xn = self.sc.transform(Xi[self.cols].astype(float)).astype("float32")
        xc = np.column_stack([X[c].astype(str).map(self.maps[c]).fillna(0).astype(int).values for c in self.cat]).astype("int64")
        return xn, xc

    def cat_sizes(self):
        return [len(m) + 1 for m in self.maps.values()]


def fit_nn(Xtr, ytr, groups, prep, params, Xva=None, yva=None, max_epochs=60, patience=8, seed=SEED, log_curve=False):
    torch.manual_seed(seed)
    np.random.seed(seed)
    itr, iva = next(GroupShuffleSplit(1, test_size=0.12, random_state=seed).split(Xtr, groups=groups))
    xn, xc = prep.transform(Xtr)
    y = ytr.astype("float32")
    ymu, ysd = y[itr].mean(), y[itr].std()
    dims = [min(32, max(4, (s + 1) // 2)) for s in prep.cat_sizes()]
    net = Net(xn.shape[1], prep.cat_sizes(), dims, params["hidden"], params["dropout"])
    opt = torch.optim.AdamW(net.parameters(), lr=params["lr"], weight_decay=params["wd"])
    sched = torch.optim.lr_scheduler.ReduceLROnPlateau(opt, factor=0.5, patience=3)
    T = lambda a: torch.from_numpy(a)
    xn_t, xc_t, y_t = T(xn), T(xc), T((y - ymu) / ysd)
    best, best_state, bad, curve = 1e9, None, 0, []
    bs = params["batch"]
    for ep in range(max_epochs):
        net.train()
        perm = np.random.permutation(itr)
        tl = 0.0
        for s in range(0, len(perm), bs):
            b = perm[s:s + bs]
            if len(b) < 8:
                continue
            opt.zero_grad()
            loss = nn.functional.mse_loss(net(xn_t[b], xc_t[b]), y_t[b])
            loss.backward()
            opt.step()
            tl += loss.item() * len(b)
        net.eval()
        with torch.no_grad():
            vl = nn.functional.mse_loss(net(xn_t[iva], xc_t[iva]), y_t[iva]).item()
        sched.step(vl)
        curve.append((ep + 1, tl / len(itr), vl))
        if vl < best - 1e-5:
            best, bad = vl, 0
            best_state = {k: v.clone() for k, v in net.state_dict().items()}
        else:
            bad += 1
            if bad >= patience:
                break
    net.load_state_dict(best_state)
    net.eval()

    def predict(X):
        a, c = prep.transform(X)
        with torch.no_grad():
            return net(T(a), T(c)).numpy() * ysd + ymu

    return (predict, curve) if log_curve else predict


def run():
    f = CACHE / "deep.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    d = load_train()
    S = d.sample(N_COURSE, random_state=SEED).reset_index(drop=True)
    feats = feature_set(S.columns, "core+quote+quality")
    y = S["log_price_eur"].values
    g = group_key(S).values

    # ---------------- réglage Optuna du réseau (3 plis groupés, 20 essais) ; LightGBM : 40 essais déjà faits en phase 6
    folds3 = list(GroupKFold(3).split(S, groups=g))

    def obj(t):
        p = dict(hidden=[t.suggest_categorical("h1", [128, 256, 512]), t.suggest_categorical("h2", [64, 128, 256]), t.suggest_categorical("h3", [32, 64])],
                 dropout=t.suggest_float("dropout", 0.0, 0.4), lr=t.suggest_float("lr", 5e-4, 5e-3, log=True), wd=t.suggest_float("wd", 1e-6, 1e-2, log=True),
                 batch=t.suggest_categorical("batch", [512, 1024, 2048]))
        sc = []
        for tr, va in folds3[:2]:  # 2 plis pour limiter le coût
            prep = TabPrep(feats).fit(S.iloc[tr])
            pr = fit_nn(S.iloc[tr], y[tr], g[tr], prep, p, max_epochs=35, patience=5)
            sc.append(rmse(y[va], pr(S.iloc[va])))
            t.report(float(np.mean(sc)), len(sc))
            if t.should_prune():
                raise optuna.TrialPruned()
        return float(np.mean(sc))

    t0 = time.time()
    st = optuna.create_study(direction="minimize", sampler=optuna.samplers.TPESampler(seed=SEED), pruner=optuna.pruners.MedianPruner(n_startup_trials=4))
    st.optimize(obj, n_trials=20)
    nn_secs = time.time() - t0
    bp = st.best_params
    best = dict(hidden=[bp["h1"], bp["h2"], bp["h3"]], dropout=bp["dropout"], lr=bp["lr"], wd=bp["wd"], batch=bp["batch"])
    print("NN meilleurs réglages", best, round(st.best_value, 4), f"{nn_secs:.0f}s", flush=True)

    # ---------------- évaluation sur les 5 plis A (mêmes plis que la course de la phase 6)
    folds = list(StratifiedGroupKFold(5, shuffle=True, random_state=SEED).split(S, S["city"], group_key(S)))
    tuned = tune()
    lp = dict(tuned["LightGBM"]["best_params"], subsample_freq=1, random_state=SEED, n_jobs=16, verbose=-1)
    oof_nn, oof_lgb = np.zeros(len(S)), np.zeros(len(S))
    curves = []
    for i, (tr, va) in enumerate(folds):
        prep = TabPrep(feats).fit(S.iloc[tr])
        pr, curve = fit_nn(S.iloc[tr], y[tr], g[tr], prep, best, max_epochs=80, patience=10, log_curve=True)
        oof_nn[va] = pr(S.iloc[va])
        curves.append(curve)
        Z, Zv, _, _ = fold_matrices(S, feats, tr, va, y, build_preprocessor)
        oof_lgb[va] = lgb.LGBMRegressor(**lp).fit(Z, y[tr]).predict(Zv)
        print(f"pli {i}: NN {rmse(y[va], oof_nn[va]):.4f} | LightGBM {rmse(y[va], oof_lgb[va]):.4f}", flush=True)
    res = dict(best=best, nn_trials=[dict(n=t.number, v=t.value, s=str(t.state)) for t in st.trials], nn_seconds=nn_secs, nn_best_cv=st.best_value, lgb_seconds=tuned["LightGBM"]["seconds"],
               lgb_best_cv=tuned["LightGBM"]["best_rmse"], oof_nn=oof_nn, oof_lgb=oof_lgb, y=y, curves=curves, folds=[(tr.tolist(), va.tolist()) for tr, va in folds],
               city=S["city"].values)
    pickle.dump(res, open(f, "wb"))
    return res


if __name__ == "__main__":
    r = run()
    print("RMSE NN", rmse(r["y"], r["oof_nn"]), "LightGBM", rmse(r["y"], r["oof_lgb"]), "moyenne des deux", rmse(r["y"], (r["oof_nn"] + r["oof_lgb"]) / 2))
```

## 19. Évaluation finale sur le jeu de test

On ouvre enfin le **jeu de test** : 69 304 annonces (14,3 %) d'**hôtes jamais vus** ni par les modèles, ni par nos choix (sélection des variables, réglages, seuils). Il a été tiré au chapitre 3, avant toute exploration, et **utilisé une seule fois** (le script `src/final_model.py` refuse de s'exécuter une seconde fois). Les modèles finaux sont ajustés sur **tout le train**, les calibrages (intervalles, seuils) sur le train seulement. Si les scores de test sont proches de ceux de la validation croisée, la méthode est saine ; s'ils sont bien plus mauvais, quelque chose a fuité.

````python
test = json.load(open(REP / "test_results.json"))
tp = pd.read_parquet(PROC / "test_predictions.parquet")
print(f"Test propre (jeu de régression, prix connu, hors outliers/incohérences) : {fmt(test['n_test_clean'])} annonces ; test tous prix : {fmt(test['n_test_all_priced'])} ; test classification : {fmt(test['superhost_n'])}")
````

### 19.1 Le prix : test contre validation croisée

````python
fin = pickle.load(open(MOD / "_reg_cache" / "final.pkl", "rb")); yv = fin["y"]; po = fin["oof"]; pe_o, ye_o = np.exp(po), np.exp(yv)
cv = {"RMSE (log)": np.sqrt(np.mean((yv - po) ** 2)), "MAE (log)": np.mean(np.abs(yv - po)), "R²": 1 - ((yv - po) ** 2).sum() / ((yv - yv.mean()) ** 2).sum(), "MAE (EUR)": np.mean(np.abs(pe_o - ye_o)),
      "à ±20 % du prix prédit": np.mean(np.abs(pe_o - ye_o) / ye_o <= 0.2), "erreur absolue médiane (EUR)": np.median(np.abs(pe_o - ye_o))}
tc, ta = test["test propre (in_reg_set)"], test["test tous prix (outliers et incohérences inclus)"]
tab = pd.DataFrame({"validation croisée (309 412 annonces)": cv,
                    "TEST propre": {"RMSE (log)": tc["rmse"], "MAE (log)": tc["mae"], "R²": tc["r2"], "MAE (EUR)": tc["mae_eur"], "à ±20 % du prix prédit": tc["within20"], "erreur absolue médiane (EUR)": tc["median_ae_eur"]},
                    "TEST tous prix": {"RMSE (log)": ta["rmse"], "MAE (log)": ta["mae"], "R²": ta["r2"], "MAE (EUR)": ta["mae_eur"], "à ±20 % du prix prédit": ta["within20"], "erreur absolue médiane (EUR)": ta["median_ae_eur"]}})
display(tab.style.format("{:.3f}"))
# intervalle de confiance : bootstrap par hôtes (on rééchantillonne des hôtes entiers)
rng = np.random.RandomState(SEED); g = tp.assign(se=(tp.log_price_eur - tp.log_pred) ** 2, ae=np.abs(tp.pred_eur - tp.price_eur), inside=(tp.price_eur >= tp.low_eur) & (tp.price_eur <= tp.high_eur))
hid = g.city + "_" + g.host_id.astype(str); grp = g.groupby(hid.values)[["se", "ae", "inside"]].agg(["sum", "count"]); n_h = len(grp)
S_ = grp[("se", "sum")].values; A_ = grp[("ae", "sum")].values; I_ = grp[("inside", "sum")].values; N_ = grp[("se", "count")].values
bs = []
for _ in range(400):
    b = rng.randint(0, n_h, n_h); n = N_[b].sum(); bs.append((np.sqrt(S_[b].sum() / n), A_[b].sum() / n, I_[b].sum() / n))
bs = np.array(bs); ci = np.percentile(bs, [2.5, 97.5], axis=0)
print(f"Intervalles de confiance à 95 % (bootstrap sur {fmt(n_h)} hôtes) : RMSE {ci[0,0]:.4f} – {ci[1,0]:.4f} | MAE {ci[0,1]:.1f} – {ci[1,1]:.1f} EUR | couverture des intervalles {ci[0,2]:.1%} – {ci[1,2]:.1%}")
````

```python
"""Phase 14 : modèle final et ouverture du jeu de test (UNE SEULE FOIS).

1. ajuste le pipeline final sur tout le train (LightGBM réglé en phase 6) + calibration conforme et correction de biais issues de la CV (hors échantillon)
2. sauvegarde models/final_model.joblib (PriceModel) ; ajuste et sauvegarde les classifieurs Superhost et tranche de prix
3. évalue sur le test final (scellé depuis la phase 2) : prix, intervalles, classifications
Usage : python -m src.final_model       (refuse de s'exécuter deux fois, sauf --force)
"""
import json
import sys
import warnings

import joblib
import lightgbm as lgb
import numpy as np
import pandas as pd

from src.classification import best_f1_threshold, bin_metrics, multi_metrics, sets_superhost
from src.data import ROOT
from src.evaluate import rmse
from src.features import build_preprocessor, feature_set
from src.pricing import PriceModel
from src.regression import final as reg_final, full_metrics, tune

warnings.filterwarnings("ignore")
SEED = 42
OUT = ROOT / "reports" / "test_results.json"


def conformal_bins(y, p, alpha=0.1, n_bins=10):
    """Intervalle conforme Mondrian : quantile des |résidus| (log) par tranche de prix prédit, avec correction de taille finie."""
    edges = np.quantile(p, np.linspace(0, 1, n_bins + 1))
    edges[0], edges[-1] = -np.inf, np.inf
    b = np.clip(np.searchsorted(edges, p, side="right") - 1, 0, n_bins - 1)
    q = []
    for k in range(n_bins):
        r = np.abs(y[b == k] - p[b == k])
        level = min(1.0, np.ceil((len(r) + 1) * (1 - alpha)) / len(r))
        q.append(float(np.quantile(r, level)))
    return edges, np.array(q)


def run(force=False):
    if OUT.exists() and not force:
        print("Le jeu de test a déjà été ouvert (reports/test_results.json). Le test ne doit servir qu'une fois : abandon (utiliser --force pour refaire).")
        return json.load(open(OUT))
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    tr = d[~d.is_test & d.in_reg_set].reset_index(drop=True)
    te_all = d[d.is_test & d.has_price].reset_index(drop=True)
    te = te_all[te_all.in_reg_set].reset_index(drop=True)
    feats = feature_set(d.columns, "core+quote+quality")
    y = tr["log_price_eur"].values

    # --------------------------------------------------------- 1. pipeline final
    pre = build_preprocessor(feats, tr, scale=False)
    Z = pre.fit_transform(tr, y)
    pl = reg_final()["params"]
    model = lgb.LGBMRegressor(**pl).fit(Z, y)
    fin = reg_final()
    smear = float(np.mean(np.exp(fin["y"] - fin["oof"])))
    edges, qs = conformal_bins(fin["y"], fin["oof"])
    tmpl = {}
    cat = ["city", "room_type", "property_type", "nbhd_key"]
    cols = [c for c in feats if c not in cat] + cat
    for c, g in tr.groupby("city"):
        s = g[[k for k in cols if k != "city"]].median(numeric_only=True)
        for k in cat[1:]:
            s[k] = g[k].mode().iloc[0]
        for k in [k for k in s.index if k.startswith("am_")]:
            s[k] = 0.0
        s["city"] = c
        tmpl[c] = s
    centers = {c: (g.latitude.median(), g.longitude.median()) for c, g in tr.groupby("city")}
    nbc = tr.groupby(["city", "neighbourhood_cleansed"])[["latitude", "longitude"]].mean().reset_index()
    vocab = json.load(open(ROOT / "models" / "amenity_vocab.json", encoding="utf-8"))
    pm = PriceModel(pre, model, feats, smear, edges, qs, tmpl, centers, nbc, vocab)
    joblib.dump(pm, ROOT / "models" / "final_model.joblib")
    json.dump({"smear": smear, "conformal_q_by_decile": qs.tolist(), "bin_edges": [None if not np.isfinite(e) else float(e) for e in edges]},
              open(ROOT / "models" / "final_model_calibration.json", "w"), indent=1)

    # --------------------------------------------------------- 2. classifieurs
    clf = {}
    allt = d[~d.is_test].dropna(subset=["is_superhost"]).reset_index(drop=True)
    sets_sh = sets_superhost(allt.columns)
    feats_sh = sets_sh["B. + notes d'avis"]
    feats_shA = sets_sh["A. sans avis (caractéristiques seules)"]
    prs = build_preprocessor(feats_sh, allt, scale=False, target_type="binary")
    ys = allt["is_superhost"].astype(int).values
    Zs = prs.fit_transform(allt, ys)
    cP = dict(n_estimators=300, learning_rate=0.06, num_leaves=63, subsample=0.8, subsample_freq=1, colsample_bytree=0.8, n_jobs=16, random_state=SEED, verbose=-1)
    from sklearn.model_selection import GroupShuffleSplit

    a, b = next(GroupShuffleSplit(1, test_size=0.2, random_state=SEED).split(Zs, groups=allt["city"] + "_" + allt["host_id"].astype(str)))
    thr = best_f1_threshold(ys[b], lgb.LGBMClassifier(**cP).fit(Zs.iloc[a], ys[a]).predict_proba(Zs.iloc[b])[:, 1])  # seuil choisi sur le train seulement
    msh = lgb.LGBMClassifier(**cP).fit(Zs, ys)
    joblib.dump(dict(pre=prs, model=msh, feats=feats_sh, thr=thr), ROOT / "models" / "final_superhost.joblib")
    # variante A (caractéristiques seules, sans circularité avec les critères d'attribution)
    prA = build_preprocessor(feats_shA, allt, scale=False, target_type="binary")
    ZA = prA.fit_transform(allt, ys)
    thrA = best_f1_threshold(ys[b], lgb.LGBMClassifier(**cP).fit(ZA.iloc[a], ys[a]).predict_proba(ZA.iloc[b])[:, 1])
    mshA = lgb.LGBMClassifier(**cP).fit(ZA, ys)
    joblib.dump(dict(pre=prA, model=mshA, feats=feats_shA, thr=thrA), ROOT / "models" / "final_superhost_A.joblib")
    ttr = d[~d.is_test & d.in_reg_set].dropna(subset=["price_tier"]).reset_index(drop=True)
    prt = build_preprocessor(feats, ttr, scale=False, target_type="multiclass")
    yt = ttr["price_tier"].astype(int).values
    Zt = prt.fit_transform(ttr, yt)
    mt = lgb.LGBMClassifier(**cP).fit(Zt, yt)
    joblib.dump(dict(pre=prt, model=mt, feats=feats), ROOT / "models" / "final_price_tier.joblib")

    # --------------------------------------------------------- 3. TEST FINAL
    res = {"n_train_reg": len(tr), "n_test_clean": len(te), "n_test_all_priced": len(te_all), "smear": smear, "threshold_superhost": thr}
    rows = []
    for name, t in [("test propre (in_reg_set)", te), ("test tous prix (outliers et incohérences inclus)", te_all)]:
        out = pm.predict_frame(t)
        yt_ = t["log_price_eur"].values
        m = full_metrics(yt_, out.log_pred.values, smear)
        pe = out.price_eur.values
        ye = t["price_eur"].values
        m.update(within20=float(np.mean(np.abs(pe - ye) / ye <= 0.2)), coverage=float(np.mean((ye >= out.low_eur) & (ye <= out.high_eur))),
                 width_eur=float(np.mean(out.high_eur - out.low_eur)), median_ae_eur=float(np.median(np.abs(pe - ye))), n=len(t))
        res[name] = m
        rows.append(dict(jeu=name, **m))
        if name.startswith("test propre"):
            t = t.assign(pred_eur=pe, low_eur=out.low_eur.values, high_eur=out.high_eur.values, log_pred=out.log_pred.values)
            t[["city", "id", "host_id", "neighbourhood_cleansed", "room_type", "price_eur", "price_tier", "pred_eur", "low_eur", "high_eur", "log_pred", "log_price_eur"]].to_parquet(
                ROOT / "data/processed/test_predictions.parquet", index=False)
    # classification Superhost
    tes = d[d.is_test].dropna(subset=["is_superhost"])
    p = msh.predict_proba(prs.transform(tes))[:, 1]
    res["superhost"] = bin_metrics(tes["is_superhost"].astype(int).values, p, thr)
    res["superhost"]["auc_without_default_thr"] = res["superhost"]["auc"]
    res["superhost_n"] = len(tes)
    pA = mshA.predict_proba(prA.transform(tes))[:, 1]
    res["superhost_A"] = bin_metrics(tes["is_superhost"].astype(int).values, pA, thrA)
    np.save(ROOT / "data/processed/test_superhost_A_proba.npy", pA)
    np.save(ROOT / "data/processed/test_superhost_proba.npy", p)
    # tranche de prix
    tt = te.dropna(subset=["price_tier"])
    P = mt.predict_proba(prt.transform(tt))
    res["price_tier"] = multi_metrics(tt["price_tier"].astype(int).values, P)
    res["price_tier_n"] = len(tt)
    json.dump(res, open(OUT, "w"), indent=1, default=float)
    print(json.dumps(res, indent=1, default=float))
    return res


if __name__ == "__main__":
    run(force="--force" in sys.argv)
```

## 20. Interprétation du modèle

Un bon score ne suffit pas : un hôte veut savoir **pourquoi** son logement est estimé à ce prix. On interprète le **modèle final** (LightGBM réglé) avec **SHAP** (contribution de chaque variable à chaque prédiction, en log-prix) sur **22 500 annonces du train** (1 500 par ville), puis avec des **graphiques de dépendance partielle** (PDP / ICE).

**Toutes les explications ci-dessous sont des associations apprises par le modèle sur des prix affichés, pas des effets causaux.** Ajouter un équipement ne garantit pas la hausse de prix qu'il « explique ».

````python
import shap
shp = pickle.load(open(MOD / "_interp_cache" / "interp.pkl", "rb")); cols = shp["cols"]; sv = shp["sv"]; Zs = pd.DataFrame(shp["Z"], columns=cols); Sidx = shp["S"]
imp = pd.Series(np.abs(sv).mean(0), index=cols).sort_values(ascending=False)
print(f"SHAP sur {fmt(len(Zs))} annonces ; prix de base (moyenne du modèle) = {np.exp(shp['base']):.0f} EUR ; 259 variables prétraitées")
````

### 20.4 Les facteurs de prix sont-ils les mêmes à Paris et à Tokyo ?

````python
imp_c = shp["imp_city"]; top = imp.head(14).index
share = imp_c[top].div(imp_c.sum(axis=1), axis=0) * 100
fig, ax = plt.subplots(1, 2, figsize=(20, 6.8), gridspec_kw={"width_ratios": [1.5, 1]})
sns.heatmap(share.T, annot=True, fmt=".1f", cmap="YlOrBr", cbar_kws={"label": "% de l'importance totale de la ville"}, ax=ax[0]); ax[0].set(title="Part de chaque variable dans l'importance (SHAP), par ville", xlabel="", ylabel="")
rk = imp_c.T.rank(ascending=False); cm = rk.corr(method="spearman")
sns.heatmap(cm, annot=True, fmt=".2f", cmap="RdYlGn", vmin=0.85, vmax=1, cbar=False, ax=ax[1], annot_kws={"size": 7}); ax[1].set(title="Corrélation des classements d'importance entre villes (Spearman)")
plt.tight_layout(); plt.show()
off = cm.values[np.triu_indices(len(cm), 1)]
print(f"Corrélation de rangs entre villes : moyenne {off.mean():.3f}, minimum {off.min():.3f} ; Paris-Tokyo {cm.loc['paris', 'tokyo']:.3f} ; Paris-Londres {cm.loc['paris', 'london']:.3f}")
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #9A7DC0;background:#9A7DC014;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#9A7DC0">Ce qu'on en retient</span><br><b>Oui, pour l'essentiel : les facteurs de prix sont les mêmes d'une ville à l'autre.</b> La corrélation des classements d'importance entre villes est de <b>0,96 en moyenne</b> (minimum 0,92), <b>0,94 entre Paris et Tokyo</b>. Dans toutes les villes, capacité, chambres, quartier et durée du séjour dominent. Les <b>différences</b> portent sur les proportions : le <b>quartier</b> pèse <b>17,5 % à Mexico</b> et <b>15,1 % à Amsterdam</b> contre <b>4,9 % à Lisbonne</b> et <b>5,4 % à Barcelone</b> (où la géographie est portée par d'autres variables, <i>explication non testée</i>) ; la <b>durée du séjour</b> pèse <b>9-10 % à Toronto, Barcelone et New York</b> mais seulement <b>3 à 3,5 % à Mexico, Amsterdam et Londres</b> ; <b>New York</b> est la seule ville où l'indicateur de ville compte (13 % de son importance : un niveau de prix spécifique). La <b>latitude</b> compte plus que la longitude dans les deux villes, mais l'écart diffère : à Paris 4,3 % contre 1,3 %, à Tokyo 4,2 % contre 2,9 %. <b>Limite</b> : ces parts viennent d'<b>un seul modèle global</b> ; la stabilité des facteurs entre villes pourrait en partie résulter de ce que ce modèle commun leur impose. Un modèle par ville serait un test plus exigeant.</div>

> **Remarque.** <b>Corrélation n'est pas causalité.</b> « Le nombre de chambres explique le prix » signifie : dans les données, les logements avec plus de chambres sont affichés plus cher, toutes choses égales par ailleurs ; pas qu'ajouter une chambre à un logement multiplie sa valeur. Le prix de nos données est un <b>prix affiché</b>, fixé par des hôtes qui ont déjà intégré ces facteurs.

```python
"""Phase 12 : interprétabilité du modèle final (SHAP global / local / par ville, PDP / ICE). Résultats en cache : models/_interp_cache.
Les « leviers » sont des ASSOCIATIONS apprises, pas des effets causaux garantis.
Exécuter après src.final_model (utilise le modèle sauvegardé ; les exemples locaux viennent du test, ouvert une seule fois en phase 14).
Usage : python -m src.interpret
"""
import pickle
import warnings

import joblib
import numpy as np
import pandas as pd
import shap

from src.data import ROOT
from src.features import feature_set

warnings.filterwarnings("ignore")
SEED = 42
CACHE = ROOT / "models" / "_interp_cache"
CACHE.mkdir(parents=True, exist_ok=True)
PDP_VARS = ["accommodates", "bedrooms", "bath_count", "dist_center_km", "quote_nights", "n_amenities", "review_scores_rating", "host_tenure_years", "latitude"]


def build():
    f = CACHE / "interp.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    pm = joblib.load(ROOT / "models" / "final_model.joblib")
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    tr = d[~d.is_test & d.in_reg_set].reset_index(drop=True)
    S = pd.concat([g.sample(min(1500, len(g)), random_state=SEED) for _, g in tr.groupby("city")]).reset_index(drop=True)
    Z = pm.pre.transform(S)
    ex = shap.TreeExplainer(pm.model)
    sv = ex.shap_values(Z)
    out = dict(cols=list(Z.columns), sv=sv, Z=Z.values, base=float(np.ravel(ex.expected_value)[0]), S=S[["city", "id", "neighbourhood_cleansed", "room_type", "price_eur"]])
    # importance par ville (|SHAP| moyen) et corrélations de rangs
    imp_city = pd.DataFrame(np.abs(sv), columns=Z.columns).groupby(S.city.values).mean()
    out["imp_city"] = imp_city
    # PDP / ICE manuels sur la matrice prétraitée (variables numériques clés + climatisation)
    pdp = {}
    rng = np.random.RandomState(SEED)
    ice_idx = rng.choice(len(Z), 120, replace=False)
    for v in PDP_VARS + ["am_air_conditioning", "am_pool", "am_elevator", "am_dishwasher"]:
        if v not in Z.columns:
            continue
        grid = np.unique(np.quantile(Z[v], np.linspace(0.02, 0.98, 25))) if Z[v].nunique() > 2 else np.array([0.0, 1.0])
        curves = []
        for g in grid:
            Zg = Z.copy()
            Zg[v] = g
            curves.append(pm.model.predict(Zg))
        C = np.array(curves)  # (grille, annonces)
        pdp[v] = dict(grid=grid, mean=C.mean(1), ice=C[:, ice_idx], by_city={c: C[:, (S.city.values == c)].mean(1) for c in ["paris", "tokyo", "new-york-city", "mexico-city"]})
    out["pdp"] = pdp
    # explications locales : 3 annonces du test (bon marché, moyenne, luxe), après ouverture du test
    tp = pd.read_parquet(ROOT / "data/processed/test_predictions.parquet")
    full = d.set_index(["city", "id"])
    picks = {}
    q = tp.price_eur.quantile([0.1, 0.5, 0.97]).values
    for name, target in zip(["bon marché", "moyenne", "luxe"], q):
        # annonce de Paris la plus proche du quantile (cohérence d'exemple)
        c = tp[tp.city == "paris"]
        row = c.iloc[(c.price_eur - target).abs().argsort().iloc[0]]
        picks[name] = row
    rows = pd.DataFrame([full.loc[(r.city, r.id)].to_dict() | {"city": r.city, "id": r.id} for r in picks.values()])
    Zl = pm.pre.transform(rows)
    svl = ex.shap_values(Zl)
    out["local"] = dict(names=list(picks), rows=rows[["city", "id", "neighbourhood_cleansed", "room_type", "accommodates", "bedrooms", "bath_count", "dist_center_km", "n_amenities", "price_eur"]],
                        sv=svl, Z=Zl.values, pred=[float(pm.predict_frame(rows.iloc[[i]]).price_eur.iloc[0]) for i in range(3)])
    pickle.dump(out, open(f, "wb"))
    return out


if __name__ == "__main__":
    o = build()
    imp = pd.Series(np.abs(o["sv"]).mean(0), index=o["cols"]).sort_values(ascending=False)
    print(imp.head(15).round(4))

Le script `predict.py` en donne un exemple d'usage.

### 21.2 Le simulateur « et si… ? »

On prend **un même logement** (le studio parisien de la cellule précédente) et on change **une seule caractéristique à la fois** ; le modèle devient un outil de discussion : *que se passerait-il si…*

````python
def simuler(base, **chg):
    a = dict(base); [a.__setitem__(k, v) for k, v in chg.items()]; o = pm.predict_price(a)
    return o["prix_recommande_eur"], o["fourchette_eur"][0], o["fourchette_eur"][1]
base_p = simuler(studio)
scen = {"Logement de base (studio Paris, 2 nuits)": {}, "+ climatisation": {"amenities": studio["amenities"] + ["Air conditioning"]}, "+ lave-vaisselle": {"amenities": studio["amenities"] + ["Dishwasher"]},
        "+ piscine": {"amenities": studio["amenities"] + ["Pool"]}, "1 chambre au lieu de 0 (capacité 2)": {"bedrooms": 1, "beds": 1}, "capacité 4, 2 chambres, 3 lits": {"accommodates": 4, "bedrooms": 2, "beds": 3},
        "séjour de 7 nuits au lieu de 2": {"quote_nights": 7}, "séjour de 30 nuits (tarif mensuel)": {"quote_nights": 30}, "excentré (8 km du centre, nord de Paris)": {"latitude": 48.92, "longitude": 2.38},
        "même logement à Lisbonne": {"city": "lisbon", "latitude": 38.7139, "longitude": -9.1394}, "même logement à Tokyo": {"city": "tokyo", "latitude": 35.6895, "longitude": 139.6917}}
rows = []
for n, c in scen.items():
    p, lo, hi = simuler(studio, **c); rows.append({"scénario": n, "prix recommandé (EUR)": p, "fourchette 90 % (EUR)": f"{lo:.0f} – {hi:.0f}", "variation vs base (%)": (p / base_p[0] - 1) * 100})
res = pd.DataFrame(rows).set_index("scénario"); display(res.style.format({"prix recommandé (EUR)": "{:.0f}", "variation vs base (%)": "{:+.1f}"}))
fig, ax = plt.subplots(figsize=(11, 5)); r_ = res.iloc[1:]["variation vs base (%)"].sort_values()
ax.barh(r_.index, r_.values, color=[PAL["vert"] if v > 0 else PAL["rose"] for v in r_.values], edgecolor="white", linewidth=0.8); ax.axvline(0, color=PAL["noir"]); ax.set(title="Simulateur : variation du prix recommandé par rapport au studio de base", xlabel="% de variation"); plt.tight_layout(); plt.show()
````

> **Remarque.** <b>Le simulateur montre des associations apprises, pas des causes.</b> Ajouter une piscine à un studio parisien ne fait pas gagner mécaniquement le pourcentage affiché : la piscine signale surtout des logements <i>d'un autre type</i> que le modèle a appris à reconnaître. De même, la durée de séjour <i>n'est pas une caractéristique de l'annonce</i> mais du devis : elle mesure l'effet des remises de longue durée. Une conclusion causale demanderait une expérience (prix variés sur des logements comparables).

<summary><code>src/pricing.py</code> (90 lignes) : Classe PriceModel et fonction predict_price (prix recommandé et fourchette)</summary>

```python
"""Outil de recommandation de prix (Phase 13) : classe PriceModel (prétraitement + LightGBM + intervalle conforme).

Le prix prédit est le prix par nuit (EUR) d'un devis de séjour ; la fourchette est un intervalle de prédiction conforme « Mondrian »
(calibré par tranche de prix prédit sur les résidus hors échantillon de la validation croisée), de couverture visée 90 %.
"""
import json
from pathlib import Path

import numpy as np
import pandas as pd

from src.cleaning import norm_amenity
from src.features import haversine

ROOT = Path(__file__).resolve().parents[1]


class PriceModel:
    def __init__(self, pre, model, feats, smear, bin_edges, bin_q, templates, centers, nb_centroids, amenity_vocab, alpha=0.1):
        self.pre, self.model, self.feats = pre, model, feats
        self.smear, self.bin_edges, self.bin_q = smear, np.asarray(bin_edges), np.asarray(bin_q)
        self.templates, self.centers, self.nb_centroids, self.amenity_vocab, self.alpha = templates, centers, nb_centroids, amenity_vocab, alpha

    # ----------------------------------------------------------------------------------------- prédiction sur une table de features
    def predict_frame(self, X: pd.DataFrame) -> pd.DataFrame:
        Z = self.pre.transform(X)
        lp = self.model.predict(Z)
        b = np.clip(np.searchsorted(self.bin_edges, lp, side="right") - 1, 0, len(self.bin_q) - 1)
        q = self.bin_q[b]
        # prix recommandé = estimation MÉDIANE exp(log_pred) (minimise l'erreur absolue en euros) ; l'intervalle conforme est défini en log, donc sans smearing ;
        # mean_eur = prix MOYEN attendu (avec correction de biais de la retransformation), utile pour des totaux, pas pour un prix individuel
        out = pd.DataFrame({"log_pred": lp, "price_eur": np.exp(lp), "low_eur": np.exp(lp - q), "high_eur": np.exp(lp + q), "mean_eur": np.exp(lp) * self.smear}, index=X.index)
        return out

    # ----------------------------------------------------------------------------------------- une annonce décrite par un dictionnaire
    def build_row(self, listing: dict) -> pd.DataFrame:
        """Complète une annonce partielle avec le profil médian de sa ville, puis recalcule les variables dérivées."""
        city = listing["city"]
        if city not in self.templates:
            raise ValueError(f"Ville inconnue du modèle : {city}. Villes : {sorted(self.templates)}. (Pour une nouvelle ville : recalibrer, voir le rapport.)")
        row = self.templates[city].copy()
        for k, v in listing.items():
            if k in row.index:
                row[k] = v
        lat, lon = float(row["latitude"]), float(row["longitude"])
        c_lat, c_lon = self.centers[city]
        row["dist_center_km"] = float(haversine(lat, lon, c_lat, c_lon))
        if "neighbourhood_cleansed" in listing:
            row["nbhd_key"] = f"{city}|{listing['neighbourhood_cleansed']}"
        else:  # quartier le plus proche des coordonnées
            nb = self.nb_centroids[self.nb_centroids.city == city]
            j = int(np.argmin(haversine(lat, lon, nb.latitude.values, nb.longitude.values)))
            row["nbhd_key"] = f"{city}|{nb.neighbourhood_cleansed.iloc[j]}"
        acc = max(float(row["accommodates"]), 1.0)
        beds, bedrooms = float(row["beds"]), float(row["bedrooms"])
        row["beds_per_person"], row["persons_per_bedroom"] = beds / acc, acc / max(bedrooms, 1.0)
        row["bath_per_person"] = float(row["bath_count"]) / acc
        row["is_studio"] = float(bedrooms == 0)
        if "minimum_nights" in listing:
            row["log_min_nights"] = np.log1p(listing["minimum_nights"])
        if "maximum_nights" in listing:
            row["log_max_nights"] = np.log1p(min(listing["maximum_nights"], 1125))
        # contexte du devis (par défaut : séjour de 2 nuits, départ dans 14 jours)
        nights = listing.get("quote_nights", 2)
        lead = listing.get("quote_lead_days", 14)
        month = listing.get("quote_checkin_month", pd.Timestamp.today().month)
        row["quote_nights"], row["log_quote_lead_days"] = nights, np.log1p(lead)
        row["quote_month_sin"], row["quote_month_cos"] = np.sin(2 * np.pi * month / 12), np.cos(2 * np.pi * month / 12)
        row["quote_is_monthly"] = float(nights >= 28)
        row["quote_is_weekend"] = float(listing.get("quote_checkin_weekend", 0))
        # équipements : liste de libellés (normalisés comme au nettoyage)
        if "amenities" in listing:
            names = {norm_amenity(a) for a in listing["amenities"]}
            for a in self.amenity_vocab:
                row["am_" + a] = float(a in names)
            row["n_amenities_other"] = len(names - set(self.amenity_vocab))
            row["n_amenities"] = len(names)
        return row.to_frame().T.astype({c: float for c in row.index if c not in ("city", "room_type", "property_type", "nbhd_key")})

    def predict_price(self, listing: dict) -> dict:
        X = self.build_row(listing)
        X = X.assign(**{c: listing["city"] if c == "city" else X[c] for c in ["city"]})
        p = self.predict_frame(X).iloc[0]
        return {"prix_recommande_eur": round(float(p.price_eur), 1), "fourchette_eur": (round(float(p.low_eur), 1), round(float(p.high_eur), 1)), "prix_moyen_attendu_eur": round(float(p.mean_eur), 1), "couverture_visee": 1 - self.alpha}


def load_model(path=ROOT / "models" / "final_model.joblib") -> PriceModel:
    import joblib

    return joblib.load(path)
```

"""Exemple d'utilisation du modèle final de prix : python predict.py

Le modèle (models/final_model.joblib) contient le Pipeline complet : imputation par ville, encodage, LightGBM réglé, intervalle conforme à 90 %.
Le prix renvoyé est le prix par nuit (EUR) d'un devis de séjour (par défaut : 2 nuits, départ dans 14 jours) ; la fourchette vise 90 % de couverture
pour des villes connues du modèle. Ce sont des ASSOCIATIONS apprises sur des prix affichés, pas des prix payés ni des effets garantis.

# Une annonce est décrite par un dictionnaire PARTIEL : les champs manquants sont complétés par le profil médian de la ville.

grand_tokyo = {"city": "tokyo", "room_type": "Entire home/apt", "accommodates": 6, "bedrooms": 2, "beds": 4, "bath_count": 1.5,
               "latitude": 35.69, "longitude": 139.70, "amenities": ["Wifi", "Kitchen", "Air conditioning", "Washer"]}

for nom, annonce in [("Studio à Paris (centre)", studio_paris), ("2 chambres à Tokyo", grand_tokyo)]:

    lo, hi = out["fourchette_eur"]
    print(f"{nom} : {out['prix_recommande_eur']:.0f} EUR / nuit, fourchette à 90 % : {lo:.0f} - {hi:.0f} EUR (prix moyen attendu : {out['prix_moyen_attendu_eur']:.0f} EUR)")

# « Et si j'ajoute la climatisation ? » : même annonce, un seul changement (association apprise, pas effet causal)
sans, avec = dict(studio_paris), dict(studio_paris, amenities=studio_paris["amenities"] + ["Air conditioning"])
a, b = pm.predict_price(sans)["prix_recommande_eur"], pm.predict_price(avec)["prix_recommande_eur"]
print(f"Studio à Paris avec climatisation : {a:.0f} -> {b:.0f} EUR ({(b / a - 1) * 100:+.1f} %)")

# Nouvelle ville ? Le modèle refuse les villes inconnues : il faut recalibrer le niveau de prix (voir le rapport, chapitre « Apport métier »).
try:
    pm.predict_price(dict(studio_paris, city="vienne"))
except ValueError as e:
    print("Ville inconnue :", e)

## 22. Apports pour les hôtes, la plateforme et les villes

Les métriques ne valent que par les **décisions** qu'elles permettent. On traduit ici les résultats en euros et en actions pour les trois publics du projet, et on teste la question qui compte le plus pour la généralisation : **le modèle est-il exploitable dans une ville nouvelle sans réentraînement ?**

````python
biz = pickle.load(open(MOD / "_biz_cache" / "biz.pkl", "rb"))
tpb = tp.assign(ae=np.abs(tp.pred_eur - tp.price_eur), rel=np.abs(tp.pred_eur - tp.price_eur) / tp.price_eur, inside=(tp.price_eur >= tp.low_eur) & (tp.price_eur <= tp.high_eur), width=tp.high_eur - tp.low_eur)
print(f"{fmt(len(tpb))} annonces du jeu de test (hôtes jamais vus)")
````

### 22.1 Les métriques traduites en euros et en décisions

````python
q = pd.qcut(tpb.price_eur, 4, labels=["quart le moins cher", "2e quart", "3e quart", "quart le plus cher"])
t = tpb.groupby(q).agg(prix_médian=("price_eur", "median"), erreur_moyenne_EUR=("ae", "mean"), erreur_médiane_EUR=("ae", lambda s: s.median()), à_20pct=("rel", lambda s: (s <= 0.2).mean() * 100), couverture_90=("inside", lambda s: s.mean() * 100), largeur_fourchette_EUR=("width", "mean")).round(1)
glob = pd.DataFrame([{"prix_médian": tpb.price_eur.median(), "erreur_moyenne_EUR": tpb.ae.mean(), "erreur_médiane_EUR": tpb.ae.median(), "à_20pct": (tpb.rel <= 0.2).mean() * 100, "couverture_90": tpb.inside.mean() * 100, "largeur_fourchette_EUR": tpb.width.mean()}], index=["ENSEMBLE"]).round(1)
display(pd.concat([t, glob]).rename(columns={"prix_médian": "prix médian (EUR)", "erreur_moyenne_EUR": "erreur moyenne (EUR)", "erreur_médiane_EUR": "erreur médiane (EUR)", "à_20pct": "% à ±20 % du prix", "couverture_90": "% dans la fourchette 90 %", "largeur_fourchette_EUR": "largeur moyenne de la fourchette (EUR)"}))
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #9A7DC0;background:#9A7DC014;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#9A7DC0">Ce qu'on en retient</span><br><b>Pour un hôte</b>, le message est le suivant : <i>« le modèle se trompe en moyenne de 68 € par nuit, de 27 € dans la moitié des cas ; le prix réel est à ±20 % du prix prédit pour 53 % des annonces, et dans la fourchette annoncée pour 90 % des annonces »</i>. Le tableau par quart de prix précise que l'erreur <b>absolue</b> croît fortement avec le prix : <b>17 € de moyenne pour le quart le moins cher, 182 € pour le plus cher</b> (environ dix fois plus). La <b>couverture</b> de la fourchette est de 90 %, 94 % et 95 % pour les trois premiers quarts de prix réel, mais de <b>84 % pour le quart le plus cher</b> : la garantie à 90 % porte sur le prix <i>prédit</i>, pas sur le prix réel. La fourchette moyenne mesure 245 € : elle dit où le prix de marché a de grandes chances de se situer, pas quel prix précis choisir.</div>

### 22.2 Pour les hôtes : quels équipements sont associés à un prix plus élevé ?

Pour chaque équipement, on prend **2 500 annonces qui ne l'ont pas**, on le leur ajoute dans les données et on mesure la variation du **prix prédit** (le reste inchangé). C'est une **association apprise par le modèle**, pas un effet garanti.

````python
w = biz["whatif"].copy(); w = w[w.part_des_annonces >= 0.015]
top = w.sort_values("effet_pct", ascending=False).head(10)[::-1]; bot = w.sort_values("effet_pct").head(5)
fig, ax = plt.subplots(1, 2, figsize=(17, 6), gridspec_kw={"width_ratios": [1.3, 1]})
ax[0].barh(top.équipement, top.effet_pct, xerr=[top.effet_pct - top.ic95_bas, top.ic95_haut - top.effet_pct], color=PAL["vert"], edgecolor="white", linewidth=0.8, capsize=3)
for i, (e, r) in enumerate(top.iterrows()): ax[0].text(r.ic95_haut + 0.15, i, f"{r.effet_pct:+.1f} %  ({r.part_des_annonces:.0%} l'ont déjà)", va="center", fontsize=8.5)
ax[0].set(title="10 équipements associés à la plus forte hausse du prix prédit", xlabel="variation du prix prédit si on l'ajoute (%, IC à 95 % sur l'échantillon)", xlim=(0, 12))
ax[1].barh(bot.équipement[::-1], bot.effet_pct[::-1], color=PAL["rose"], edgecolor="white", linewidth=0.8); ax[1].axvline(0, color=PAL["noir"]); ax[1].set(title="5 équipements associés à une BAISSE du prix prédit", xlabel="variation (%)")
plt.tight_layout(); plt.show()
display(top[::-1][["équipement", "part_des_annonces", "effet_pct", "ic95_bas", "ic95_haut"]].rename(columns={"part_des_annonces": "part des annonces qui l'ont", "effet_pct": "effet (%)", "ic95_bas": "IC95 bas", "ic95_haut": "IC95 haut"}).style.format({"part des annonces qui l'ont": "{:.0%}", "effet (%)": "{:+.1f}", "IC95 bas": "{:+.1f}", "IC95 haut": "{:+.1f}"}).hide(axis="index"))
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #9A7DC0;background:#9A7DC014;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#9A7DC0">Ce qu'on en retient</span><br>Les équipements associés à la plus forte hausse du prix prédit sont <b>la piscine (+7,9 %)</b>, <b>le lave-vaisselle (+7,5 %)</b>, <b>la cheminée intérieure (+6,8 %)</b>, <b>la climatisation centralisée (+6,2 %)</b> et <b>la climatisation (+6,1 %)</b>, puis le coffre-fort (+4,6 %), l'ascenseur (+3,9 %), la vue sur la ligne d'horizon (+3,7 %), le salon privé (+3,5 %) et le petit-déjeuner (+3,4 %). Pour un hôte, les leviers <b>accessibles</b> sont ceux que beaucoup de logements n'ont pas encore mais qu'on peut installer : la <b>climatisation</b> (44 % des annonces l'ont : plus de la moitié ne l'ont pas) et le <b>lave-vaisselle</b> (36 % l'ont). La piscine et la cheminée (4-6 % des annonces) signalent plutôt un <b>autre type de bien</b> qu'un équipement ajoutable. <b>À l'inverse, certains équipements sont associés à une baisse</b> : le <b>verrou sur la porte de chambre</b> (−7,3 %) est un marqueur de <b>chambres louées chez l'habitant</b> (le modèle reconnaît le type de logement), l'<b>eau chaude</b> (−3,9 %) ou le <b>micro-ondes</b> (−3,1 %) ne font évidemment pas baisser un prix : ce sont des équipements presque universels (84 % et 66 % des annonces) que le modèle semble utiliser comme marqueur de logement <i>standard</i> plutôt que haut de gamme (<i>hypothèse non testée</i>). Ce contraste est la meilleure illustration de la <b>différence entre association et causalité</b>.</div>

> **Remarque.** <b>Avertissement.</b> Ces pourcentages décrivent ce que le <i>modèle</i> a appris, sur des logements existants. Installer une climatisation ne produira pas mécaniquement +6 % : l'effet réel dépend du logement, de la saison, de la ville, et du fait qu'elle soit mise en avant. Les intervalles de confiance ne mesurent que la variabilité de l'échantillon de 2 500 annonces, <b>pas l'incertitude causale</b>.

C'est cohérent avec une demande qui réagit à un prix trop élevé, mais c'est une <b>association</b>, pas une preuve. ````python
d_tr = pd.read_parquet(PROC / "listings_features.parquet", columns=["is_test", "is_superhost", "review_scores_rating", "review_scores_cleanliness", "review_scores_communication", "number_of_reviews", "number_of_reviews_ltm",
                                                                 "host_tenure_years", "host_is_pro", "n_amenities", "log_price_eur", "city", "availability_365", "host_identity_verified"])
d_tr = d_tr[~d_tr.is_test].dropna(subset=["is_superhost"]); d_tr["price_rel"] = np.exp(d_tr.log_price_eur - d_tr.groupby("city").log_price_eur.transform("median"))
prof = d_tr.groupby("is_superhost").agg(**{"note globale (médiane)": ("review_scores_rating", "median"), "propreté (médiane)": ("review_scores_cleanliness", "median"), "communication (médiane)": ("review_scores_communication", "median"),
                                           "avis au total (médiane)": ("number_of_reviews", "median"), "avis sur 12 mois (médiane)": ("number_of_reviews_ltm", "median"), "ancienneté de l'hôte (années, médiane)": ("host_tenure_years", "median"),
                                           "hôte pro (≥ 5 annonces) %": ("host_is_pro", lambda s: s.mean() * 100), "nb équipements (médiane)": ("n_amenities", "median"), "prix relatif à la ville (médiane)": ("price_rel", "median"),
                                           "identité vérifiée %": ("host_identity_verified", lambda s: s.mean() * 100), "jours disponibles / an (médiane)": ("availability_365", "median")}).T
prof.columns = ["non Superhost", "Superhost"]; prof["écart"] = prof["Superhost"] - prof["non Superhost"]
display(prof.style.format("{:.2f}"))
import joblib
mA = joblib.load(MOD / "final_superhost_A.joblib"); dfe = pd.read_parquet(PROC / "listings_features.parquet"); dfe = dfe[~dfe.is_test].dropna(subset=["is_superhost"]).sample(3000, random_state=SEED)
cols_sh = list(mA["pre"].transform(dfe).columns); gain = pd.Series(mA["model"].booster_.feature_importance("gain"), index=cols_sh); gain = (gain / gain.sum() * 100).sort_values(ascending=False).head(10)
fig, ax = plt.subplots(figsize=(9, 4.6)); ax.barh(gain.index[::-1], gain.values[::-1], color=PAL["rose"], edgecolor="white", linewidth=0.8); ax.set(title="Superhost (caractéristiques seules) : variables les plus utilisées par le modèle", xlabel="part du gain total (%)"); plt.tight_layout(); plt.show()
````

### 22.4 Pour les investisseurs et les villes

````python
nb = biz["neighbourhoods"]; sig = nb[(nb.ic_bas > 0) | (nb.ic_haut < 0)]
sous = nb[nb.ic_haut < 0].head(8); sur = nb[nb.ic_bas > 0].tail(8)[::-1]
fig, ax = plt.subplots(1, 2, figsize=(18, 5.6))
for a, d_, col, t in [(ax[0], sous[::-1], PAL["bleu"], "Quartiers SOUS-évalués (prix observé inférieur au prix attendu)"), (ax[1], sur[::-1], PAL["rose"], "Quartiers SUR-évalués (prix observé supérieur au prix attendu)")]:
    lab = d_.city + " · " + d_.neighbourhood_cleansed
    a.barh(lab, d_.ecart_pct, xerr=[d_.ecart_pct - d_.ic_bas, d_.ic_haut - d_.ecart_pct], color=col, edgecolor="white", linewidth=0.8, capsize=3); a.axvline(0, color=PAL["noir"]); a.set(title=t, xlabel="écart moyen observé − prédit (%, IC à 95 %)")
plt.tight_layout(); plt.show()
print(f"{len(nb)} quartiers de ≥ 40 annonces ; {len(sig)} ({len(sig) / len(nb):.0%}) ont un écart significatif (IC à 95 % qui exclut zéro).")
display(pd.concat([sous.assign(sens="sous-évalué"), sur.assign(sens="sur-évalué")])[["sens", "city", "neighbourhood_cleansed", "n", "prix_median", "ecart_pct", "ic_bas", "ic_haut"]].rename(columns={"neighbourhood_cleansed": "quartier", "n": "annonces", "prix_median": "prix médian (EUR)", "ecart_pct": "écart (%)", "ic_bas": "IC95 bas", "ic_haut": "IC95 haut"}).style.format({"prix médian (EUR)": "{:.0f}", "écart (%)": "{:+.1f}", "IC95 bas": "{:+.1f}", "IC95 haut": "{:+.1f}"}).hide(axis="index"))
````

### 22.5 Et une ville que le modèle n'a jamais vue ?

````python
rc = biz["recal_curve"]; rcm = rc.groupby(["variante", "n_calib"]).rmse.mean().unstack(0)
# couverture : décalage seul, ou décalage + largeur de l'intervalle recalibrés sur n annonces de la nouvelle ville
rng2 = np.random.RandomState(SEED); fb = pickle.load(open(MOD / "_reg_cache" / "final.pkl", "rb")); yb, ib = fb["y"], fb["d_index"]
covs = {}
for n in [25, 50, 100, 200, 500]:
    acc = []
    for c in sorted(ib.city.unique()):
        m = np.where((ib.city == c).values)[0]; r = yb[m] - fb["oofB"]["V1"][m]; vals = []
        for _ in range(30):
            cal = rng2.choice(len(m), n, replace=False); te = np.setdiff1d(np.arange(len(m)), cal); sh = r[cal].mean(); q = np.quantile(np.abs(r[cal] - sh), min(1, np.ceil((n + 1) * 0.9) / n)); vals.append(np.mean(np.abs(r[te] - sh) <= q))
        acc.append(np.mean(vals))
    covs[n] = np.mean(acc)
fig, ax = plt.subplots(1, 2, figsize=(16, 5.2))
ax[0].plot(rcm.index, rcm["V1"], "o-", color=PAL["rose"], mec="white", lw=2.5, label="toutes les variables (V1)"); ax[0].plot(rcm.index, rcm["V3"], "s--", color=PAL["bleu"], mec="white", lw=2, label="sans identifiants géographiques absolus (V3)")
ax[0].axhline(0.356, color=PAL["noir"], ls=":"); ax[0].text(300, 0.362, "ville connue (CV : 0,356)", fontsize=8)
ax[0].set(xscale="symlog", title="Ville inconnue : erreur selon le nombre d'annonces utilisées pour recaler le niveau", xlabel="annonces de la nouvelle ville utilisées pour recalibrer (0 = aucune)", ylabel="RMSE moyen par ville (log-prix)"); ax[0].legend()
cc = biz["recal_cov"].groupby(["variante", "cas"]).couverture.mean().unstack()
ax[1].bar(["sans\nrecalibration", "décalage de niveau\n(200 annonces)", "niveau + largeur de\nl'intervalle (200)"], [cc.loc["V1", "sans recalibration"] * 100, cc.loc["V1", "recalibré (200 annonces)"] * 100, covs[200] * 100], color=[PAL["gris"], PAL["jaune"], PAL["vert"]], edgecolor="white", linewidth=0.8)
ax[1].axhline(90, color=PAL["noir"], ls="--"); ax[1].set(title="Ville inconnue : couverture de la fourchette annoncée à 90 %", ylabel="% de prix réels dans la fourchette", ylim=(0, 100))
for i, v in enumerate([cc.loc["V1", "sans recalibration"], cc.loc["V1", "recalibré (200 annonces)"], covs[200]]): ax[1].text(i, v * 100 + 1.5, f"{v:.0%}", ha="center", fontsize=11)
plt.tight_layout(); plt.show()
tab = rcm.rename(columns={"V1": "RMSE V1", "V3": "RMSE V3"}); tab["couverture (niveau + largeur)"] = pd.Series(covs); tab.index.name = "annonces utilisées"; display(tab.T.style.format("{:.3f}", na_rep="—"))
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #9A7DC0;background:#9A7DC014;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#9A7DC0">Ce qu'on en retient</span><br><b>Sans aucune recalibration, le modèle n'est pas exploitable dans une ville nouvelle</b> : le RMSE moyen par ville est de <b>0,555</b> (contre 0,356 en ville connue) et la fourchette annoncée à 90 % ne contient le prix réel que <b>65 % du temps</b> (jusqu'à <b>14 % à Mexico</b> et 39 % à New York). <b>Mais le défaut est presque entièrement un décalage de niveau, facile à corriger</b> : avec seulement <b>25 annonces</b> de la nouvelle ville (leur prix moyen contre le prix prédit), le RMSE tombe à <b>0,419</b> ; avec 100 à 200, à <b>0,41</b>, soit à peine plus que 0,356 en ville connue. La <b>fourchette</b> demande de recaler aussi sa <b>largeur</b> : un simple décalage donne 83 % de couverture, <b>décalage + largeur estimés sur 100 à 200 annonces ramènent la couverture à 90 %</b> (minimum par ville : 89 %). Retirer les identifiants géographiques absolus (V3) ne change presque rien après recalibration (0,418 contre 0,412).</div>

**Que faut-il recalibrer pour une nouvelle ville ?**
1. **Le niveau de prix** : prix moyen observé moins prix moyen prédit sur 25 à 100 annonces de la nouvelle ville (une correction additive sur le log-prix).
2. **La largeur de l'intervalle** : quantile à 90 % des résidus absolus de ces 100 à 200 annonces, après correction du niveau.
3. **La devise** : taux de change BCE à la date de collecte (`data/external`, `config.yaml`).
4. **Les références géographiques** : centre-ville (médiane des coordonnées) et table des quartiers, pour la distance au centre et l'encodage du quartier (un quartier inconnu est encodé par la moyenne globale).
5. **À vérifier** : la couverture du vocabulaire d'équipements (libellés locaux) et la définition de la tranche de prix (terciles de la ville).

```python
"""Phase 13 : apport métier (calculs lourds en cache : models/_biz_cache).
 - simulation « et si j'ajoute cet équipement ? » (association, pas causalité)
 - annonces sur / sous-évaluées (résidus hors échantillon de la CV), quartiers sous-évalués
 - régulation : concentration des hôtes, licences, annonces fantômes
 - ville inconnue : combien d'annonces faut-il pour recalibrer le niveau de prix ?
Usage : python -m src.business   (après src.regression et src.final_model)
"""
import pickle
import warnings

import joblib
import numpy as np
import pandas as pd

from src.data import ROOT
from src.regression import final as reg_final

warnings.filterwarnings("ignore")
SEED = 42
CACHE = ROOT / "models" / "_biz_cache"
CACHE.mkdir(parents=True, exist_ok=True)


def amenity_whatif(pm, tr, n=2500):
    """Pour chaque équipement du vocabulaire : variation moyenne du prix prédit quand on l'ajoute à des annonces qui ne l'ont pas."""
    rows = []
    rng = np.random.RandomState(SEED)
    base_all = tr.sample(min(20000, len(tr)), random_state=SEED).reset_index(drop=True)
    p0_all = pm.predict_frame(base_all).log_pred.values
    for a in pm.amenity_vocab:
        col = "am_" + a
        if col not in base_all.columns:
            continue
        m = base_all[col] == 0
        if m.sum() < 200:
            continue
        idx = np.where(m)[0]
        idx = rng.choice(idx, min(n, len(idx)), replace=False)
        X = base_all.iloc[idx].copy()
        X[col] = 1.0
        X["n_amenities"] = X["n_amenities"] + 1
        d = pm.predict_frame(X).log_pred.values - p0_all[idx]
        share = float((base_all[col] == 1).mean())
        rows.append(dict(équipement=a, part_des_annonces=share, effet_pct=float(np.expm1(d.mean()) * 100), ic95_bas=float(np.expm1(d.mean() - 1.96 * d.std() / np.sqrt(len(d))) * 100),
                         ic95_haut=float(np.expm1(d.mean() + 1.96 * d.std() / np.sqrt(len(d))) * 100), n=len(d)))
    return pd.DataFrame(rows).sort_values("effet_pct", ascending=False)


def gini(x):
    x = np.sort(np.asarray(x, float))
    n = len(x)
    return float((2 * np.arange(1, n + 1) - n - 1).dot(x) / (n * x.sum()))


def build():
    f = CACHE / "biz.pkl"
    if f.exists():
        return pickle.load(open(f, "rb"))
    pm = joblib.load(ROOT / "models" / "final_model.joblib")
    d = pd.read_parquet(ROOT / "data/processed/listings_features.parquet")
    tr = d[~d.is_test & d.in_reg_set].reset_index(drop=True)
    fin = reg_final()
    out = {}
    out["whatif"] = amenity_whatif(pm, tr)

    # ---- sur / sous-évaluation (résidus hors échantillon, schéma A)
    idx = fin["d_index"].copy()
    idx["res"] = fin["y"] - fin["oof"]  # > 0 : prix observé au-dessus du prix prédit (sur-évalué), < 0 : en dessous (sous-évalué)
    idx["ratio"] = np.exp(idx["res"])
    out["res_by_city"] = idx.groupby("city").agg(n=("res", "size"), pct_sur=("res", lambda s: (s > 0.5).mean() * 100), pct_sous=("res", lambda s: (s < -0.5).mean() * 100),
                                                 rmse=("res", lambda s: np.sqrt((s ** 2).mean())), biais=("res", "mean")).round(2)
    nb = idx.groupby(["city", "neighbourhood_cleansed"]).agg(n=("res", "size"), res_moy=("res", "mean"), res_sd=("res", "std"), prix_median=("price_eur", "median")).reset_index()
    nb = nb[nb.n >= 40].copy()
    nb["se"] = nb.res_sd / np.sqrt(nb.n)
    nb["ecart_pct"] = np.expm1(nb.res_moy) * 100
    nb["ic_bas"], nb["ic_haut"] = np.expm1(nb.res_moy - 1.96 * nb.se) * 100, np.expm1(nb.res_moy + 1.96 * nb.se) * 100
    out["neighbourhoods"] = nb.sort_values("ecart_pct")
    # annonces extrêmes + croisement avec les annonces fantômes
    ext = d[~d.is_test & d.in_reg_set][["city", "id", "availability_365", "number_of_reviews_ltm", "host_is_pro"]].reset_index(drop=True)
    ext["res"] = idx["res"].values
    ext["ghost"] = (ext.availability_365 == 0) & (ext.number_of_reviews_ltm == 0)
    out["ext_vs_ghost"] = ext.assign(grp=np.where(ext.res > 0.5, "sur-évaluées (>+65 %)", np.where(ext.res < -0.5, "sous-évaluées (<-39 %)", "dans la norme"))).groupby("grp").agg(
        n=("res", "size"), part_fantomes=("ghost", "mean"), part_pro=("host_is_pro", "mean"), dispo_moy=("availability_365", "mean")).round(3)

    # ---- régulation : concentration, licences, fantômes (toutes annonces)
    reg = []
    for c, g in d.groupby("city"):
        n_host = g.groupby("host_id").size().sort_values(ascending=False)
        top10 = n_host.head(max(1, int(len(n_host) * 0.1))).sum() / n_host.sum() * 100
        reg.append(dict(city=c, annonces=len(g), hôtes=len(n_host), part_hôtes_5plus_pct=(g.calculated_host_listings_count >= 5).mean() * 100, top10pct_hôtes_part_annonces=top10,
                        gini_annonces_par_hôte=gini(n_host.values), logements_entiers_pct=(g.room_type == "Entire home/apt").mean() * 100, licence_pct=g.has_license.mean() * 100,
                        fantômes_pct=((g.availability_365 == 0) & (g.number_of_reviews_ltm == 0)).mean() * 100, prix_median_eur=g.price_eur.median()))
    out["regulation"] = pd.DataFrame(reg).set_index("city").round(1)

    # ---- ville inconnue : recalibrer le niveau avec n annonces
    rng = np.random.RandomState(SEED)
    q_dec = joblib.load(ROOT / "models" / "final_model.joblib")
    rows = []
    cov = []
    for v in ["V1", "V3"]:
        p = fin["oofB"][v]
        for c in sorted(fin["d_index"].city.unique()):
            m = (fin["d_index"].city == c).values
            r = fin["y"][m] - p[m]
            for n in [0, 10, 25, 50, 100, 200, 500, 1000]:
                vals = []
                for _ in range(40):
                    shift = r[rng.choice(len(r), n, replace=False)].mean() if n > 0 else 0.0
                    vals.append(np.sqrt(np.mean((r - shift) ** 2)))
                rows.append(dict(variante=v, city=c, n_calib=n, rmse=float(np.mean(vals))))
            # couverture de l'intervalle conforme (calibré sur les villes connues) avant et après recalibration (n = 200)
            lp = p[m]
            edges, q = q_dec.bin_edges, q_dec.bin_q
            b = np.clip(np.searchsorted(edges, lp, side="right") - 1, 0, len(q) - 1)
            for lab, shift in [("sans recalibration", 0.0), ("recalibré (200 annonces)", float(r[rng.choice(len(r), 200, replace=False)].mean()))]:
                yv = fin["y"][m]
                cov.append(dict(variante=v, city=c, cas=lab, couverture=float(np.mean(np.abs(yv - (lp + shift)) <= q[b]))))
    out["recal_curve"] = pd.DataFrame(rows)
    out["recal_cov"] = pd.DataFrame(cov)
    pickle.dump(out, open(f, "wb"))
    return out


if __name__ == "__main__":
    o = build()
    print(o["whatif"].head(10).round(2).to_string())
    print(o["whatif"].tail(5).round(2).to_string())
    print(o["res_by_city"].to_string())
    print(o["regulation"].to_string())
    print(o["recal_cov"].groupby(["variante", "cas"]).couverture.mean().round(3))
```

## 23. Limites, éthique et recommandations

Un modèle qui prédit bien en moyenne peut être injuste ou trompeur pour certains cas, et un outil de prix n'est jamais neutre. On dresse ici le bilan honnête : limites des données, limites du modèle, risques d'usage, puis les actions concrètes que les résultats justifient.

### 23.1 Le modèle est-il aussi bon pour tout le monde ?

````python
g2 = tpb.assign(rmse2=(np.log(tpb.price_eur) - np.log(tpb.pred_eur)) ** 2)
by_city = g2.groupby("city").agg(RMSE=("rmse2", lambda s: np.sqrt(s.mean())), MAPE_médian=("rel", "median"), couverture=("inside", "mean"))
by_type = g2.groupby("room_type").agg(n=("rmse2", "size"), RMSE=("rmse2", lambda s: np.sqrt(s.mean())), MAPE_médian=("rel", "median"), couverture=("inside", "mean")).query("n > 200")
by_dec = g2.groupby(pd.qcut(g2.price_eur, 10, labels=False)).agg(RMSE=("rmse2", lambda s: np.sqrt(s.mean())), MAPE_médian=("rel", "median"), couverture=("inside", "mean"))
fig, ax = plt.subplots(1, 3, figsize=(20, 5.4))
b = by_city.sort_values("MAPE_médian"); ax[0].barh(b.index, b.MAPE_médian * 100, color=PAL["jaune"], edgecolor="white", linewidth=0.8); ax[0].set(title="Erreur relative médiane par ville (test)", xlabel="% du prix")
ax[1].bar(by_type.index, by_type.MAPE_médian * 100, color=PAL["rose"], edgecolor="white", linewidth=0.8); ax[1].tick_params(axis="x", labelsize=8); ax[1].set(title="Erreur relative médiane par type de logement", ylabel="% du prix")
ax[2].plot(range(1, 11), by_dec.couverture * 100, "o-", color=PAL["bleu"], mec="white", label="couverture de la fourchette"); ax[2].axhline(90, color=PAL["noir"], ls="--"); ax2 = ax[2].twinx(); ax2.bar(range(1, 11), by_dec.MAPE_médian * 100, color=PAL["gris"], alpha=0.4, edgecolor="white", linewidth=0.8); ax2.set_ylabel("erreur relative médiane (%)")
ax[2].set(title="Par décile de prix RÉEL (test)", xlabel="décile de prix réel (1 = moins cher)", ylabel="couverture (%)", ylim=(40, 100)); ax[2].legend(loc="lower left")
plt.tight_layout(); plt.show()
display(by_type.round(3))
````

<div style="font-family:'Segoe UI',Calibri,Helvetica,Arial,sans-serif;border-left:4px solid #9A7DC0;background:#9A7DC014;padding:10px 16px;margin:10px 0 20px;color:#3E4A55;font-size:14.5px;line-height:1.6">
<span style="font-size:11px;letter-spacing:.12em;text-transform:uppercase;font-weight:700;color:#9A7DC0">Ce qu'on en retient</span><br><b>Non, le modèle n'est pas aussi bon pour tout le monde.</b> (1) <b>Par prix réel</b>, la couverture de la fourchette est de l'ordre de 93-95 % au milieu de la distribution mais tombe à <b>74 % pour le décile le plus cher</b> (et à 85 % pour le moins cher) : le modèle est <b>le moins fiable pour les biens de luxe</b>, justement ceux dont l'enjeu financier est le plus fort. (2) <b>Par ville</b>, l'erreur relative est comparable, mais la couverture de la fourchette varie de 85 % (Mexico) à 93 % (Toronto). (3) <b>Par type de logement</b>, la couverture est de 91,5 % pour les logements entiers, 88 % pour les chambres privées, mais seulement <b>80 % pour les chambres partagées et 75 % pour les chambres d'hôtel</b> (types rares, moins présents à l'entraînement). Un outil déployé doit donc <b>afficher sa fourchette et dire quand il est hors de son domaine de validité</b>.</div>

### 23.2 Limites et risques

**Limites des données**
- **Le prix est un devis pour un séjour précis** (1 à 31 nuits, check-in de 0 à 369 jours après la collecte), **pas un prix payé**. On n'a aucune donnée de **réservation réelle**.
- **Biais de sélection** : le prix manque pour **25 % des annonces**, et ces annonces sont surtout indisponibles ou non rafraîchies. Le modèle ne connaît que des annonces **réservables**.
- **Une seule collecte par ville** (septembre 2026) : on ne peut ni mesurer la dérive dans le temps, ni isoler la saisonnalité annuelle (effet d'horizon mélangé). Le **calendar ne contient pas de prix**.
- **Variables absentes** qui font le prix d'un logement de luxe : photos, vue, surface, rénovation, avis détaillés. Cela explique la sous-estimation des prix très élevés.
- **Échantillons** : sélection de variables sur 60 000 annonces ; NLP sur ~8 000 annonces bien documentées (≥ 8 avis) ; course des modèles sur 80 000 annonces. Les résultats ne valent pas tels quels pour des annonces sans avis.

**Limites du modèle**
- **Régression vers la moyenne** : surestimation des logements bon marché (−21 % au premier décile), sous-estimation des plus chers (+47 % au dernier).
- **Ville inconnue** : sans recalibration, l'erreur passe de 0,36 à 0,56 (RMSE sur le log-prix) et la fourchette ne couvre que 65 % des prix.
- **Associations, pas causes** : le simulateur « et si j'ajoute… » décrit ce que le modèle a appris, pas ce qui arriverait réellement.
- Les **intervalles conformes** garantissent leur couverture sur des hôtes **semblables** à ceux de la calibration, pas sur n'importe quel cas.

**Risques éthiques et réglementaires**
- **Données personnelles (RGPD)** : le nom des hôtes a été **retiré** ; le numéro de licence est réduit à un indicateur ; les identifiants d'hôte ne servent qu'à **grouper** les validations ; les avis sont agrégés par annonce (les noms des auteurs ne sont pas utilisés). Un déploiement réel exigerait une base légale et une analyse d'impact.
- **Effets sur le logement local** : recommander des prix aux hôtes peut **encourager la conversion de logements en meublés touristiques**. Les chiffres de concentration (jusqu'à **74 %** des annonces détenues par des hôtes à ≥ 5 annonces à Tokyo, 59 % pour les 10 % d'hôtes les plus gros à Barcelone) montrent que le marché est déjà très professionnalisé.
- **Homogénéisation des prix et coordination algorithmique** : si tous les hôtes suivent le même outil, les prix convergent et la concurrence s'atténue. Le modèle doit rester une **aide à la décision avec une fourchette**, jamais un prix imposé.
- **Évolutions réglementaires** : les licences, plafonds de nuitées et interdictions varient selon la ville et **changent vite** ; la colonne de licence n'est même pas renseignée à Londres et à Mexico.
- **Biais de langue du NLP** : 62 % des avis de l'échantillon sont en anglais ; les langues minoritaires sont mal couvertes ; les avis sont majoritairement très positifs (médiane 4,84), ce qui rend la détection des problèmes plus difficile.

### 23.3 Cinq actions concrètes, chacune reliée à un résultat chiffré

### Synthèse des choix méthodologiques
**Les décisions à justifier**
1. **Pourquoi prédire `log(prix)`** : distribution écrasée (asymétrie 57 → 0,6), erreurs relatives plutôt qu'absolues.
2. **Pourquoi un découpage par hôte** : 253 996 hôtes pour 485 128 annonces ; un split par annonce ferait fuiter l'hôte et gonflerait les scores.
3. **Pourquoi le prix est un devis** : le contexte du devis (durée, délai, saison) est une variable majeure (RMSE 0,396 → 0,374) ; sans lui on compare des prix qui ne mesurent pas la même chose.
4. **Pourquoi exclure disponibilités et avis récents** : conséquences du prix (fuite indirecte) ; le jeu « full » donne un meilleur RMSE (0,354) mais inutilisable par un hôte qui fixe son prix.
5. **Pourquoi des taux BCE à la date de collecte** : le symbole « $ » du brut est trompeur ; les taux sont sourcés et reproductibles (`data/external`).
6. **Pourquoi le target encoding avec validation croisée interne** : 1 294 couples ville-quartier ; l'encodage naïf triche (l'annonce se voit elle-même).
7. **Pourquoi le boosting** : meilleur en validation (0,375 contre 0,428 pour le linéaire), les interactions comptent ; le réseau de neurones ne le bat pas (chapitre 18).
8. **Pourquoi la MAE en euros sans correction de biais** : la correction (smearing) corrige la moyenne mais dégrade la MAE (70 → 74 €) ; on recommande la médiane.
9. **Pourquoi des intervalles conformes adaptatifs** : la régression quantile ne couvre que 82 % ; le conformal global sous-couvre les logements chers (83,5 %) ; Mondrian donne 90 % dans chaque décile.
10. **Pourquoi deux schémas de validation** : l'effet ville est massif (RMSE 0,366 contre 0,583) ; latitude et longitude sont des identifiants de ville déguisés.
11. **Pourquoi deux usages pour le Superhost** : avec les notes, on lit en partie les critères d'attribution (AUC 0,875) ; sans, on mesure ce que le logement et l'hôte annoncent (0,817).
12. **Pourquoi aucune ligne n'est supprimée** : les cas douteux sont signalés, les modèles filtrent, et on peut mesurer ce que le filtrage change.

**Les limites à dire spontanément** : prix affiché et non payé ; biais de sélection (25 % sans prix) ; une seule collecte par ville ; calendar sans prix ; sous-estimation du luxe ; recalibration nécessaire pour une nouvelle ville ; associations et non causalité.

```python
"""Exécute le projet de bout en bout (reproductible, random_state = 42, configuration dans config.yaml).

Usage :
    python run_all.py              # tout (plusieurs heures sur un portable à 16 cœurs ; les étapes lourdes sont mises en cache)
    python run_all.py --from 5     # reprendre à l'étape 5
    python run_all.py --only 3     # une seule étape
Les téléchargements initiaux (download_airbnb.py, merge_airbnb.py) sont supposés déjà faits : data/processed/listings_all.parquet et reviews_all.parquet.
"""
import subprocess
import sys
import time

STEPS = [
    ("0  Audit des données", ["-m", "src.audit"]),
    ("1  Analyse exploratoire", ["-m", "src.eda"]),
    ("2  Nettoyage + split train/test (test scellé)", ["-m", "src.cleaning"]),
    ("2b Impact de la règle d'outliers", ["-m", "src.outlier_impact"]),
    ("3  Variables + contrôles anti-fuite", ["-m", "src.features"]),
    ("3b Contrôles des features", ["-m", "src.features_check"]),
    ("4  Sélection de variables", ["-m", "src.selection"]),
    ("4b Rapport de sélection", ["-m", "src.selection_report"]),
    ("5  Schémas de validation A/B/C", ["-m", "src.validation"]),
    ("5b Rapport de validation", ["-m", "src.validation_report"]),
    ("6  Régression (course, réglage Optuna, OOF, LOCO)", ["-m", "src.regression", "all"]),
    ("7  Classification (Superhost, tranche de prix)", ["-m", "src.classification"]),
    ("8  Clustering, PCA, UMAP, anomalies", ["-m", "src.clustering"]),
    ("9  NLP : échantillon, sentiment, embeddings", ["-m", "src.nlp", "all"]),
    ("9b NLP : analyses et valeur ajoutée", ["-m", "src.nlp_analysis"]),
    ("10 Séries temporelles (calendar)", ["-m", "src.timeseries"]),
    ("11 Deep learning", ["-m", "src.deep"]),
    ("14 Modèle final + ouverture UNIQUE du test", ["-m", "src.final_model"]),
    ("12 Interprétabilité (SHAP, PDP)", ["-m", "src.interpret"]),
    ("13 Apport métier", ["-m", "src.business"]),
    ("   Notebook final", ["tools/build_notebook.py"]),
]

if __name__ == "__main__":
    a = sys.argv[1:]
    lo = int(a[a.index("--from") + 1]) if "--from" in a else 1
    only = int(a[a.index("--only") + 1]) if "--only" in a else None
    for i, (name, cmd) in enumerate(STEPS, 1):
        if (only and i != only) or i < lo:
            continue
        t0 = time.time()
        print(f"\n=== [{i}/{len(STEPS)}] {name} ===", flush=True)
        r = subprocess.run([sys.executable] + cmd)
        print(f"--- {name} : {'OK' if r.returncode == 0 else 'ÉCHEC'} en {time.time() - t0:.0f}s", flush=True)
        if r.returncode != 0:
            sys.exit(r.returncode)
